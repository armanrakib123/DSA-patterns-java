# Math & Geometry (গণিত এবং জ্যামিতি) - The FAANG Standard Theory 📐🔢

## ১. ভূমিকা (Detailed Introduction)
কম্পিউটার সায়েন্সের মূলে রয়েছে গণিত। অনেক জটিল অ্যালগরিদম আসলে গণিতের একটি রূপান্তর মাত্র। FAANG ইন্টারভিউতে ম্যাথ এবং জ্যামিতি সংক্রান্ত সমস্যাগুলো মূলত আপনার লজিক্যাল রিজনিং এবং গাণিতিক দক্ষতা যাচাই করার জন্য দেওয়া হয়। এটি আপনাকে জটিল সমস্যাগুলোকে গাণিতিক সূত্রে (Mathematical Formula) রূপান্তর করতে সাহায্য করে।

সহজ কথায়, ম্যাথ হলো কোডিংয়ের "পাওয়ার সোর্স"। আপনি যদি বুঝতে পারেন একটি প্রবলেমের পেছনে কোন গাণিতিক সূত্র কাজ করছে, তবে আপনার কোড অনেক ছোট এবং ফাস্ট হবে।

---

## ২. কেন গণিত এবং জ্যামিতি? (The Foundation of Complexity)
অনেক সমস্যা আছে যা লুপ চালিয়ে সমাধান করা সম্ভব, কিন্তু গাণিতিক সূত্র ব্যবহার করলে তা $O(1)$ এ করা যায়।
- **ব্যবহার:** Primality Test, GCD/LCM, Modular Arithmetic, Coordinate Geometry, Line Intersection.

## ৩. গুরুত্বপূর্ণ গাণিতিক কনসেপ্ট (Core Concepts)
১. **Sieve of Eratosthenes:** প্রাইম নাম্বার (Prime Numbers) দ্রুত খুঁজে বের করার উপায়।
2. **Euclidean Algorithm:** GCD বের করার সবচেয়ে ফাস্ট পদ্ধতি।
৩. **Modular Arithmetic:** বড় সংখ্যা নিয়ে কাজ করার সময় ওভারফ্লো (Overflow) এড়ানোর কৌশল।
৪. **Geometry Basics:** দূরত্ব বের করা, ভেক্টর ক্রস প্রোডাক্ট, এবং এরিয়া ক্যালকুলেশন।

## ৪. রিয়েল-ওয়ার্ল্ড FAANG উদাহরণ (Real-World Case Studies)

### ৪.১. সাইবার সিকিউরিটি - RSA এনক্রিপশন
গুগল বা আমাজনের সিকিউরিটি সিস্টেমে যখন ডেটা এনক্রিপ্ট করা হয়, তখন বিশাল বিশাল প্রাইম নাম্বার এবং মডুলার অ্যারিথমেটিক ব্যবহার করা হয়। এটি নিশ্চিত করে যে আপনার ডেটা হ্যাকারদের কাছ থেকে নিরাপদ থাকবে। এটি সম্পূর্ণভাবে নাম্বার থিওরির ওপর নির্ভরশীল।

### ৪.২. গুগল ম্যাপস - কোঅর্ডিনেট জিওমেট্রি
গুগল ম্যাপস যখন দুটি জায়গার মধ্যে দূরত্ব বের করে, তখন তারা 'Haversine Formula' ব্যবহার করে যা গোলাকার পৃথিবীর ওপর দুটি পয়েন্টের দূরত্ব নির্ণয় করে। এটি জ্যামিতির একটি অ্যাডভান্সড অ্যাপ্লিকেশন।

### ৪.৩. কম্পিউটার গ্রাফিক্স ও গেমিং - ভেক্টর ক্যালকুলেশন
পিক্সার (Pixar) বা এনভিডিয়া (Nvidia) যখন থ্রিডি গ্রাফিক্স রেন্ডার করে, তারা কোটি কোটি জ্যামিতিক ক্যালকুলেশন করে। প্রতিটি ফ্রেম জেনারেট করার জন্য ভেক্টর মাল্টিপ্লিকেশন এবং ম্যাট্রিক্স ট্রান্সফর্মেশন ব্যবহার করা হয়।

# Math & Geometry (গণিত ও জ্যামিতি) 🧮📐

## Math in DSA
কম্পিটিটিভ প্রোগ্রামিং এবং ইন্টারভিউতে সরাসরি ম্যাথ (Math) বা জ্যামিতির (Geometry) কিছু প্রবলেম আসে, যেগুলো জানা না থাকলে সলভ করা প্রায় অসম্ভব।

## 🔢 Common Math Topics

১. **Prime Numbers (মৌলিক সংখ্যা):**
   - একটি সংখ্যা Prime কি না তা চেক করতে $O(\sqrt{N})$ সময় লাগে। 
   - অনেকগুলো প্রাইম নাম্বার একসাথে বের করতে **Sieve of Eratosthenes** (উচ্চারণ: সিভ অফ এরাতোসথেনিস) অ্যালগরিদম ব্যবহার করা হয়, যা প্রায় $O(N \log(\log N))$ এ কাজ করে।

২. **GCD ও LCM (গসাগু ও লসাগু):**
   - GCD (Greatest Common Divisor) বের করার জন্য **Euclidean Algorithm** (ইউক্লিডিয়ান অ্যালগরিদম) সবচেয়ে ফাস্ট। 
   - `LCM(a, b) = (a * b) / GCD(a, b)`

৩. **Fast Exponentiation (দ্রুত পাওয়ার বের করা):**
   - $x^n$ বের করতে $O(N)$ সময় লাগে। কিন্তু **Binary Exponentiation** ব্যবহার করে এটি $O(\log N)$ সময়ে বের করা যায়।

৪. **Modulo Arithmetic:**
   - যখন উত্তর অনেক বড় হয়ে যায়, তখন $10^9 + 7$ দিয়ে মডুলো (Modulo) করতে বলা হয়। 
   - `(a + b) % M = ((a % M) + (b % M)) % M`
   - `(a * b) % M = ((a % M) * (b % M)) % M`

## 📐 Common Geometry Topics

১. **Rectangle Overlap (আয়তক্ষেত্রের ওভারল্যাপ):**
   - দুটি আয়তক্ষেত্র ওভারল্যাপ করে কি না তা চেক করার লজিক (Intervals এর 2D ভার্সন)।
   - `Overlap Area = max(0, min(right1, right2) - max(left1, left2)) * max(0, min(top1, top2) - max(bottom1, bottom2))`

২. **Points on a Line (একই সরলরেখায় বিন্দু):**
   - তিনটি বিন্দু (Points) একই সরলরেখায় (Collinear) আছে কি না তা তাদের ঢাল (Slope) চেক করে বের করা যায়।
   - Slope $m = (y_2 - y_1) / (x_2 - x_1)$। 
   - ডিভাইড বাই জিরো (Divide by Zero) এরর এড়াতে গুণ (Cross Multiplication) ব্যবহার করা ভালো: $(y_2 - y_1) \times (x_3 - x_2) == (y_3 - y_2) \times (x_2 - x_1)$।

পরবর্তী ফাইলগুলোতে আমরা এই ট্রিকসগুলোর ইমপ্লিমেন্টেশন দেখব।
