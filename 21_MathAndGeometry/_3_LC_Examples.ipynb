৩. এটি টু-পয়েন্টার (Fast & Slow) দিয়েও করা সম্ভব।

## ২. Pow(x, n) (LC 50) 🟡
**টাইপ:** Binary Exponentiation

### প্রবলেম স্টেটমেন্ট:
$x$ এর পাওয়ার $n$ বের করুন ($x^n$)।

### সমাধান:
১. ব্রুট ফোর্সে $O(N)$ সময় লাগে। কিন্তু বাইনারি এক্সপোনেনশিয়েশন ব্যবহার করে $O(\log N)$ এ করা সম্ভব।
২. যদি $n$ ইভেন (Even) হয়, তবে $x^n = (x^{n/2})^2$।
৩. যদি $n$ অড (Odd) হয়, তবে $x^n = x \times x^{n-1}$।
৪. নেগেটিভ পাওয়ার হ্যান্ডেল করার জন্য $1/x$ ব্যবহার করুন।

## ৩. Count Primes (LC 204) 🟡
**টাইপ:** Sieve of Eratosthenes

### প্রবলেম স্টেটমেন্ট:
$N$ এর চেয়ে ছোট কতগুলো প্রাইম নাম্বার আছে?

### সমাধান:
১. 'Sieve of Eratosthenes' অ্যালগরিদম ব্যবহার করুন।
২. একটি বুলিয়ান অ্যারে নিন এবং ২ থেকে শুরু করে প্রতিটি প্রাইম নাম্বারের গুণিতকগুলোকে (Multiples) 'Not Prime' মার্ক করুন।
৩. শেষে কতগুলো মার্ক করা হয়নি তা কাউন্ট করুন।

## ৪. Palindrome Number (LC 9) 🟢
**টাইপ:** Number Reversal

### প্রবলেম স্টেটমেন্ট:
একটি সংখ্যা প্যালেড্রোম কি না তা চেক করুন। স্ট্রিং কনভার্ট করা যাবে না।

### সমাধান:
১. সংখ্যাটির অর্ধেক রিভার্স (Reverse) করুন।
২. রিভার্সড অর্ধেক যদি বাকি অর্ধেকের সমান হয়, তবে এটি প্যালেড্রোম।
৩. নেগেটিভ সংখ্যা প্যালেড্রোম হতে পারে না।
