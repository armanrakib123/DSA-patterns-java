অনেকে মনে করেন কোডিং মানে শুধু লজিক, কিন্তু Software Company লেভেলের ইন্টারভিউতে ম্যাথ অ্যালগরিদমগুলো খুব গুরুত্বপূর্ণ। অনেক সময় একটি জটিল প্রবলেম শুধুমাত্র একটি ম্যাথমেটিক্যাল ফর্মুলা দিয়ে $O(1)$-এ সলভ করা যায়।

## Key Areas to Focus
১. **Prime Numbers (মৌলিক সংখ্যা):** সংখ্যাটি প্রাইম কি না এবং একটি রেঞ্জের মধ্যে কতগুলো প্রাইম আছে।
২. **GCD & LCM:** দুটি সংখ্যার গসাগু এবং লসাগু বের করা।
৩. **Modular Arithmetic:** বড় বড় ক্যালকুলেশন হ্যান্ডেল করার জন্য `% MOD` ব্যবহার করা।
৪. **Power & Root:** $x^n$ বা $\sqrt{x}$ বের করার অপ্টিমাইজড পদ্ধতি।
৫. **Base Conversion:** ডেসিমাল থেকে বাইনারি বা অন্য বেস-এ কনভার্ট করা।

## Prime Numbers: Sieve of Eratosthenes
যদি আপনাকে ১ থেকে ১০০-এর মধ্যে সব প্রাইম বের করতে বলা হয়, তবে প্রতিটি সংখ্যা আলাদা করে চেক করা অনেক সময়সাপেক্ষ ($O(N \sqrt{N})$)। এর বদলে **Sieve** ব্যবহার করলে আমরা মাত্র $O(N \log \log N)$ টাইমে সব প্রাইম পেয়ে যাই।


# GCD & Modular Arithmetic

## 1. GCD (Greatest Common Divisor)
গসাগু বের করার সবচেয়ে সেরা পদ্ধতি হলো **Euclidean Algorithm**। 
লজিক: $GCD(a, b) = GCD(b, a \% b)$। এটি ততক্ষণ চলে যতক্ষণ না রিমেইন্ডার ০ হয়।

```java
int gcd(int a, int b) {
    return b == 0 ? a : gcd(b, a % b);
}
```

## 2. Modular Arithmetic Properties
ইন্টারভিউতে বড় সংখ্যার গুণ বা যোগ করার সময় রেইঞ্জ ওভারফ্লো হতে পারে। তাই এই প্রপার্টিগুলো মনে রাখা জরুরি:
- $(a + b) \% M = ((a \% M) + (b \% M)) \% M$
- $(a \times b) \% M = ((a \% M) \times (b \% M)) \% M$
- $(a - b) \% M = ((a \% M) - (b \% M) + M) \% M$ (বিয়োগের ক্ষেত্রে $M$ যোগ করা হয় নেগেটিভ ভ্যালু এড়াতে)

## 3. Fast Exponentiation (Binary Exponentiation)
$2^{100}$ বের করার জন্য ১০০ বার গুণ না করে আমরা একে $2^{50} \times 2^{50}$ এভাবে ভাগ করতে পারি। এটি টাইম কমপ্লেক্সিটি $O(N)$ থেকে $O(\log N)$-এ নামিয়ে আনে।
