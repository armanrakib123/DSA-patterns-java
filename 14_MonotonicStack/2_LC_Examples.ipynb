## ১. Next Greater Element I (LC 496) 🟢
**টাইপ:** Monotonic Decreasing Stack

### প্রবলেম স্টেটমেন্ট:
একটি অ্যারের প্রতিটি ইলিমেন্টের জন্য তার ডান পাশের প্রথম বড় সংখ্যাটি খুঁজে বের করুন।

### সমাধান:
১. একটি স্ট্যাক ব্যবহার করুন। যদি বর্তমান সংখ্যা স্ট্যাকের উপরের সংখ্যার চেয়ে বড় হয়, তবে বর্তমান সংখ্যাটিই হলো স্ট্যাকের ওই সংখ্যার জন্য "Next Greater Element"।
২. স্ট্যাক থেকে পপ করুন এবং রেজাল্ট ম্যাপে সেভ করুন।
৩. বর্তমান সংখ্যাটি স্ট্যাকে পুশ করুন।

## ২. Daily Temperatures (LC 739) 🟡
**টাইপ:** Index-based Monotonic Stack

### প্রবলেম স্টেটমেন্ট:
প্রতিদিনের টেম্পারেচার দেওয়া আছে। কতদিন পর এর চেয়ে বেশি টেম্পারেচার পাওয়া যাবে তা বের করুন।

### সমাধান:
১. স্ট্যাকে টেম্পারেচারের বদলে ইনডেক্স (Index) স্টোর করুন।
২. যখন বর্তমান টেম্পারেচার স্ট্যাকের টেম্পারেচারের চেয়ে বেশি হবে, তখন ইনডেক্সের পার্থক্যই (`current_index - stack_index`) হবে আপনার উত্তর।

## ৩. Largest Rectangle in Histogram (LC 84) 🔴
**টাইপ:** Advanced Monotonic Stack

### প্রবলেম স্টেটমেন্ট:
একটি হিস্টোগ্রামে সবচেয়ে বড় রেকট্যাঙ্গেলের এরিয়া কত হবে?

### সমাধান:
১. প্রতিটি বারের (Bar) জন্য তার বামে এবং ডানে প্রথম ছোট বারটি খুঁজে বের করুন (Next Smaller Element)।
২. এই দুই সীমানার মাঝখানের ডিস্ট্যান্সই হবে রেকট্যাঙ্গেলের উইডথ (Width)।
৩. এরিয়া = `height * width`।

## ৪. Trapping Rain Water (LC 42) 🔴
**টাইপ:** Two Pointer vs Monotonic Stack

### প্রবলেম স্টেটমেন্ট:
একটি অ্যারেতে হাইট দেওয়া আছে। বৃষ্টির পর কতটুকু পানি জমা হবে?

### সমাধান:
১. একটি ডিক্রিজিং স্ট্যাক ব্যবহার করুন।
২. যখন একটি বড় হাইট পাওয়া যাবে, তখন স্ট্যাকের টপ ইলিমেন্টটি হবে পানির "Bottom"।
৩. বাম ও ডানের মিনিমাম হাইট থেকে বটম বিয়োগ করে পানির গভীরতা বের করুন।
