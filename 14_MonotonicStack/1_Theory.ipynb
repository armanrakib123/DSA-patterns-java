# Monotonic Stack (মনোটোনিক স্ট্যাক) - The FAANG Standard Theory 📊

## ১. ভূমিকা (Detailed Introduction)
মনোটোনিক স্ট্যাক (Monotonic Stack) হলো স্ট্যাক ডেটা স্ট্রাকচারের একটি বিশেষ ব্যবহার। সাধারণ স্ট্যাকের মতো এটি শুধু ডেটা স্টোর করে না, বরং এটি নিশ্চিত করে যে স্ট্যাকের ভেতরে ডেটাগুলো সব সময় একটি নির্দিষ্ট অর্ডারে (ক্রম) থাকবে—হয় ছোট থেকে বড় (Increasing) অথবা বড় থেকে ছোট (Decreasing)। FAANG ইন্টারভিউতে "Next Greater Element" বা "Histogram" সংক্রান্ত সমস্যার এটিই একমাত্র অপ্টিমাইজড সমাধান।

সহজ কথায়, মনোটোনিক স্ট্যাক হলো একটি "স্মার্ট লাইন"। যদি লাইনে কেউ নতুন আসে এবং সে যদি আগের জনের চেয়ে শক্তিশালী হয়, তবে সে আগের জনকে সরিয়ে নিজের জায়গা করে নেয়। এই লজিকটিই আমরা কোডিংয়ে ব্যবহার করি।

## ২. কেন মনোটোনিক স্ট্যাক? (The Power of O(N))
অনেক সমস্যা আছে যা নেস্টেড লুপ দিয়ে $O(N^2)$ এ সমাধান করা যায়। কিন্তু মনোটোনিক স্ট্যাক ব্যবহার করে আমরা প্রতিটি ইলিমেন্টকে স্ট্যাকে মাত্র একবার পুশ এবং একবার পপ করে $O(N)$ টাইমে সমাধান করতে পারি।
- **ব্যবহার:** Next Greater Element, Next Smaller Element, Largest Rectangle in Histogram, Trapping Rain Water.

## ৩. মনোটোনিক স্ট্যাকের প্রকারভেদ
ইন্টারভিউতে মূলত দুই ধরনের মনোটোনিক স্ট্যাক ব্যবহৃত হয়:

১. **Monotonic Increasing Stack:** এখানে স্ট্যাকের নিচে সবচেয়ে ছোট এবং উপরে সবচেয়ে বড় ইলিমেন্ট থাকে। এটি "Next Smaller Element" খুঁজে বের করতে সাহায্য করে।
২. **Monotonic Decreasing Stack:** এখানে স্ট্যাকের নিচে সবচেয়ে বড় এবং উপরে সবচেয়ে ছোট ইলিমেন্ট থাকে। এটি "Next Greater Element" খুঁজে বের করতে সাহায্য করে।

### ৪.২. আমাজন - শিপিং কন্টেইনার স্ট্যাকিং
আমাজনের ওয়্যারহাউসে যখন কন্টেইনারগুলো একটির ওপর একটি রাখা হয়, তখন তাদের হাইট এবং ওয়েট ম্যানেজ করার জন্য মনোটোনিক স্ট্যাক লজিক ব্যবহার করা হয়। এটি নিশ্চিত করে যে ভারী কন্টেইনারগুলো নিচে থাকে এবং স্ট্যাকটি স্টেবল থাকে।

### ৪.৩. ফিন্যান্সিয়াল অ্যানালিটিক্স - মার্কেট ব্রেকআউট ডিটেকশন

# Monotonic Stack (মনোটোনিক স্ট্যাক) 🥞

## Monotonic Stack কী?
"Monotonic" শব্দের অর্থ হলো একমুখী (যা শুধু বাড়ে বা শুধু কমে)। 
**Monotonic Stack** হলো একটি সাধারণ Stack ডেটা স্ট্রাকচার, যার ভেতরের ইলিমেন্টগুলো সবসময় একটি নির্দিষ্ট অর্ডারে (ছোট থেকে বড় অথবা বড় থেকে ছোট) সাজানো থাকে।

এটি Array এর প্রবলেমগুলোকে (যেখানে $O(N^2)$ সময় লাগে) খুব সহজে $O(N)$ এ সলভ করতে সাহায্য করে।

## ২ প্রকারের Monotonic Stack
১. **Monotonic Increasing Stack:**
   - স্ট্যাকের নিচের ইলিমেন্ট সবচেয়ে ছোট এবং ওপরের ইলিমেন্ট সবচেয়ে বড় থাকে।
   - **নিয়ম:** যদি নতুন ইলিমেন্ট স্ট্যাকের ওপরের (Top) ইলিমেন্টের চেয়ে ছোট হয়, তবে স্ট্যাক থেকে ওপরের ইলিমেন্টগুলো পপ (Pop) করতে হবে, যতক্ষণ না নতুন ইলিমেন্টটি বসার উপযুক্ত হয় (অর্থাৎ স্ট্যাক আবার increasing হয়)।
   - **ব্যবহার:** "Next Smaller Element" (পরবর্তী ছোট সংখ্যা) বের করার জন্য।

২. **Monotonic Decreasing Stack:**
   - স্ট্যাকের নিচের ইলিমেন্ট সবচেয়ে বড় এবং ওপরের ইলিমেন্ট সবচেয়ে ছোট থাকে।
   - **নিয়ম:** যদি নতুন ইলিমেন্ট স্ট্যাকের ওপরের ইলিমেন্টের চেয়ে বড় হয়, তবে ছোট ইলিমেন্টগুলো পপ (Pop) করতে হবে।
   - **ব্যবহার:** "Next Greater Element" (পরবর্তী বড় সংখ্যা) বের করার জন্য।

## কখন বুঝবেন যে Monotonic Stack ব্যবহার করতে হবে? (Pattern Recognition)
ইন্টারভিউতে নিচের লক্ষণগুলো দেখলে বুঝবেন এখানে Monotonic Stack লাগবে:
১. **"Next Greater" বা "Next Smaller":** প্রবলেমে সরাসরি বা পরোক্ষভাবে বলা থাকবে যে ডানদিকের বা বামদিকের প্রথম বড়/ছোট সংখ্যা বের করতে হবে।
২. **Daily Temperatures / Stock Spans:** "পরবর্তী গরম দিনটি আসতে কত দিন লাগবে?" বা "স্টক প্রাইস এর আগে কতদিন পর্যন্ত কম ছিল?"
৩. **Histogram / Area:** "সবচেয়ে বড় আয়তক্ষেত্র (Rectangle) বের করো" (Largest Rectangle in Histogram)।
৪. **O(N) Requirement:** প্রবলেমটি Nested Loop দিয়ে $O(N^2)$ এ সলভ করা যাচ্ছে, কিন্তু আপনাকে $O(N)$ এ করতে বলা হয়েছে।
