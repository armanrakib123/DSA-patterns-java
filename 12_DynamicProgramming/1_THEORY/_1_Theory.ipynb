# 🧠 Pattern 45: Dynamic Programming (DP) - মহাজাগতিক বিশ্লেষণ

ডাইনামিক প্রোগ্রামিং (Dynamic Programming) বা সংক্ষেপে DP হলো অ্যালগরিদমের জগতে এক জাদুকরী পদ্ধতি। এটি মূলত কোনো বড় এবং জটিল সমস্যাকে ছোট ছোট উপ-সমস্যায় (Sub-problems) ভাগ করে সমাধান করে এবং সেই সমাধানগুলো একটি টেবিল বা মেমোরিতে জমিয়ে রাখে (Memoization) যাতে ভবিষ্যতে একই সমস্যা বারবার সমাধান করতে না হয়।

---

## 🧐 DP এর মূল দর্শন (Core Philosophy)

DP এর মূল মন্ত্র হলো: **"যারা অতীত মনে রাখতে পারে না, তারা তা পুনরাবৃত্তি করতে বাধ্য।"** 
(Those who cannot remember the past are condemned to repeat it.)

### ১. ওভারল্যাপিং সাব-প্রবলেম (Overlapping Sub-problems)
যদি একটি সমস্যা সমাধান করতে গিয়ে দেখেন যে একই ছোট সমস্যা বারবার ফিরে আসছে (যেমন- ফিবোনাচ্চি সিরিজে F(3) বারবার বের করতে হয়), তবে সেখানে DP ব্যবহার করা সম্ভব।

### ২. অপ্টিমাল সাব-স্ট্রাকচার (Optimal Sub-structure)
যদি কোনো বড় সমস্যার সমাধান তার ছোট ছোট উপ-সমস্যার অপ্টিমাল সমাধান থেকে পাওয়া যায়, তবে সেটি DP এর জন্য উপযুক্ত।

---

## 🚀 DP সমাধানের দুটি প্রধান অ্যাপ্রোচ

### ১. টপ-ডাউন (Top-Down / Memoization)
এটি মূলত রিকার্সন (Recursion) ভিত্তিক। আমরা বড় সমস্যা থেকে শুরু করি এবং ছোট সমস্যাগুলোর উত্তর একটি `Map` বা `Array` তে সেভ করে রাখি।
*ইন্টুইশন: "যদি উত্তরটা জানি তবে দিয়ে দাও, না জানলে বের করে মনে রাখো।"*

### ২. বটম-আপ (Bottom-Up / Tabulation)
এটি ইটারেটিভ (Iterative) বা লুপ ভিত্তিক। আমরা সবচেয়ে ছোট সমস্যা (Base case) থেকে শুরু করি এবং একটি টেবিল পূর্ণ করতে করতে বড় সমস্যার দিকে যাই।
*ইন্টুইশন: "ছোট থেকে বড়র দিকে যাও।"*

---

## 🛠️ কেন DP ইন্টারভিউতে সবচেয়ে ভীতিজাগানিয়া এবং গুরুত্বপূর্ণ?

- ১. **Logic Complexity:** DP প্রবলেমগুলো সরাসরি বোঝা কঠিন। এটি আপনার গভীর গাণিতিক এবং লজিক্যাল ক্ষমতা যাচাই করে।
- ২. **Efficiency:** এটি এক্সপোনেনশিয়াল টাইম কমপ্লেক্সিটিকে (O(2^N)) পলিনোমিয়াল টাইমে (O(N^2) বা O(N)) নামিয়ে আনে।
- ৩. **Decision Making:** এটি মূলত এমন সমস্যা সমাধানে ব্যবহৃত হয় যেখানে আপনাকে অনেকগুলো অপশনের মধ্যে সেরাটি (Min/Max/Count) বেছে নিতে হয়।

---

## 🚧 DP প্রবলেম সলভ করার "৫-ধাপের ফ্রেমওয়ার্ক"

- ১. **State Definition:** `dp[i]` বা `dp[i][j]` বলতে আমরা আসলে কী বোঝাচ্ছি তা পরিষ্কার করুন।
- ২. **Base Case:** সবচেয়ে ছোট সমস্যাটির উত্তর কী? (যেমন- `dp[0] = 0`)
- ৩. **Transition Relation:** বড় সমস্যাটি ছোট সমস্যাগুলোর সাথে কীভাবে সম্পর্কিত? (যেমন- `dp[i] = dp[i-1] + dp[i-2]`)
- ৪. **Result:** টেবিলের কোন ইনডেক্সে আমাদের ফাইনাল উত্তর থাকবে?
- ৫. **Complexity:** টাইম এবং স্পেস কমপ্লেক্সিটি হিসাব করুন।

---

## 📝 DP টেমপ্লেট (Tabulation)

```java
int[] dp = new int[n + 1];
dp[0] = base_case_value;

for (int i = 1; i <= n; i++) {
    dp[i] = transition_logic(dp[i-1], dp[i-2]...);
}
return dp[n];
```

## ২. কেন ডাইনামিক প্রোগ্রামিং? (Breaking the Exponential Barrier)
অনেক সমস্যা আছে যা রিকার্সন দিয়ে সমাধান করলে টাইম কমপ্লেক্সিটি এক্সপোনেনশিয়াল ($2^N$) হয়ে যায়। DP ব্যবহার করে আমরা তাকে পলিনোমিয়াল ($O(N^2)$ বা $O(N)$) এ নামিয়ে আনতে পারি।
- **উপকারিতা:** এটি অকল্পনীয়ভাবে পারফরম্যান্স ইমপ্রুভ করে।
- **মূল শর্ত:**
  - ১. **Overlapping Subproblems:** একই সাব-প্রবলেম বারবার আসতে হবে।
  - ২. **Optimal Substructure:** ছোট সমস্যার সেরা সমাধান ব্যবহার করে বড় সমস্যার সেরা সমাধান তৈরি করা যাবে।

---

## ৫. DP ক্যাটাগরি (The Frameworks)
ইন্টারভিউতে DP মূলত ৫-৬টি প্যাটার্নে আসে:
- ১. **0/1 Knapsack:** আইটেম নিব কি নিব না?
- ২. **Unbounded Knapsack:** একই আইটেম বারবার নেওয়া যাবে।
- ৩. **Longest Common Subsequence (LCS):** স্ট্রিং ম্যাচিং।
- ৪. **Longest Increasing Subsequence (LIS):** নাম্বার সিকোয়েন্স।
- ৫. **Matrix Chain Multiplication:** চেইন ডিসিশন।
- ৬. **DP on Trees/Graphs:** গ্রাফ বা ট্রিতে অপ্টিমাইজেশন।

---

## ৬. ইন্টারভিউ চেকলিস্ট (Interview Checklist)
- ১. **Define State:** `dp[i]` বা `dp[i][j]` বলতে আপনি কী বোঝাচ্ছেন তা ক্লিয়ারলি ডিফাইন করুন।
- ২. **Recursive Relation:** বর্তমান স্টেট বের করার জন্য আগের কোন স্টেটগুলো লাগবে? (যেমন: `dp[i] = dp[i-1] + dp[i-2]`)।
- ৩. **Base Case:** সবচেয়ে ছোট সমস্যার উত্তর কী? (যেমন: `dp[0] = 0`)।
- ৪. **Optimization:** আপনি কি স্পেস কমিয়ে $O(N)$ থেকে $O(1)$ এ আনতে পারেন? (Space Optimization)।

# Dynamic Programming (ডাইনামিক প্রোগ্রামিং - DP) 🧠

## Dynamic Programming (DP) কী?
ডাইনামিক প্রোগ্রামিং (DP) হলো একটি অপ্টিমাইজেশন টেকনিক। সহজ কথায়, **"যারা অতীত থেকে শেখে না, তারা একই ভুল বারবার করে!"** DP ঠিক এই কাজটিই করে। এটি অতীতকে মনে রাখে।

কোনো বড় প্রবলেমকে ছোট ছোট সাব-প্রবলেমে (Sub-problems) ভাগ করে সলভ করার পর, সেই ছোট প্রবলেমগুলোর উত্তর যদি ভবিষ্যতে আবার লাগে, তবে DP সেই উত্তরগুলোকে সেভ (Store) করে রাখে। এতে একই কাজ বারবার করতে হয় না।

*উদাহরন:* 
কেউ জিজ্ঞেস করল, "২+২+২+২ = কত?" আপনি গুনে বললেন "৮"।
এরপর সে সাথে সাথে আবার জিজ্ঞেস করল, "২+২+২+২ + ২ = কত?"
আপনি কি আবার প্রথম থেকে গুনবেন? না! আপনি আগের "৮" এর সাথে "২" যোগ করে "১০" বলে দেবেন। এটাই DP! (আগের উত্তর মনে রাখা)।

## কখন বুঝবেন যে DP ব্যবহার করতে হবে? (Pattern Recognition)
দুটি শর্ত (Properties) পূরণ হলে সেখানে DP ব্যবহার করা যায়:
- ১. **Overlapping Subproblems (ওভারল্যাপিং সাব-প্রবলেম):** বড় প্রবলেমকে ছোট করতে গেলে একই ছোট প্রবলেম বারবার আসে। (যেমন: ফিবোনাচি সিরিজে $F(5)$ বের করতে $F(3)$ বারবার লাগে)।
- ২. **Optimal Substructure (অপ্টিমাল সাবস্ট্রাকচার):** ছোট ছোট সাব-প্রবলেমের সবচেয়ে সেরা উত্তরগুলো (Optimal solutions) জোড়া লাগিয়ে বড় প্রবলেমের সেরা উত্তর পাওয়া যায়। (যেমন: Shortest path)।


## DP এর দুটি পদ্ধতি (The Two Approaches)
DP প্রবলেম মূলত দুইভাবে সলভ করা যায়:

১. **Top-Down Approach (Memoization):**
   - আমরা বড় প্রবলেম (Top) থেকে শুরু করি।
   - রিকার্সন (Recursion) ব্যবহার করে ছোট প্রবলেমের দিকে যাই।
   - ফেরার সময় উত্তরগুলো একটি অ্যারে বা ম্যাপে (Memo) সেভ করে রাখি।
   - *সুবিধা:* চিন্তা করা সহজ, রিকার্সন ট্রি আঁকলেই কোড লেখা যায়।

২. **Bottom-Up Approach (Tabulation):**
   - আমরা একদম ছোট প্রবলেম (Bottom) থেকে শুরু করি।
   - একটি টেবিল (Array/Matrix) তৈরি করে ছোট থেকে বড়র দিকে লুপ চালিয়ে (Iterative) মান পূরণ করি।
   - *সুবিধা:* রিকার্সন নেই, তাই Stack Overflow এর ভয় নেই। এটি সাধারণত Memoization এর চেয়ে সামান্য ফাস্ট হয়।
