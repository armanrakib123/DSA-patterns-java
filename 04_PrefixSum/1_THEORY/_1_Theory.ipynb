# ➕ Pattern 04: Prefix Sum (প্রিফিক্স সাম)

প্রিফিক্স সাম (Prefix Sum) হচ্ছে অ্যারের প্রবলেমগুলোকে `O(N)` টাইমে অপ্টিমাইজ করার একটি অত্যন্ত ফেমাস এবং সহজ ম্যাথমেটিক্যাল প্যাটার্ন। যখন একটি অ্যারের কোনো নির্দিষ্ট রেঞ্জ বা সাবঅ্যারের (Subarray) যোগফল বারবার বের করতে বলা হয়, তখন এই প্যাটার্নটি ম্যাজিকের মতো কাজ করে।

---

## 🧠 ইন্টুইশন (Intuition)

ধরে নিন একটি অ্যারে দেওয়া আছে: `arr = [2, 4, 6, 8, 10]`
আপনাকে বারবার প্রশ্ন করা হচ্ছে:
- "ইনডেক্স ১ থেকে ৩ পর্যন্ত যোগফল কত?" (উত্তর: ৪+৬+৮ = ১৮)
- "ইনডেক্স ০ থেকে ৪ পর্যন্ত যোগফল কত?" (উত্তর: ৩০)

যদি আপনি প্রতিবার লুপ চালিয়ে যোগ করতে যান, তবে প্রতিটি প্রশ্নের উত্তর দিতে `O(N)` সময় লাগবে। যদি ১০০০০টি প্রশ্ন থাকে, তবে টাইম কমপ্লেক্সিটি হবে `O(Q * N)`, যা ইন্টারভিউতে রিজেক্টেড।

**Prefix Sum এর ম্যাজিক:**
আমরা শুরুতে একটি নতুন অ্যারে (prefix[]) তৈরি করব, যেখানে প্রতিটি ইনডেক্সে তার আগের সব এলিমেন্টের যোগফল সেভ করা থাকবে।
`prefix[i] = prefix[i-1] + arr[i]`

আমাদের অ্যারের জন্য প্রিফিক্স অ্যারেটি হবে:
`prefix = [2, 6, 12, 20, 30]`
(অর্থাৎ `prefix[2]` মানে হলো ইনডেক্স ০ থেকে ২ পর্যন্ত যোগফল ১২)

এখন যদি কেউ জিজ্ঞেস করে "ইনডেক্স ১ থেকে ৩ পর্যন্ত যোগফল কত?"
লজিক: `sum(i to j) = prefix[j] - prefix[i - 1]`
যোগফল = `prefix[3] - prefix[0]` = `20 - 2 = 18`। 
আমরা লুপ ছাড়াই `O(1)` সময়ে উত্তর পেয়ে গেলাম!

---

## 🛠️ প্রিফিক্স সামের ব্যবহার

- ১. **1D Range Queries:** উপরে যে উদাহরণ দেওয়া হলো।
- ২. **2D Matrix Range Queries:** ইমেজ প্রসেসিং বা গ্রিড প্রবলেমে একটি নির্দিষ্ট আয়তক্ষেত্রের যোগফল বের করতে 2D Prefix Sum লাগে।
- ৩. **Subarray with given Target:** যখন বলা হয় এমন সাবঅ্যারে বের করতে যার যোগফল `K` এর সমান। তখন Prefix Sum এবং HashMap এর কম্বিনেশন ব্যবহার করা হয়। (Software Company এ সবচেয়ে বেশি আসে)।
- ৪. **Equilibrium Index:** অ্যারের এমন একটি ইনডেক্স বের করা যার বাম দিকের যোগফল এবং ডান দিকের যোগফল সমান।

---

## 🚨 কখন বুঝবেন Prefix Sum ব্যবহার করতে হবে?

- ১. প্রবলেমে যদি **Subarray Sum** (সাবঅ্যারের যোগফল) নিয়ে কথা বলা হয়, কিন্তু স্লাইডিং উইন্ডো ব্যবহার করা যাচ্ছে না কারণ অ্যারেতে নেগেটিভ সংখ্যা আছে। (স্লাইডিং উইন্ডো নেগেটিভ সংখ্যা থাকলে কাজ করে না, তখন প্রিফিক্স সাম + হ্যাশম্যাপ লাগে)।
- ২. যদি অ্যারের একটি নির্দিষ্ট রেঞ্জের (Range) ওপর বারবার কোনো অপারেশন (Query) চালাতে হয়।
- ৩. যদি `O(N)` সময়ে সাবঅ্যারে খোঁজার দরকার পড়ে।

---

## 📝 কোড টেমপ্লেট

### বেসিক 1D Prefix Sum
```java
int[] prefix = new int[arr.length];
prefix[0] = arr[0];

for (int i = 1; i < arr.length; i++) {
    prefix[i] = prefix[i - 1] + arr[i];
}

public int rangeSum(int[] prefix, int i, int j) {
    if (i == 0) return prefix[j];
    return prefix[j] - prefix[i - 1];
}
```

### Prefix Sum + HashMap (খুব গুরুত্বপূর্ণ)
"Total subarrays with sum equals K"
```java
public int subarraySum(int[] nums, int k) {
    int count = 0, currentSum = 0;
    HashMap<Integer, Integer> map = new HashMap<>();
    
    map.put(0, 1); 
    
    for (int num : nums) {
        currentSum += num;
        
        if (map.containsKey(currentSum - k)) {
            count += map.get(currentSum - k);
        }
        
        map.put(currentSum, map.getOrDefault(currentSum, 0) + 1);
    }
    return count;
}
```

## 🏆 এই ফোল্ডারের প্রবলেমগুলো
1. **RangeSumQuery.java (Easy):** 1D রেঞ্জ কুয়েরির সরাসরি ইমপ্লিমেন্টেশন।
2. **SubarraySumEqualsK.java (Medium):** Prefix Sum + HashMap এর সবচেয়ে ক্লাসিক এবং ফেমাস প্রবলেম। এটি নেগেটিভ সংখ্যার জন্য স্লাইডিং উইন্ডোর বিকল্প।
3. **ContinuousSubarraySum.java (Medium):** একটু ট্রিকি ভার্সন যেখানে মডুলো (Modulo / `%`) অ্যারিথমেটিক ব্যবহার করতে হয়।

প্রথমে নিজে নিজে ট্রাই করুন, না পারলে কোড এবং কমেন্টগুলো পড়ুন!


# Prefix Sum (প্রিফিক্স সাম) - The Software Company Standard Theory ➕

## ১. ভূমিকা (Detailed Introduction)
প্রিফিক্স সাম (Prefix Sum) হলো একটি অত্যন্ত শক্তিশালী প্রি-প্রসেসিং (Pre-processing) টেকনিক যা মূলত রেঞ্জ কোয়েরি (Range Queries) এবং সাব-অ্যারে সাম (Subarray Sum) সংক্রান্ত সমস্যা দ্রুত সমাধান করতে ব্যবহৃত হয়। Software Company ইন্টারভিউতে যখনই আপনাকে অনেকগুলো কোয়েরি হ্যান্ডেল করতে বলা হবে বা নেগেটিভ নাম্বার যুক্ত সাব-অ্যারে সাম বের করতে হবে, তখনই প্রিফিক্স সাম আপনার সবচেয়ে বড় বন্ধু হয়ে দাঁড়াবে।

সহজ কথায়, প্রিফিক্স সাম হলো অ্যারের শুরু থেকে একটি নির্দিষ্ট ইনডেক্স পর্যন্ত সব ইলিমেন্টের যোগফল আগে থেকেই ক্যালকুলেট করে রাখা। এটি "Compute once, use many times" ফিলোসফিতে কাজ করে।

---

## ২. কেন প্রিফিক্স সাম? (The Optimization Logic)
ধরা যাক আমাদের কাছে একটি বিশাল অ্যারে আছে এবং আমাদের বারবার জিজ্ঞেস করা হচ্ছে, "ইনডেক্স $i$ থেকে $j$ পর্যন্ত সব সংখ্যার যোগফল কত?"
- **ব্রুট ফোর্স:** প্রতিবার লুপ চালিয়ে যোগ করা। যদি $Q$ টি কোয়েরি থাকে, তবে টাইম কমপ্লেক্সিটি হবে $O(Q \times N)$।
- **প্রিফিক্স সাম:** একবার $O(N)$ টাইমে সব যোগফল বের করে রাখলে, প্রতিটি কোয়েরি মাত্র $O(1)$ টাইমে উত্তর দেওয়া যায়। টোটাল টাইম: $O(N + Q)$।

এটি বিশাল পরিমাণ ডেটা রিড (Read) করার সময় পারফরম্যান্স কয়েক হাজার গুণ বাড়িয়ে দিতে পারে।

---

## ৩. প্রিফিক্স সাম কিভাবে কাজ করে? (Mathematical Logic)
ধরা যাক আমাদের অ্যারে হলো: `A = [2, 4, 1, 3]`
প্রিফিক্স সাম অ্যারে হবে: `P = [2, 6, 7, 10]`
এখানে `P[i]` হলো `A[0]` থেকে `A[i]` পর্যন্ত যোগফল।

এখন যদি আমাদের ২ নম্বর ইনডেক্স থেকে ৩ নম্বর ইনডেক্সের যোগফল লাগে:
$Sum(2, 3) = P[3] - P[1] = 10 - 6 = 4$
**সূত্র:** $Sum(i, j) = P[j] - P[i-1]$

# Prefix Sum (প্রিফিক্স সাম) ➕

## Prefix Sum কী?
"Prefix" মানে হলো শুরুর দিক থেকে বা আগে থেকে। **Prefix Sum** হলো এমন একটি টেকনিক যেখানে আমরা একটি Array এর শুরু থেকে বর্তমান ইনডেক্স পর্যন্ত সকল ইলিমেন্টের যোগফল আগে থেকেই হিসাব করে (pre-compute) একটি নতুন Array তে সেভ করে রাখি।

ধরি একটি Array আছে: `A = [1, 2, 3, 4, 5]`
এর Prefix Sum Array হবে: `P = [1, 3, 6, 10, 15]`
এখানে `P[i]` হলো `A[0]` থেকে `A[i]` পর্যন্ত সব ইলিমেন্টের যোগফল।

## কখন বুঝবেন যে Prefix Sum ব্যবহার করতে হবে? (Pattern Recognition)
১. **Range Queries (রেঞ্জ কোয়েরি):** যদি আপনাকে বারবার জিজ্ঞাসা করা হয় যে "Array এর ইনডেক্স $i$ থেকে $j$ পর্যন্ত ইলিমেন্টগুলোর যোগফল কত?"
২. **Subarray Sum:** যদি বলা হয় এমন সাব-অ্যারে খুঁজে বের করতে যার যোগফল নির্দিষ্ট একটি টার্গেট ($K$) এর সমান, অথবা যেটির যোগফল $K$ দ্বারা বিভাজ্য (Divisible)।
৩. **Multiple Queries:** যখন একই Array এর উপর অনেকগুলো যোগফলের কোয়েরি থাকে, তখন Brute Force $O(N)$ এ কাজ করলে পুরো কোয়েরি প্রসেস করতে $O(N \times Q)$ সময় লাগে। Prefix Sum এটি $O(1)$ এ নামিয়ে আনে।

## Range Sum কীভাবে কাজ করে?
ধরা যাক আপনাকে বলা হলো `i` থেকে `j` পর্যন্ত ইলিমেন্টের যোগফল বের করতে।
- **Brute Force:** `i` থেকে `j` পর্যন্ত লুপ চালিয়ে যোগ করা ($O(N)$)।
- **Prefix Sum:** শুধু `P[j] - P[i-1]` হিসাব করা! (Time Complexity: $O(1)$)

*উদাহরন:* `A = [1, 2, 3, 4, 5]`, `P = [1, 3, 6, 10, 15]`
বলা হলো ইনডেক্স 1 থেকে 3 (অর্থাৎ 2+3+4 = 9) এর যোগফল কত।
সূত্র অনুযায়ী: `P[3] - P[0]` = `10 - 1` = 9!
ম্যাজিক, তাই না?

## HashMap + Prefix Sum (The Ultimate Combo) 🔥
ইন্টারভিউতে শুধু সাধারণ Prefix Sum আসে না। বরং **Prefix Sum এর সাথে HashMap** মিলিয়ে প্রবলেম দেওয়া হয়। 
- আপনি যখন কোনো সাব-অ্যারের যোগফল $K$ খুঁজছেন, তার মানে হলো: `current_prefix_sum - previous_prefix_sum = K`
- সমীকরণটি একটু ঘুরালে: `previous_prefix_sum = current_prefix_sum - K`
- আমরা একটি HashMap এ সব `previous_prefix_sum` সেভ করে রাখি। যদি দেখি `current_prefix_sum - K` ম্যাপে আছে, তার মানে আমরা $K$ যোগফলের সাব-অ্যারে পেয়ে গেছি!

পরবর্তী ফাইলগুলোতে আমরা এর জেনেরিক টেমপ্লেট এবং উদাহরণ দেখব।
