## ১. Course Schedule (LC 207) 🟡
**টাইপ:** Cycle Detection in Directed Graph

### প্রবলেম স্টেটমেন্ট:
আপনাকে কিছু কোর্স এবং তাদের প্রিরিকুইজিট (Prerequisites) দেওয়া আছে। সব কোর্স কি শেষ করা সম্ভব?

### সমাধান:
এটি একটি ক্ল্যাসিক সাইকেল ডিটেকশন প্রবলেম। যদি কোর্সের ডিপেন্ডেন্সি গ্রাফে কোনো সাইকেল থাকে, তবে কোনো কোর্সই শেষ করা যাবে না। আমরা কান'স অ্যালগরিদম (Kahn's Algorithm) ব্যবহার করে যদি সব নোড ভিজিট করতে পারি, তবে উত্তর 'True', নাহলে 'False'।

## ৩. Alien Dictionary (LC 269) 🔴
**টাইপ:** Dependency Graph Construction + Topo Sort

### প্রবলেম স্টেটমেন্ট:
একটি এলিয়েন ল্যাঙ্গুয়েজের সর্টেড ওয়ার্ড লিস্ট দেওয়া আছে। ক্যারেক্টারগুলোর অর্ডার (যেমন: a এর পর b) বের করুন।

### সমাধান:
১. পাশাপাশি দুটি ওয়ার্ডের ক্যারেক্টার তুলনা করে ডিপেন্ডেন্সি গ্রাফ তৈরি করুন।
২. ক্যারেক্টারগুলোর ওপর টপোলজিক্যাল সর্ট চালান।
৩. যদি কোনো ইনভ্যালিড কেস (যেমন: সাইকেল বা ছোট ওয়ার্ড বড় ওয়ার্ডের পরে থাকা) পাওয়া যায়, তবে খালি স্ট্রিং রিটার্ন করুন।
