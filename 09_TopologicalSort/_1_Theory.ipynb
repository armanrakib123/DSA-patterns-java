# 🎓 Pattern 33: Topological Sort (টপোলজিকাল সর্ট)

টপোলজিকাল সর্ট (Topological Sort) হলো একটি ডিরেক্টেড অ্যাসাইক্লিক গ্রাফের (Directed Acyclic Graph - DAG) নোডগুলোকে এমনভাবে সাজানো যেন প্রতিটি এজ `u -> v` এর জন্য নোড `u` সবসময় নোড `v` এর আগে থাকে। এটি মূলত কোনো কাজের ধারাক্রম (Dependency order) ঠিক করার জন্য ব্যবহৃত হয়।

---

## 🧠 ইন্টুইশন (Intuition)

সহজ কথায়, টপোলজিকাল সর্ট হলো "নির্ভরশীলতা" (Dependency) অনুযায়ী সাজানো। 
**উদাহরণ:**
- ইউনিভার্সিটির একটি কোর্স করার আগে আপনাকে তার "Pre-requisite" কোর্সটি করতে হবে। 
- ডাটাবেস শেখার আগে আপনাকে হয়তো প্রোগ্রামিং ল্যাংগুয়েজ শিখতে হবে। 

**শর্ত:**
- ১. গ্রাফটি অবশ্যই **Directed** হতে হবে।
- ২. গ্রাফে কোনো **Cycle** থাকতে পারবে না (Acyclic)। যদি সাইকেল থাকে, তবে সর্টিং করা সম্ভব নয়।

---

## 🚀 সলভিং স্ট্র্যাটেজি

টপোলজিকাল সর্ট করার দুটি প্রধান পদ্ধতি আছে:

### ১. DFS পদ্ধতি (Postorder logic)
- একটি DFS চালান। 
- যখন কোনো নোডের সব নেইবার ভিজিট করা শেষ হবে (অর্থাৎ ওই নোডের ওপর আর কেউ ডিপেন্ডেন্ট নয়), তখন নোডটিকে একটি **স্ট্যাকে (Stack)** পুশ করুন। 
- শেষে স্ট্যাকটি উল্টো করে প্রিন্ট করলেই সর্টেড অর্ডার পাওয়া যাবে।

### ২. Kahn's Algorithm (BFS পদ্ধতি)
- এটি **In-degree** (একটি নোডে কতগুলো ইনকামিং এজ আছে) এর ওপর ভিত্তি করে কাজ করে। 
- যেসব নোডের ইন-ডিগ্রি ০ (অর্থাৎ তারা কারো ওপর ডিপেন্ডেন্ট নয়), তাদের কিউতে অ্যাড করুন। 
- কিউ থেকে নোড বের করার সময় তার নেইবারদের ইন-ডিগ্রি ১ কমিয়ে দিন। 
- যাদের ইন-ডিগ্রি ০ হবে, তাদের কিউতে অ্যাড করুন।

---

## 🚨 কখন বুঝবেন Topological Sort ব্যবহার করতে হবে?

- ১. যখনই প্রবলেমে **"Dependency"**, **"Ordering"**, বা **"Prerequisite"** এর কথা থাকবে।
- ২. সফটওয়্যার প্যাকেজ ইন্সটলেশন বা কম্পাইলেশন অর্ডার ঠিক করার জন্য।
- ৩. কোর্স সিডিউলিং প্রবলেমে।
- ৪. টাস্ক ম্যানেজমেন্ট সিস্টেমে কোন কাজের পর কোন কাজ করতে হবে তা বের করতে।

# 🎓 Pattern 33: Topological Sort (টপোলজিকাল সর্ট)

সহজ কথায়, টপোলজিকাল সর্ট হলো "নির্ভরশীলতা" (Dependency) অনুযায়ী সাজানো। 
**উদাহরণ:**
- ইউনিভার্সিটির একটি কোর্স করার আগে আপনাকে তার "Pre-requisite" কোর্সটি করতে হবে। 
- ডাটাবেস শেখার আগে আপনাকে হয়তো প্রোগ্রামিং ল্যাংগুয়েজ শিখতে হবে। 

**শর্ত:**

টপোলজিকাল সর্ট করার দুটি প্রধান পদ্ধতি আছে:

- একটি DFS চালান।

- শেষে স্ট্যাকটি উল্টো করে প্রিন্ট করলেই সর্টেড অর্ডার পাওয়া যাবে।

- এটি **In-degree** (একটি নোডে কতগুলো ইনকামিং এজ আছে) এর ওপর ভিত্তি করে কাজ করে। 
- যেসব নোডের ইন-ডিগ্রি ০ (অর্থাৎ তারা কারো ওপর ডিপেন্ডেন্ট নয়), তাদের কিউতে অ্যাড করুন। 
- কিউ থেকে নোড বের করার সময় তার নেইবারদের ইন-ডিগ্রি ১ কমিয়ে দিন। 
- যাদের ইন-ডিগ্রি ০ হবে, তাদের কিউতে অ্যাড করুন।

# Topological Sort (টপোলজিক্যাল সর্ট) 🔄

## Topological Sort কী?
Topological Sort হলো একটি Directed Acyclic Graph (DAG) এর নোডগুলোকে এমন একটি লিনিয়ার (Linear) অর্ডারে সাজানো, যেখানে প্রতিটি এজ $U \rightarrow V$ এর জন্য, অর্ডারিংয়ে $U$ অবশ্যই $V$ এর আগে থাকবে।

সহজ কথায়: **কাজের ধারাবাহিকতা (Dependency Order)**। 
*উদাহরন:* আপনি যদি মোজা না পরে জুতা পরতে না পারেন, তবে "মোজা পরা" হলো $U$ এবং "জুতা পরা" হলো $V$। টপোলজিক্যাল সর্টে "মোজা পরা" অবশ্যই আগে আসবে।

## DAG (Directed Acyclic Graph) কী?
টপোলজিক্যাল সর্ট শুধুমাত্র DAG তেই সম্ভব।
- **Directed:** গ্রাফের এজে অ্যারো (Arrow/Direction) থাকতে হবে ($A \rightarrow B$ মানে A এর ওপর B নির্ভরশীল)।
- **Acyclic:** গ্রাফে কোনো সাইকেল (Cycle/Loop) থাকতে পারবে না। যদি A এর জন্য B লাগে, এবং B এর জন্য আবার A লাগে, তবে আপনি কোনো কাজই শুরু করতে পারবেন না (Deadlock)।

## কখন বুঝবেন যে Topological Sort ব্যবহার করতে হবে? (Pattern Recognition)
১. **Dependencies/Prerequisites:** প্রবলেমে যদি বলা হয় "Task A must be done before Task B" বা "Course A is a prerequisite of Course B" (যেমন: Course Schedule)।
২. **Ordering/Sorting:** এমন কিছু সর্ট করতে বলা হলে যেখানে একের ওপর অন্যের নির্ভরতা আছে (যেমন: Alien Dictionary)।

## অ্যালগরিদম কীভাবে কাজ করে? (Kahn's Algorithm)
আমরা Kahn's Algorithm (BFS এপ্রোচ) ব্যবহার করে টপোলজিক্যাল সর্ট করব। এর জন্য আমাদের **In-degree** (ভিতরের দিকে আসা এজের সংখ্যা) জানতে হবে।
- **In-degree = 0** মানে হলো এই নোডটি কারো ওপর নির্ভরশীল নয়। আমরা এটি দিয়েই কাজ শুরু করতে পারি।

**স্টেপস:**
১. গ্রাফ তৈরি করুন (Adjacency List) এবং প্রতিটি নোডের In-degree কাউন্ট করুন।
২. যাদের In-degree $0$, তাদের সবাইকে একটি Queue তে ঢুকিয়ে দিন।
৩. Queue থেকে একটি নোড বের করুন এবং রেজাল্টে (লিস্টে) যোগ করুন।
৪. ওই নোডের সব প্রতিবেশীর (যাদের দিকে সে পয়েন্ট করে আছে) In-degree ১ করে কমিয়ে দিন (কারণ তাদের একটি ডিপেন্ডেন্সি পূরণ হয়ে গেছে)।
৫. কমানোর পর যদি কোনো প্রতিবেশীর In-degree $0$ হয়ে যায়, তবে তাকেও Queue তে ঢুকিয়ে দিন।
৬. Queue খালি হওয়া পর্যন্ত ৩-৫ পুনরাবৃত্তি করুন।

**Cycle Detection:**
লুপ শেষে যদি রেজাল্ট লিস্টের সাইজ মোট নোডের সাইজের সমান না হয়, তার মানে গ্রাফে একটি Cycle ছিল এবং সবগুলো কাজ করা সম্ভব হয়নি!

পরবর্তী ফাইলগুলোতে আমরা Kahn's Algorithm এবং DFS এপ্রোচ দেখব।

## ১. ভূমিকা (Detailed Introduction)
টপোলজিক্যাল সর্ট (Topological Sort) হলো গ্রাফ থিওরির একটি বিশেষ ধরনের লিনিয়ার অর্ডারিং (Linear Ordering)। এটি শুধুমাত্র ডাইরেক্টেড অ্যাসাইক্লিক গ্রাফ (Directed Acyclic Graph বা DAG) এর ওপর কাজ করে। এর মূল উদ্দেশ্য হলো গ্রাফের নোডগুলোকে এমনভাবে সাজানো যাতে প্রতিটি ডাইরেক্টেড এজ $u \to v$ এর জন্য, নোড $u$ সবসময় নোড $v$ এর আগে থাকে।

সহজ কথায়, টপোলজিক্যাল সর্ট হলো "ডিপেন্ডেন্সি" (Dependency) ম্যানেজ করার একটি উপায়। আপনি যদি একটি কাজ করার আগে অন্য একটি কাজ শেষ করতে চান, তবে সেই কাজের ক্রম বের করার জন্য টপোলজিক্যাল সর্ট অপরিহার্য।

## ২. কেন টপোলজিক্যাল সর্ট? (The Dependency Logic)
FAANG কোম্পানিগুলোতে হাজার হাজার মাইক্রোসার্ভিস (Microservices) একে অপরের ওপর নির্ভরশীল। একটি সার্ভিস চালু করার আগে তার ডিপেন্ডেন্ট সার্ভিসগুলো চালু করতে হয়। এই জটিল ক্রম বের করার জন্য টপোলজিক্যাল সর্ট ব্যবহার করা হয়। এটি নিশ্চিত করে যে কোনো কাজ তার ডিপেন্ডেন্সি শেষ না করে শুরু হবে না।

## ৩. ব্যবহারের শর্ত (The DAG Rule)
টপোলজিক্যাল সর্ট শুধুমাত্র DAG এর ওপর কাজ করে।
- **Directed:** এজগুলোর নির্দিষ্ট দিক থাকতে হবে।
- **Acyclic:** গ্রাফে কোনো লুপ বা সাইকেল থাকতে পারবে না। যদি গ্রাফে সাইকেল থাকে, তবে টপোলজিক্যাল সর্ট করা অসম্ভব, কারণ সেখানে একটি ডেডলক তৈরি হবে।

## ৪. কান'স অ্যালগরিদম (Kahn's Algorithm - BFS)
কান'স অ্যালগরিদম হলো একটি ইন-ডিগ্রি (In-degree) ভিত্তিক পদ্ধতি।
১. প্রতিটি নোডের ইন-ডিগ্রি (কতগুলো এজ ওই নোডে ঢুকেছে) ক্যালকুলেট করুন।
২. যে নোডগুলোর ইন-ডিগ্রি ০, তাদের কিউ-তে (Queue) পুশ করুন।
৩. কিউ থেকে নোড বের করুন এবং তার নেইবারগুলোর ইন-ডিগ্রি ১ কমিয়ে দিন।
৪. যদি কোনো নেইবারের ইন-ডিগ্রি ০ হয়, তবে তাকে কিউ-তে পুশ করুন।

এটি মূলত BFS এর একটি মডিফাইড ভার্সন যা খুব সহজেই সাইকেল ডিটেক্ট করতে পারে।

## ৫. রিয়েল-ওয়ার্ল্ড FAANG উদাহরণ (Real-World Case Studies)

### ৫.১. গুগল ক্লাউড - বিল্ড সিস্টেম (Bazel)
গুগলের বিল্ড সিস্টেম 'Bazel' হাজার হাজার লাইব্রেরি এবং মডিউল কম্পাইল করার জন্য টপোলজিক্যাল সর্ট ব্যবহার করে। এটি নিশ্চিত করে যে একটি লাইব্রেরি কম্পাইল করার আগে তার সব ডিপেন্ডেন্সি লাইব্রেরিগুলো কম্পাইল হয়েছে। এটি বিল্ড টাইম অনেক কমিয়ে দেয়।

### ৫.২. আমাজন - সাপ্লাই চেইন ম্যানেজমেন্ট
আমাজনের লজিস্টিক সিস্টেম যখন একটি অর্ডার প্রসেস করে, তখন অনেকগুলো ধাপ (Payment, Inventory, Packing, Shipping) থাকে। এই ধাপগুলোর মধ্যে যে ডিপেন্ডেন্সি আছে, তা টপোলজিক্যাল সর্ট ব্যবহার করে শিডিউল করা হয় যাতে কোনো ভুল না হয়।

### ৫.৩. অপারেটিং সিস্টেম - প্যাকেজ ম্যানেজার (apt, npm)
আপনি যখন `npm install` বা `apt-get install` করেন, তখন ওই প্যাকেজের সাথে আরও অনেকগুলো প্যাকেজ ডাউনলোড হয়। প্যাকেজ ম্যানেজার একটি ডিপেন্ডেন্সি গ্রাফ তৈরি করে এবং টপোলজিক্যাল সর্ট চালিয়ে বের করে কোন অর্ডারে প্যাকেজগুলো ইনস্টল করতে হবে।
