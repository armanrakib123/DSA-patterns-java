# Dynamic (Variable) Sliding Window 📏

Fixed Window তে আমাদের উইন্ডোর সাইজ `K` আগে থেকেই বলে দেওয়া থাকতো। কিন্তু **Dynamic Sliding Window** তে উইন্ডোর সাইজ ফিক্সড থাকে না। এটি প্রবলেমের একটি নির্দিষ্ট শর্ত (Condition) পূরণ করার ওপর ভিত্তি করে বড় (Grow) বা ছোট (Shrink) হতে থাকে।

## কখন বুঝবেন এটি Dynamic Window এর প্রবলেম?
ইন্টারভিউতে প্রবলেম স্টেটমেন্টে সাধারণত দুটি জিনিস চাওয়া হয়:
1. **Longest / Maximum Subarray/Substring:** "Find the longest substring with at most K distinct characters."
2. **Shortest / Minimum Subarray/Substring:** "Find the minimum length subarray with a sum greater than or equal to S."

অর্থাৎ, আপনাকে উইন্ডোর সাইজ (Length) টাই বের করতে হবে, যা একটি নির্দিষ্ট শর্ত পূরণ করে।

## লজিক (The General Template)
Dynamic Window এর কোর লজিকটি খুবই চমৎকার এবং এটি একটি নির্দিষ্ট টেমপ্লেট ফলো করে। আমরা দুটি পয়েন্টার ব্যবহার করি: `left` এবং `right`। উভয়ই `0` ইনডেক্স থেকে শুরু হয়।

1. **Grow the Window (উইন্ডো বড় করো):** 
   - `right` পয়েন্টারটিকে ডান দিকে সরাতে থাকি এবং নতুন এলিমেন্টগুলোকে আমাদের উইন্ডোতে (বা HashMap/Sum এ) অ্যাড করতে থাকি।
   - এতে উইন্ডোর সাইজ বড় হতে থাকে।
   
2. **Check Condition (শর্ত চেক করো):**
   - প্রতিবার নতুন এলিমেন্ট অ্যাড করার পর চেক করি উইন্ডোটি কি প্রবলেমের শর্ত ভঙ্গ (Violate) করেছে?
   
3. **Shrink the Window (উইন্ডো ছোট করো):**
   - যদি শর্ত ভঙ্গ হয় (Invalid Window), তবে আমরা `left` পয়েন্টারটিকে ডান দিকে সরাতে শুরু করি। 
   - `left` সরানোর সময় আমরা উইন্ডো থেকে এলিমেন্টগুলো রিমুভ (Subtract/Delete) করতে থাকি।
   - কতক্ষণ `left` সরাব? যতক্ষণ না উইন্ডোটি আবার ভ্যালিড (Valid) হচ্ছে।
   
4. **Update Result (রেজাল্ট আপডেট করো):**
   - উইন্ডোটি যখন ভ্যালিড অবস্থায় থাকে, তখন আমরা আমাদের ম্যাক্সিমাম বা মিনিমাম লেংথ আপডেট করি।

## উদাহরণ:
Longest Substring Without Repeating Characters:
- `right` বাড়াতে থাকব এবং ক্যারেক্টার অ্যাড করব।
- যদি দেখি একই ক্যারেক্টার দুইবার চলে এসেছে (Invalid), তখন `left` কে বাড়িয়ে আগের ক্যারেক্টারগুলো ডিলিট করতে থাকব যতক্ষণ না ডুপ্লিকেটটি রিমুভ হচ্ছে (Valid)।
- প্রতি ভ্যালিড স্টেপে `max_len = max(max_len, right - left + 1)` আপডেট করব।
