# Optimization & Complexity (Sliding Window) ⚙️

Sliding Window প্যাটার্নটি শেখার মূল উদ্দেশ্যই হলো $O(N^2)$ বা $O(N^3)$ কমপ্লেক্সিটির কোডকে $O(N)$ এ নামিয়ে আনা। ইন্টারভিউতে যখন আপনি আপনার সলিউশনের কমপ্লেক্সিটি বলবেন, তখন কিছু বিষয়ে সতর্ক থাকতে হবে।

## Time Complexity: O(N)
প্রায় সব Sliding Window প্রবলেমের টাইম কমপ্লেক্সিটি $O(N)$ হয়। 
এখানে একটি কনফিউশন প্রায়ই হয়:
> *"আমার কোডে তো একটি `for` লুপের ভেতরে আরেকটি `while` লুপ আছে। তাহলে এটি O(N^2) নয় কেন?"*

**FAANG Answer:**
যদিও এখানে নেস্টেড লুপ আছে, কিন্তু প্রতিটি এলিমেন্টকে (Array বা String এর) সর্বোচ্চ **দুইবার** প্রোসেস করা হচ্ছে।
1. একবার `right` পয়েন্টার দ্বারা উইন্ডোতে অ্যাড করার সময়।
2. আরেকবার `left` পয়েন্টার দ্বারা উইন্ডো থেকে রিমুভ করার সময়।

যদি Array এর সাইজ $N$ হয়, তবে `right` পয়েন্টার $N$ বার মুভ করবে এবং `left` পয়েন্টারও সর্বোচ্চ $N$ বার মুভ করবে। তাই মোট অপারেশন সংখ্যা হবে $2N$। 
Big-O নোটিশনে কনস্ট্যান্ট ২ বাদ দিলে, টাইম কমপ্লেক্সিটি দাঁড়ায় $O(N)$। 

## Space Complexity
Sliding Window এর Space Complexity প্রবলেমের ধরনের ওপর নির্ভর করে।

1. **O(1) Space (Constant Space):**
   যদি আপনি শুধু যোগফল (Sum) বা গুণফল (Product) মেইনটেইন করেন, তবে আপনার এক্সট্রা স্পেস লাগবে $O(1)$। (যেমন: Minimum Size Subarray Sum)।

2. **O(K) বা O(1) Space (HashMap বা Array Frequency):**
   String এর প্রবলেমগুলোতে (যেমন Longest Substring Without Repeating Characters) ক্যারেক্টার ট্র্যাক করার জন্য আমাদের `HashMap` বা `int[256]` (বা `int[26]`) সাইজের Frequency Array ব্যবহার করতে হয়। 
   - যদি ক্যারেক্টার সেট ফিক্সড থাকে (যেমন শুধু ইংরেজি ছোট হাতের অক্ষর), তবে সাইজ হবে 26, যা $O(1)$।
   - যদি সমস্ত ASCII ক্যারেক্টার হয়, তবে সাইজ হবে 128 বা 256, যা ও $O(1)$।
   - তবে যদি কোনো আনলিমিটেড রেঞ্জের অবজেক্ট বা ডাটা ট্র্যাক করতে হয়, তবে স্পেস কমপ্লেক্সিটি $O(K)$ হতে পারে (যেখানে K হলো ইউনিক এলিমেন্টের সংখ্যা)।

## Edge Cases to Consider ⚠️
Sliding Window ইমপ্লিমেন্ট করার সময় নিচের বিষয়গুলো খেয়াল রাখবেন:
1. **Window Size > Array Size:** Fixed window এর ক্ষেত্রে যদি `K` এর মান Array এর সাইজের চেয়ে বড় হয়, তবে আগেই `return 0` বা কোনো এরর হ্যান্ডেল করতে হবে।
2. **Negative Numbers:** ম্যাক্সিমাম সাম সাব-অ্যারে বের করার ক্ষেত্রে যদি Array তে নেগেটিভ সংখ্যা থাকে, তবে সাধারণ Sliding Window কাজ নাও করতে পারে (সেক্ষেত্রে Kadane's Algorithm ব্যবহার করতে হয়)। Sliding Window (Variable) সাধারণত তখন ভালো কাজ করে যখন এলিমেন্টগুলো **Positive** হয় (যাতে উইন্ডো বাড়লে যোগফল বাড়ে এবং উইন্ডো কমলে যোগফল কমে, অর্থাৎ Monotonic property বজায় থাকে)।
