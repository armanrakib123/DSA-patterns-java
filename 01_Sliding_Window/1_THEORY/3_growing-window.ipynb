# Growing Window (Maximum Window) 📈

Dynamic Sliding Window এর আরেকটি ধরন হলো **Growing Window**। এটি ব্যবহার করা হয় যখন প্রবলেমে **"Maximum length"** বা **"Longest subarray/substring"** বের করতে বলা হয়।

## মূল কনসেপ্ট
Growing Window এর প্রবলেমগুলোতে আমরা চাই আমাদের উইন্ডোটি যতটা সম্ভব বড় হোক, তবে একটি নির্দিষ্ট শর্তের (Constraints) ভেতরে থেকে। 

এক্ষেত্রে আমরা `right` পয়েন্টার বাড়িয়ে উইন্ডোটিকে বড় করতে থাকি। যতক্ষণ উইন্ডোটি **Valid** থাকে, আমরা আমাদের ম্যাক্সিমাম রেজাল্ট আপডেট করতে থাকি।
কিন্তু যদি নতুন এলিমেন্ট অ্যাড করার ফলে উইন্ডোটি **Invalid** হয়ে যায় (অর্থাৎ শর্ত ভঙ্গ হয়), তখন আমরা `left` পয়েন্টার বাড়িয়ে উইন্ডোটিকে ছোট করতে থাকি, ঠিক যতক্ষণ না উইন্ডোটি আবার **Valid** অবস্থায় ফিরে আসে।

## স্টেপ-বাই-স্টেপ লজিক
ধরা যাক প্রবলেম: "Longest Subarray with sum $\le$ K".

1. **Grow:** 
   `right` বাড়াও এবং এলিমেন্টগুলো উইন্ডোতে যোগ করো।
2. **Valid Check:**
   যদি বর্তমান যোগফল $\le$ K হয়, তার মানে উইন্ডোটি ভ্যালিড। আমরা আমাদের `maxLength` আপডেট করব: `maxLength = Math.max(maxLength, right - left + 1)`.
3. **Invalid State:**
   যদি যোগফল > K হয়ে যায়, তখন উইন্ডোটি ইনভ্যালিড। 
4. **Shrink to make Valid:**
   যেহেতু উইন্ডোটি ইনভ্যালিড, আমাদের `left` পয়েন্টার ডান দিকে সরিয়ে কিছু এলিমেন্ট বাদ দিতে হবে, যাতে যোগফল আবার $\le$ K হয়।

## Code Structure (While Loop for Shrinking)
```java
int maxLen = 0;
for (int right = 0; right < arr.length; right++) {
    // Add arr[right] to window state
    
    while (Condition Is Invalid) {
        // Remove arr[left] from window state
        left++; // Shrink window until it becomes valid again
    }
    
    // Window is valid now, update maximum length
    maxLen = Math.max(maxLen, right - left + 1);
}
```

## Important Note for FAANG
অনেক সময় ম্যাক্সিমাম উইন্ডো বের করার ক্ষেত্রে ভেতরের `while` লুপটিকে একটি `if` দিয়ে রিপ্লেস করা যায়, যদি আমরা `maxLen` আপডেট করার বদলে উইন্ডোটির সাইজ ফিক্সড রেখে স্লাইড করাই। তবে ইন্টারভিউতে উপরের `while` লুপের এপ্রোচটি (যেখানে উইন্ডোটি আক্ষরিক অর্থেই ছোট-বড় হয়) বোঝানো অনেক সহজ এবং এটি সব ধরনের টেস্ট কেস কভার করে।
