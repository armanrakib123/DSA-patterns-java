# Shrinking Window (Minimum Window) 📉

Dynamic Sliding Window এর একটি ধরন হলো **Shrinking Window**। এটি মূলত ব্যবহার করা হয় যখন প্রবলেমে **"Minimum length"** বা **"Shortest subarray/substring"** বের করতে বলা হয়।

## মূল কনসেপ্ট
Shrinking Window এর প্রবলেমগুলোতে শুরুতে আমাদের উইন্ডোটি **Invalid** থাকে (কারণ শর্ত পূরণ হয় না)। আমরা `right` পয়েন্টার বাড়িয়ে উইন্ডোটিকে বড় করতে থাকি যতক্ষণ না এটি প্রবলেমের শর্ত পূরণ করে (অর্থাৎ, **Valid** হয়)।

একবার উইন্ডোটি Valid হয়ে গেলে, আমরা আমাদের রেজাল্ট আপডেট করি এবং এরপর উইন্ডোটিকে ছোট (Shrink) করার চেষ্টা করি `left` পয়েন্টার বাড়িয়ে। উদ্দেশ্য হলো: "আমি তো একটি ভ্যালিড উইন্ডো পেয়েছি, দেখি এর চেয়েও ছোট কোনো উইন্ডো বানানো যায় কিনা যা এখনও শর্ত পূরণ করে!"

## স্টেপ-বাই-স্টেপ লজিক
ধরা যাক প্রবলেম: "Minimum Size Subarray Sum" (Find the minimal length of a subarray whose sum $\ge$ target).
Array = `[2, 3, 1, 2, 4, 3]`, Target = `7`.

1. **Grow:** 
   `right` বাড়াও। 
   `2` -> `2+3=5` -> `2+3+1=6` -> `2+3+1+2=8`.
2. **Valid! (8 $\ge$ 7):**
   আমরা একটি ভ্যালিড উইন্ডো পেলাম `[2, 3, 1, 2]`, যার length = 4. `minLength = 4` সেভ করলাম।
3. **Shrink:**
   এখন `left` (যা 2 কে পয়েন্ট করে আছে) সরাব।
   `sum = 8 - 2 = 6`. 
   এখন `6 < 7`, তাই উইন্ডোটি আবার Invalid হয়ে গেলো।
4. **Grow again:**
   `right` বাড়িয়ে `4` নিলাম। `sum = 6 + 4 = 10`.
5. **Valid! (10 $\ge$ 7):**
   এখন উইন্ডো `[3, 1, 2, 4]`, length = 4.
6. **Shrink:**
   `left` সরিয়ে `3` বাদ দিলাম। `sum = 10 - 3 = 7`.
7. **Still Valid! (7 $\ge$ 7):**
   উইন্ডো `[1, 2, 4]`, length = 3. `minLength = 3` সেভ করলাম!
8. এভাবে চলতে থাকবে।

## Code Structure (While Loop for Shrinking)
এ ধরনের প্রবলেমে সব সময় একটি `while` লুপের ভেতরে আরেকটি `while` লুপ থাকে shrinking এর জন্য।
```java
int minLen = Integer.MAX_VALUE;
for (int right = 0; right < arr.length; right++) {
    // Add arr[right] to window state
    
    while (Condition Is Valid) {
        // Update minimum length
        minLen = Math.min(minLen, right - left + 1);
        
        // Remove arr[left] from window state
        left++; // Shrink window
    }
}
```
