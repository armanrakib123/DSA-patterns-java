# Fixed Sliding Window 🪟

Sliding Window প্যাটার্ন মূলত Two Pointer এর একটি স্পেশালাইজড রূপ, যা Array বা String এর "Subarray" বা "Substring" এর প্রবলেমগুলো সলভ করতে ব্যবহৃত হয়। 

Sliding Window মূলত দুই প্রকার:
1. **Fixed Window (নির্দিষ্ট সাইজের উইন্ডো)**
2. **Variable / Dynamic Window (পরিবর্তনশীল সাইজের উইন্ডো)**

এই ফাইলে আমরা **Fixed Window** নিয়ে আলোচনা করব।

## Fixed Window কী?
যখন প্রবলেমে সরাসরি বলে দেওয়া থাকে যে, আপনাকে ঠিক `K` সাইজের একটি সাব-অ্যারে (Subarray) নিয়ে কাজ করতে হবে, তখন সেটি Fixed Sliding Window এর প্রবলেম।

- *উদাহরণ:* "Find the maximum sum of a subarray of size `K`."

## লজিক ও মুভমেন্ট
ধরি, Array = `[2, 1, 5, 1, 3, 2]` এবং `K = 3`। আমাদের বের করতে হবে ৩ সাইজের কোন সাব-অ্যারের যোগফল সবচেয়ে বেশি।

**Brute Force:**
আপনি প্রতিটি এলিমেন্ট থেকে শুরু করে ৩ সাইজের সাব-অ্যারে বানাতে পারেন।
- `[2, 1, 5]` = 8
- `[1, 5, 1]` = 7
- `[5, 1, 3]` = 9 (Max!)
- এটি করতে $O(N \times K)$ সময় লাগবে। কিন্তু আমরা চাইলে $O(N)$ এ করতে পারি।

**Sliding Window Approach:**
এখানে উইন্ডোর সাইজ ফিক্সড (`K = 3`)।
১. প্রথমে আমরা `0` থেকে `K-1` পর্যন্ত যোগ করে একটি প্রাথমিক উইন্ডো (Initial Window) বানাব। `sum = 2 + 1 + 5 = 8`।
২. এরপর আমরা উইন্ডোটিকে এক ঘর করে ডান দিকে সরাব (Slide করব)।
৩. স্লাইড করার নিয়ম হলো:
   - নতুন যে এলিমেন্টটি উইন্ডোতে ঢুকছে, তাকে যোগ করব।
   - উইন্ডোর একদম বামের যে এলিমেন্টটি বের হয়ে যাচ্ছে, তাকে বিয়োগ করব।
   - **New Sum = (Old Sum) + (New Element) - (Outgoing Element)**
৪. নতুন উইন্ডোর ক্ষেত্রে:
   - `sum = 8 + 1 (নতুন এলিমেন্ট) - 2 (বের হয়ে যাওয়া এলিমেন্ট) = 7`।
   - এভাবে পুরো Array একবার ট্রাভার্স করলেই উত্তর পেয়ে যাব।

## সুবিধা
- টাইম কমপ্লেক্সিটি $O(N \times K)$ থেকে কমে $O(N)$ হয়ে যায়, কারণ আমরা বারবার একই এলিমেন্টগুলো যোগ করছি না।
- এটি $O(1)$ এক্সট্রা স্পেস নেয়।
