# 🐢🐇 Pattern 02: Fast & Slow Pointers (Tortoise and Hare Algorithm)

ফাস্ট অ্যান্ড স্লো পয়েন্টার (যাকে Floyd's Cycle Detection Algorithm বা কচ্ছপ এবং খরগোশ অ্যালগরিদমও বলা হয়) হলো টু-পয়েন্টার প্যাটার্নের একটি বিশেষ রূপ। এই প্যাটার্নটি মূলত লিংকড লিস্ট (Linked List) এবং নির্দিষ্ট কিছু সাইক্লিক (Cyclic) অ্যারে প্রবলেমে জাদুর মতো কাজ করে।

---

## 🧠 ইন্টুইশন (Intuition)

ধরে নিন একটি রেসট্র্যাকে (Track) একটি কচ্ছপ (Slow pointer) এবং একটি খরগোশ (Fast pointer) দৌড়াচ্ছে। 
- কচ্ছপ প্রতি ধাপে ১টি করে ঘর এগোয়।
- খরগোশ প্রতি ধাপে ২টি করে ঘর এগোয়।

যদি ট্র্যাকটি সোজা (Linear) হয়, তবে খরগোশ খুব দ্রুত ফিনিশ লাইনে (Null) পৌঁছে যাবে। 
কিন্তু ট্র্যাকটি যদি গোলাকার (Cyclic বা Loop) হয়, তবে দৌড়াতে দৌড়াতে কোনো এক সময় খরগোশ পিছন দিক থেকে এসে কচ্ছপকে ধরে ফেলবে বা ওভারল্যাপ করবে। 

অ্যালগরিদমের ভাষায়: 
**যদি Slow এবং Fast পয়েন্টার কোনো একসময় একে অপরের সমান (slow == fast) হয়ে যায়, তার মানে ওই ডেটা স্ট্রাকচারে একটি সাইকেল বা লুপ (Cycle/Loop) আছে!**

---

## 🚀 কেন এটি এত পাওয়ারফুল?

- ১. **মেমোরি অপ্টিমাইজেশন:** লুপ বের করার জন্য ব্রুট ফোর্স অ্যাপ্রোচ হলো `HashSet` ব্যবহার করা এবং দেখা যে একই নোড দুবার ভিজিট হচ্ছে কিনা। এতে `O(N)` স্পেস লাগে। কিন্তু Fast/Slow পয়েন্টারে কোনো এক্সট্রা স্পেস লাগে না, তাই Space Complexity হয় `O(1)`।
- ২. **মিডল এলিমেন্ট বের করা:** ফাস্ট পয়েন্টার যখন শেষ মাথায় পৌঁছায়, স্লো পয়েন্টার ঠিক তখন মাঝামাঝি (Middle) পজিশনে থাকে (যেহেতু ফাস্ট দ্বিগুণ স্পিডে যায়)। লিংকড লিস্টের মাঝখানের নোড বের করার এটি বেস্ট উপায়।

---

## 🚨 কখন বুঝবেন Fast & Slow Pointer ব্যবহার করতে হবে?

- ১. প্রবলেমটি যদি **Linked List** এর হয় এবং "Cycle", "Loop", "Middle", "N-th node from end" বা "Palindrome" এর কথা বলা থাকে।
- ২. যদি এমন একটি অ্যারে দেওয়া থাকে যেখানে ভ্যালুগুলো অ্যারের ইনডেক্স হিসেবে কাজ করে (যেমন প্রবলেম: Find the Duplicate Number) এবং সেখানে `O(1)` স্পেসে সলভ করতে বলা হয়।
- ৩. যখন কিছু সিকোয়েন্স রিপিট হচ্ছে এবং বলা হয়েছে সেটি নির্দিষ্ট কোনো জায়গায় গিয়ে থামবে কিনা চেক করতে (যেমন Happy Number)।

## 📝 বেসিক কোড টেমপ্লেট (Cycle Detection)

```java
public boolean hasCycle(ListNode head) {
    if (head == null || head.next == null) return false;
    
    ListNode slow = head;       // কচ্ছপ
    ListNode fast = head;       // খরগোশ
    
    while (fast != null && fast.next != null) {
        slow = slow.next;       // ১ স্টেপ
        fast = fast.next.next;  // ২ স্টেপ
        
        if (slow == fast) {
            return true;        // সাইকেল পাওয়া গেছে!
        }
    }
    return false; // ফাস্ট পয়েন্টার নাল এ পৌঁছেছে, অর্থাৎ কোনো সাইকেল নেই
}
```

## 🔍 সাইকেলের শুরুর জায়গা বের করার ম্যাজিক (Floyd's Math)

শুধু সাইকেল আছে কিনা তা বের করাই যথেষ্ট নয়। অনেক সময় প্রশ্ন করা হয়, সাইকেলটি কোন নোড থেকে শুরু হয়েছে?
এটি বের করার একটি সুন্দর ম্যাথ আছে:
- ১. যখন slow এবং fast প্রথমবার মিলিত হবে, তখন **fast কে ওখানেই রাখুন, আর slow কে আবার লিস্টের শুরুতে (head) পাঠিয়ে দিন**।
- ২. এবার slow এবং fast দুজনকেই **১ স্পিডে (1 step at a time)** সামনের দিকে বাড়ান।
- ৩. তারা দুজন আবার যে নোডে গিয়ে মিলিত হবে, সেটাই সাইকেলের শুরুর নোড!
