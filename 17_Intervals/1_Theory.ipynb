# Intervals (ইন্টারভালস) - The Software Company Standard Theory ⏳

## ১. ভূমিকা (Detailed Introduction)
ইন্টারভালস (Intervals) সংক্রান্ত সমস্যাগুলো মূলত টাইম-বেজড (Time-based) বা রেঞ্জ-বেজড (Range-based) ডেটা ম্যানেজমেন্টের ওপর কাজ করে। ইন্টারভাল বলতে আমরা একটি শুরুর সময় (Start) এবং একটি শেষের সময়কে (End) বুঝি। Software Company ইন্টারভিউতে ক্যালেন্ডার শিডিউলিং, মিটিং ম্যানেজমেন্ট বা রিসোর্স অ্যালোকেশন সংক্রান্ত সমস্যাগুলো ইন্টারভাল প্যাটার্ন ব্যবহার করে সমাধান করা হয়।

সহজ কথায়, ইন্টারভাল হলো একটি "সময়কাল"। আপনি যদি জানতে চান আপনার দুটি অ্যাপয়েন্টমেন্টের মধ্যে ওভারল্যাপ আছে কি না, তবে আপনি ইন্টারভাল লজিক ব্যবহার করবেন।

---

## ২. কেন ইন্টারভালস? (The Sorting & Merging Power)
ইন্টারভাল প্রবলেমের সবচেয়ে বড় চ্যালেঞ্জ হলো ডেটাগুলো এলোমেলোভাবে থাকতে পারে। সর্টিং (Sorting) ব্যবহার করে আমরা এই ডেটাগুলোকে একটি লিনিয়ার অর্ডারে নিয়ে আসি, যা আমাদের খুব সহজে ডিসিশন নিতে সাহায্য করে।
- **ব্যবহার:** Merge Intervals, Insert Interval, Meeting Rooms, Interval List Intersections.

---

## ৩. ইন্টারভাল রিলেশনশিপ (The 6 Possible Relations)
দুটি ইন্টারভাল $[s1, e1]$ এবং $[s2, e2]$ এর মধ্যে মূলত ৬টি সম্পর্ক হতে পারে:
১. তারা একদম আলাদা (s2 > e1)।
২. তারা ওভারল্যাপ করছে (s2 <= e1)।
৩. একটি ইন্টারভাল অন্যটির ভেতরে (Nested)।
৪. তারা একই সময়ে শুরু হচ্ছে কিন্তু আলাদা সময়ে শেষ হচ্ছে।
৫. তারা একই সময়ে শেষ হচ্ছে কিন্তু আলাদা সময়ে শুরু হচ্ছে।
৬. তারা একদম একই (Identical)।

ইন্টারভিউতে এই ওভারল্যাপ কন্ডিশন (`start2 <= end1`) চেক করা সবচেয়ে গুরুত্বপূর্ণ।

## ৪. রিয়েল-ওয়ার্ল্ড Software Company উদাহরণ (Real-World Case Studies)

### ৪.১. গুগল ক্যালেন্ডার - বিজি স্লট ডিটেকশন
গুগল ক্যালেন্ডার যখন চেক করে আপনি কখন বিজি এবং কখন ফ্রি, তখন তারা ইন্টারভাল মার্জিং অ্যালগরিদম ব্যবহার করে। আপনার সব মিটিংকে তারা মার্জ করে বের করে টোটাল "Busy Blocks" কোনগুলো। এটি আপনাকে নতুন মিটিং শিডিউল করতে সাহায্য করে।

### ৪.২. আমাজন AWS - রিসোর্স শিডিউলিং
AWS যখন তাদের সার্ভারগুলো বিভিন্ন ক্লায়েন্টকে অ্যালোকেট করে, তারা চেক করে কার কতক্ষণ সার্ভার লাগবে। ইন্টারভাল লজিক ব্যবহার করে তারা নিশ্চিত করে যে একই সার্ভার একই সময়ে দুজন ক্লায়েন্টকে দেওয়া না হয় (Double Booking Prevention)।

### ৪.৩. অপারেটিং সিস্টেম - প্রসেস শিডিউলিং (Gantt Chart)
আপনার কম্পিউটারে যখন বিভিন্ন প্রসেস রান করে, তখন CPU একটি Gantt Chart মেইনটেইন করে। এটি মূলত ইন্টারভালের একটি সিরিজ যা নির্দেশ করে কোন সময়ে কোন প্রসেস CPU ব্যবহার করছে।

## ৫. ইন্টারভিউ চেকলিস্ট (Interview Checklist)
১. **Sort by Start Time:** ইন্টারভাল প্রবলেমের প্রথম ধাপই হলো স্টার্ট টাইম অনুযায়ী সর্ট করা।
২. **Handling Edge Cases:** যদি একটি ইন্টারভাল অন্যটির ঠিক পরেই শুরু হয় (`e1 == s2`), তবে কি তারা ওভারল্যাপ ধরবেন? এটি প্রবলেম স্টেটমেন্ট অনুযায়ী ঠিক করুন।
৩. **In-place vs New List:** আপনি কি অরিজিনাল লিস্ট মডিফাই করবেন নাকি নতুন লিস্ট তৈরি করবেন? (ইন্টারভিউয়ারকে এটি আগে জিজ্ঞেস করে নিন)।

# Intervals (ইন্টারভালস) 🗓️

## Interval কী?
Interval (ইন্টারভাল) বলতে সাধারণত একটি শুরু এবং একটি শেষ সময়কে (Start, End) বোঝায়। যেমন: একটি মিটিং শুরু হলো ২টায় এবং শেষ হলো ৪টায় `[2, 4]`। 

ইন্টারভাল প্রবলেমগুলোতে আপনাকে অনেকগুলো এমন পেয়ার (Pair) বা 2D Array দেওয়া থাকবে।

## কখন বুঝবেন যে Interval Pattern ব্যবহার করতে হবে?
১. **Overlapping:** প্রবলেমে ওভারল্যাপিং (Overlapping) বা সংঘর্ষের কথা বলা থাকলে। যেমন: "কতগুলো মিটিং ওভারল্যাপ করছে?", "ওভারল্যাপ করা মিটিংগুলোকে এক করে দাও (Merge)"।
২. **Scheduling:** কাজ বা মিটিং শিডিউলিং করার কথা থাকলে।
৩. **Insert/Remove:** একটি নতুন ইন্টারভাল মাঝখানে বসানো (Insert) বা বাদ দেওয়ার কথা থাকলে।

## ইন্টারভাল ওভারল্যাপ করার শর্ত (The Golden Rule)
দুটি ইন্টারভাল $A = [startA, endA]$ এবং $B = [startB, endB]$ ওভারল্যাপ (Overlap) করবে কি না, তা বোঝার সবচেয়ে সহজ উপায় হলো:
**তারা ওভারল্যাপ করবে যদি:** `max(startA, startB) < min(endA, endB)` হয়।

কিন্তু আমরা সাধারণত প্রবলেম সলভ করার সময় ইন্টারভালগুলোকে **Start Time অনুযায়ী সর্ট (Sort)** করে নিই। 
সর্ট করার পর: $startA \le startB$ হবে। 
তখন ওভারল্যাপ চেক করা একদম সহজ:
**যদি $startB \le endA$ হয়, তবে তারা ওভারল্যাপ করছে!**

## কমন অপারেশনস
১. **সর্টিং (Sorting):** ৯৯% ইন্টারভাল প্রবলেমের প্রথম ধাপ হলো Start Time অনুযায়ী সর্ট করা। (Greedy মডিউলে আমরা End Time অনুযায়ী সর্ট করেছিলাম, সেটি শুধু Activity Selection এর জন্য। বাকি সব ক্ষেত্রে Start Time)।
২. **মার্জিং (Merging):** যদি তারা ওভারল্যাপ করে, তবে নতুন ইন্টারভালটি হবে: `[startA, max(endA, endB)]`।

পরবর্তী ফাইলগুলোতে আমরা এর ক্লাসিক কিছু প্রবলেম এবং ইমপ্লিমেন্টেশন দেখব।
