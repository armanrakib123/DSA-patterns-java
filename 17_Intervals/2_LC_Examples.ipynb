# LeetCode Interval Examples - Software Company Master Class 🚀

এই ফাইলে আমরা ইন্টারভাল (Intervals) প্যাটার্নের সেরা কিছু লেটকোড প্রবলেম এবং সেগুলোর বিস্তারিত সমাধান দেখব।

## ১. Merge Intervals (LC 56) 🟡
**টাইপ:** Sorting + Merging

### প্রবলেম স্টেটমেন্ট:
কিছু ওভারল্যাপিং ইন্টারভাল দেওয়া আছে। তাদের মার্চ (Merge) করে একটি লিস্ট তৈরি করুন।

### সমাধান:
১. ইন্টারভালগুলো স্টার্ট টাইম (Start Time) অনুযায়ী সর্ট করুন।
২. প্রথম ইন্টারভালটি রেজাল্ট লিস্টে অ্যাড করুন।
৩. পরবর্তী ইন্টারভালগুলোর জন্য চেক করুন: যদি বর্তমান ইন্টারভালের স্টার্ট টাইম আগের ইন্টারভালের এন্ড টাইমের চেয়ে ছোট বা সমান হয়, তবে এন্ড টাইম আপডেট করুন: `prev.end = max(prev.end, curr.end)`।
৪. নাহলে নতুন ইন্টারভাল হিসেবে লিস্টে অ্যাড করুন।

## ২. Insert Interval (LC 57) 🟡
**টাইপ:** Linear Scan + Merging

### প্রবলেম স্টেটমেন্ট:
একটি অলরেডি সর্টেড এবং নন-ওভারল্যাপিং ইন্টারভাল লিস্টে একটি নতুন ইন্টারভাল ইনসার্ট করুন এবং প্রয়োজন হলে মার্চ করুন।

### সমাধান:
১. নতুন ইন্টারভালের আগে থাকা সব ছোট ইন্টারভাল লিস্টে অ্যাড করুন।
২. নতুন ইন্টারভালের সাথে ওভারল্যাপ করা সব ইন্টারভাল মার্চ করুন।
৩. শেষে বাকি বড় ইন্টারভালগুলো লিস্টে অ্যাড করুন।

## ৩. Meeting Rooms II (LC 253) 🟡
**টাইপ:** Sorting + Min-Heap

### প্রবলেম স্টেটমেন্ট:
মিটিংয়ের টাইম লিস্ট দেওয়া আছে। মিনিমাম কতগুলো কনফারেন্স রুম লাগবে সব মিটিং শেষ করতে?

### সমাধান:
১. মিটিংগুলো স্টার্ট টাইম অনুযায়ী সর্ট করুন।
২. একটি মিন-হিপ (Min-Heap) ব্যবহার করুন যা মিটিংয়ের এন্ড টাইম স্টোর করবে।
৩. যদি নতুন মিটিংয়ের স্টার্ট টাইম হিপের টপ (সবচেয়ে আগে শেষ হওয়া মিটিং) এর চেয়ে বেশি হয়, তবে একই রুম ব্যবহার করা সম্ভব। হিপ থেকে পপ করে নতুন এন্ড টাইম পুশ করুন।
৪. হিপের সাইজই হবে আপনার প্রয়োজনীয় রুম সংখ্যা।

## ৪. Interval List Intersections (LC 986) 🟡
**টাইপ:** Two Pointer + Intersection Logic

### প্রবলেম স্টেটমেন্ট:
দুটি সর্টেড ইন্টারভাল লিস্টের মধ্যে কমন বা ইন্টারসেকশন অংশগুলো খুঁজে বের করুন।

### সমাধান:
১. দুটি পয়েন্টার ব্যবহার করুন দুই লিস্টের জন্য।
২. ইন্টারসেকশন স্টার্ট হবে: `max(start1, start2)` এবং এন্ড হবে: `min(end1, end2)`।
৩. যদি `start <= end` হয়, তবে এটি একটি ভ্যালিড ইন্টারসেকশন।
৪. যার এন্ড টাইম ছোট, তার পয়েন্টার এক ঘর ডানে সরান।
