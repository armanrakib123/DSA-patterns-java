# Trie (ট্রাই / প্রিফিক্স ট্রি) - The FAANG Standard Theory 🌳🔤

## ১. ভূমিকা (Detailed Introduction)
ট্রাই (Trie), যা "Prefix Tree" নামেও পরিচিত, একটি বিশেষ ধরনের সার্চ ট্রি ডেটা স্ট্রাকচার যা মূলত স্ট্রিং (String) প্রসেস করার জন্য ব্যবহৃত হয়। সাধারণ HashMap এর তুলনায় ট্রাই অনেক বেশি কার্যকর যখন আমাদের অনেকগুলো স্ট্রিংয়ের মধ্যে কমন প্রিফিক্স (Prefix) খুঁজে বের করতে হয়। FAANG ইন্টারভিউতে অটো-কমপ্লিট (Auto-complete), স্পেল চেকার (Spell Checker) বা ডিকশনারি সংক্রান্ত সমস্যার এটিই সবচেয়ে সেরা সমাধান।

সহজ কথায়, ট্রাই হলো একটি "অক্ষরের গাছ"। প্রতিটি নোড একটি ক্যারেক্টার রিপ্রেজেন্ট করে এবং রুট থেকে একটি নোড পর্যন্ত পথ একটি স্ট্রিং বা প্রিফিক্স তৈরি করে।

---

## ২. কেন ট্রাই? (The Power of Constant Time Prefix Search)
যদি আমাদের কাছে একটি বিশাল ডিকশনারি থাকে এবং আমরা জানতে চাই "apple" শব্দটি সেখানে আছে কি না:
- **HashMap:** স্ট্রিংটি হ্যাশ করতে হবে এবং খুঁজতে হবে। এটি মোটামুটি ফাস্ট হলেও এতে প্রিফিক্স সার্চ করা কঠিন।
- **Trie:** স্ট্রিংয়ের লেন্থ $L$ হলে, মাত্র $O(L)$ টাইমে শব্দটি খুঁজে পাওয়া সম্ভব, ডিকশনারিতে কতগুলো শব্দ আছে তার ওপর এটি নির্ভর করে না।

ট্রাই আমাদের মেমরি সাশ্রয় করে কারণ কমন প্রিফিক্সগুলো (যেমন: "apple" এবং "apply") একই নোড শেয়ার করে।

## ৩. ট্রাই নোডের গঠন (Structure)
একটি ট্রাই নোডে সাধারণত দুটি জিনিস থাকে:
১. **Children Array/Map:** পরবর্তী ক্যারেক্টারগুলোর লিংক। সাধারণত `TrieNode[26]` (যদি শুধু ছোট হাতের অক্ষর থাকে)।
২. **isEndOfWord:** একটি বুলিয়ান ভেরিয়েবল যা নির্দেশ করে এখানে একটি শব্দ শেষ হয়েছে কি না।

## ৪. রিয়েল-ওয়ার্ল্ড FAANG উদাহরণ (Real-World Case Studies)

### ৪.১. গুগল সার্চ - অটো-সাজেশন (Auto-complete)
গুগল সার্চ বক্সে যখন আপনি কিছু লিখতে শুরু করেন, গুগল আপনাকে সাথে সাথে কিছু শব্দ সাজেস্ট করে। এটি ব্যাক-এন্ডে একটি বিশাল ট্রাই ডেটা স্ট্রাকচার ব্যবহার করে। আপনি যা টাইপ করছেন তা ট্রাইয়ের একটি প্রিফিক্স হিসেবে ধরা হয় এবং ওই নোড থেকে নিচের সব শব্দগুলো আপনাকে সাজেশন হিসেবে দেখানো হয়।

### ৪.২. আমাজন - কি-ওয়ার্ড ইন্ডেক্সিং
আমাজনের প্রোডাক্ট সার্চ সিস্টেমে কোটি কোটি কি-ওয়ার্ড ইনডেক্স করার জন্য ট্রাই ব্যবহার করা হয়। এটি নিশ্চিত করে যে কাস্টমাররা যখন সার্চ করছে, তখন তারা যেন মিলি-সেকেন্ডের মধ্যে রেজাল্ট পায়।

### ৪.৩. অ্যাপল - অটো-কারেক্ট এবং প্রেডিক্টিভ টেক্সট
আপনার আইফোন (iPhone) বা ম্যাকবুক (MacBook) যখন আপনি টাইপ করার সময় ভুল শব্দ ঠিক করে দেয় বা পরের শব্দটি প্রেডিক্ট করে, তখন সেখানে ট্রাইয়ের সাথে প্রোবাবিলিটি মডেল ব্যবহার করা হয়। এটি ক্যারেক্টার বাই ক্যারেক্টার ট্রাই ট্রাভার্স করে সম্ভাব্য সঠিক শব্দটি খুঁজে বের করে।

## ৫. ইন্টারভিউ চেকলিস্ট (Interview Checklist)

২. **Deletion Logic:** একটি শব্দ ডিলিট করার সময় খেয়াল রাখতে হবে যে ওই নোডগুলো অন্য কোনো শব্দের অংশ কি না।
৩. **Prefix vs Word:** `search()` এবং `startsWith()` ফাংশনের পার্থক্য ক্লিয়ারলি বুঝুন।

# Trie (ট্রাই বা প্রিফিক্স ট্রি) 🌳

## Trie কী?
Trie (উচ্চারণ "Try" বা "Tree") হলো একটি বিশেষ ধরনের ট্রি ডেটা স্ট্রাকচার, যা মূলত **স্ট্রিং (String) বা ক্যারেক্টার** স্টোর (Store) করতে এবং খুব দ্রুত (Fast) সার্চ করতে ব্যবহৃত হয়। একে অনেক সময় **Prefix Tree** ও বলা হয়।

## কেন Trie ব্যবহার করব? (Hash Map বনাম Trie)
ধরা যাক আমাদের কাছে একটি ডিকশনারি (Dictionary) আছে, যেখানে অনেক শব্দ আছে। আমরা একটি শব্দ খুঁজতে চাই।
- **Hash Map/Set:** শব্দ খুঁজতে $O(1)$ সময় লাগলেও, যদি আমরা জানতে চাই "এমন কোনো শব্দ আছে কি না যা `cat` দিয়ে শুরু হয়েছে (Prefix)?", তবে Hash Map এটি সহজে করতে পারে না।
- **Trie:** Trie তে `startsWith("cat")` অপারেশনটি খুব ফাস্ট ($O(L)$, যেখানে $L$ হলো প্রিফিক্সের সাইজ) করা যায়!

*বাস্তব উদাহরণ:* গুগল সার্চে যখন আপনি কিছু টাইপ করেন (যেমন: "how to tie..."), তখন নিচে যে অটো-সাজেশন (Auto-complete) গুলো আসে ("how to tie a tie", "how to tie shoes"), সেগুলো Trie এর মাধ্যমেই কাজ করে!

## Trie কীভাবে কাজ করে?
- Trie এর প্রতিটি নোডে (Node) একটি ক্যারেক্টার থাকে না, বরং নোডের সাথে সংযুক্ত **এজ (Edge / Link)** গুলো ক্যারেক্টার রিপ্রেজেন্ট করে।
- প্রতিটি নোডে সাধারণত একটি Array (`TrieNode[] children = new TrieNode[26]`) থাকে ইংরেজি a-z ক্যারেক্টারগুলো ট্র্যাক করার জন্য। (অথবা HashMap ব্যবহার করা যায়)।
- প্রতিটি নোডে একটি `boolean isEndOfWord` ভেরিয়েবল থাকে, যা বোঝায় যে এখানে কোনো শব্দ শেষ হয়েছে কি না।

**উদাহরণ:** "cat" ইনসার্ট (Insert) করা:
Root $\rightarrow$ 'c' $\rightarrow$ 'a' $\rightarrow$ 't' (isEndOfWord = true)

## কখন বুঝবেন যে Trie ব্যবহার করতে হবে? (Pattern Recognition)
১. **Prefix Search:** "Prefix", "Starts with", "Auto-complete" বা ডিকশনারি জাতীয় প্রবলেম।
২. **Word Games:** Boggle, Word Search II (2D Grid এর সাথে Trie)।
৩. **XOR / Bitwise Trie:** নাম্বারগুলোর ম্যাক্সিমাম XOR বের করতে বললে (এটি একটু অ্যাডভান্সড)।

পরবর্তী ফাইলগুলোতে আমরা TrieNode এবং Trie এর ইমপ্লিমেন্টেশন দেখব।
