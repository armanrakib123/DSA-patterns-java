# Pointer Movement Logic 🚦

Two Pointer প্যাটার্নে সবচেয়ে ক্রিটিক্যাল অংশ হলো "কখন কোন পয়েন্টারটি মুভ (Move) করব?" 
আপনি যদি পয়েন্টার মুভমেন্টের লজিক ঠিকমতো না বুঝতে পারেন, তবে কোড অসীম লুপে (Infinite Loop) আটকে যেতে পারে অথবা ভুল উত্তর দিতে পারে।

এখানে আমরা Opposite Direction Two Pointer এর মুভমেন্ট লজিক নিয়ে বিস্তারিত আলোচনা করব।

## The Core Concept (Sorted Array এর ক্ষেত্রে)

ধরুন আমাদের কাছে একটি সর্টেড Array আছে: `[2, 5, 7, 11, 15]` এবং আমাদের এমন দুটি সংখ্যা খুঁজতে হবে যাদের যোগফল `Target = 18`।

আমরা দুটি পয়েন্টার নেব:
- `left` = 0 (প্রথম এলিমেন্ট, 2)
- `right` = 4 (শেষ এলিমেন্ট, 15)

### লজিক:
যেহেতু Array টি সর্টেড, তাই:
- বাম দিকের সংখ্যাগুলো ছোট।
- ডান দিকের সংখ্যাগুলো বড়।

এখন, বর্তমান যোগফল `current_sum` = `arr[left] + arr[right]`

1. **যদি `current_sum == target` হয়:**
   - বুম! আমরা আমাদের উত্তর পেয়ে গেছি।
2. **যদি `current_sum < target` হয়:**
   - এর মানে হলো আমাদের বর্তমান যোগফল টার্গেটের চেয়ে **ছোট**।
   - যোগফল বাড়াতে হলে আমাদের কী করতে হবে? আমাদের অবশ্যই ছোট সংখ্যাটিকে বাদ দিয়ে বড় একটি সংখ্যা নিতে হবে।
   - তাই, আমরা `left` পয়েন্টারটিকে ডান দিকে সরাব (`left++`)। কারণ ডান দিকে বড় সংখ্যা আছে।
3. **যদি `current_sum > target` হয়:**
   - এর মানে হলো আমাদের বর্তমান যোগফল টার্গেটের চেয়ে **বড়**।
   - যোগফল কমাতে হলে আমাদের কী করতে হবে? আমাদের বড় সংখ্যাটিকে বাদ দিয়ে ছোট একটি সংখ্যা নিতে হবে।
   - তাই, আমরা `right` পয়েন্টারটিকে বাম দিকে সরাব (`right--`)। কারণ বাম দিকে ছোট সংখ্যা আছে।

### ভিজ্যুয়ালাইজেশন:

`Array: [2, 5, 7, 11, 15], Target = 18`

**Step 1:**
- `left` = 0 (value: 2)
- `right` = 4 (value: 15)
- `sum` = 2 + 15 = 17. 
- 17 < 18, তাই আমাদের যোগফল বাড়াতে হবে। `left++` করব।

**Step 2:**
- `left` = 1 (value: 5)
- `right` = 4 (value: 15)
- `sum` = 5 + 15 = 20.
- 20 > 18, তাই আমাদের যোগফল কমাতে হবে। `right--` করব।

**Step 3:**
- `left` = 1 (value: 5)
- `right` = 3 (value: 11)
- `sum` = 5 + 11 = 16.
- 16 < 18, তাই `left++` করব।

**Step 4:**
- `left` = 2 (value: 7)
- `right` = 3 (value: 11)
- `sum` = 7 + 11 = 18.
- 18 == 18. Target Matched!

## Same Direction Pointer (Fast and Slow) Movement

In-place মডিফিকেশনের ক্ষেত্রে (যেমন: Remove Duplicates):
- `slow` পয়েন্টারটি ট্র্যাক রাখে কোথায় নতুন (বা ভ্যালিড) এলিমেন্ট বসাতে হবে।
- `fast` পয়েন্টারটি সামনের দিকে এগিয়ে গিয়ে চেক করে এলিমেন্টটি ভ্যালিড কিনা।
- যদি `fast` ভ্যালিড এলিমেন্ট পায়, তবে সেটিকে `slow` এর পজিশনে কপি করে দেয় এবং `slow++` করে।
- `fast` সবসময় প্রতি ধাপে এক ঘর করে এগোয় (`fast++`)।
