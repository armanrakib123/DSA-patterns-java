## Two Pointer কেন এত গুরুত্বপূর্ণ?
Software Company ইন্টারভিউতে Two Pointer থেকে প্রায়ই প্রশ্ন আসে কারণ এটি একটি সাধারণ $O(N^2)$ (ব্রুট-ফোর্স) সলিউশনকে সহজেই $O(N)$ বা $O(N \log N)$ টাইমে নামিয়ে আনতে পারে। 

ধরা যাক, আপনাকে একটি Array দেওয়া হলো এবং বলা হলো এমন দুটি সংখ্যা খুঁজে বের করতে যাদের যোগফল একটি নির্দিষ্ট Target এর সমান (Two Sum Problem)। 
- **Brute Force:** আপনি দুটি নেস্টেড লুপ চালাতে পারেন। প্রথম লুপ `i` এর জন্য এবং দ্বিতীয় লুপ `j` এর জন্য। এটি $O(N^2)$ সময় নেবে।
- **Two Pointer:** আপনি যদি Array টিকে সর্ট করে নেন, তবে একটি পয়েন্টার শুরুতে (`left`) এবং একটি শেষে (`right`) রেখে মাত্র একবার ট্রাভার্স করেই $O(N)$ টাইমে (সর্টিং বাদে) উত্তর বের করতে পারবেন।

## Two Pointer এর মূল ধরন (Types)

Two Pointer মূলত দুটি উপায়ে ব্যবহার করা হয়:

1. **Opposite Direction (বিপরীত দিক থেকে):** 
   - একটি পয়েন্টার থাকে Array এর একদম শুরুতে (Index 0)।
   - অন্য পয়েন্টারটি থাকে Array এর একদম শেষে (Index N-1)।
   - তারা ধীরে ধীরে একে অপরের দিকে এগোতে থাকে যতক্ষণ না তারা মিলিত হয় (`left < right`)।
   - *উদাহরণ:* Array Reverse করা, Palindrome চেক করা, Two Sum in Sorted Array।

2. **Same Direction (একই দিকে) / Fast & Slow Pointer:**
   - দুটি পয়েন্টারই শুরু থেকে (বা একই দিক থেকে) যাত্রা শুরু করে।
   - একটি পয়েন্টার দ্রুত আগায় (Fast), অন্যটি ধীরে আগায় (Slow)।
   - এটি Linked List এর Cycle (লুপ) বের করার জন্য খুব জনপ্রিয়।
   - *উদাহরণ:* Move Zeroes, Remove Duplicates, Linked List Cycle।
