# Edge Cases to Watch Out For ⚠️

যেকোনো অ্যালগরিদম ইমপ্লিমেন্ট করার সময় Edge Cases (অস্বাভাবিক ইনপুট) হ্যান্ডেল করা একজন প্রো-লেভেলের সফটওয়্যার ইঞ্জিনিয়ারের অন্যতম প্রধান কাজ। Two Pointer প্যাটার্নেও বেশ কিছু কমন Edge Cases আছে যা ইন্টারভিউতে আপনার কোড ব্রেক করে দিতে পারে।

কোড সাবমিট করার আগে অবশ্যই নিচের কেসগুলো আপনার কোডের মাধ্যমে মেন্টালি রান (Dry run) করিয়ে দেখবেন।

## ১. Empty Array বা String (খালি ইনপুট)
ইনপুট যদি `[]` বা `""` (Empty String) হয়, তাহলে আপনার কোড কি ক্র্যাশ করবে?
- `int left = 0, right = arr.length - 1;` 
- যদি `arr.length` 0 হয়, তবে `right` এর ভ্যালু হবে `-1`। এরপর লুপের ভেতরে `arr[-1]` অ্যাক্সেস করতে গেলে `ArrayIndexOutOfBoundsException` খাবেন।
- **সমাধান:** কোডের শুরুতেই চেক করে নিন।
  ```java
  if (arr == null || arr.length == 0) return 0; // or return arr
  ```

## ২. Array with Only 1 Element (মাত্র একটি এলিমেন্ট)
ইনপুটে যদি মাত্র একটি এলিমেন্ট থাকে, তবে `left` এবং `right` উভয়ই `0` তম ইনডেক্সে পয়েন্ট করবে।
- প্রবলেমের ওপর নির্ভর করে আপনার লুপের কন্ডিশন `while (left < right)` নাকি `while (left <= right)` হবে তা সতর্কতার সাথে বেছে নিতে হবে।
- যদি আপনার জোড়া (Pair) লাগে, তবে ১টি এলিমেন্ট দিয়ে জোড়া হবে না। তখন `left < right` ব্যবহার করতে হবে।

## ৩. All Duplicate Elements (সবগুলো এলিমেন্ট একই)
ইনপুট যদি `[2, 2, 2, 2, 2]` হয় এবং আপনাকে Unique Triplet (3Sum) বের করতে বলা হয়, তখন আপনার কোড কি বারবার একই রেজাল্ট রিটার্ন করবে?
- **সমাধান:** ডুপ্লিকেট স্কিপ করার লজিক লিখতে হবে।
  ```java
  while (left < right && arr[left] == arr[left + 1]) left++;
  while (left < right && arr[right] == arr[right - 1]) right--;
  ```

## ৪. Overlapping Pointers (পয়েন্টার একে অপরকে ক্রস করা)
লুপের ভেতরে যদি আপনি `left++` বা `right--` একাধিকবার করেন (যেমন ডুপ্লিকেট স্কিপ করার সময়), তবে `left` কি `right` কে ক্রস করে চলে যাচ্ছে?
- সব সময় ভেতরের লুপেও `left < right` কন্ডিশনটি চেক করতে হবে।
- **ভুল কোড:** `while(arr[left] == arr[left+1]) left++;` (এখানে left বাড়তে বাড়তে Array এর বাইরে চলে যেতে পারে)।
- **সঠিক কোড:** `while(left < right && arr[left] == arr[left+1]) left++;`

## ৫. Integer Overflow (পূর্ণসংখ্যার সীমা অতিক্রম)
Two Pointer দিয়ে যদি আপনি যোগফল (Sum) বা গুণফল (Product) বের করেন এবং ইনপুটের ভ্যালুগুলো অনেক বড় হয় (যেমন $10^9$), তবে দুটি Integer এর যোগফল Integer এর সর্বোচ্চ সীমা ($2^{31}-1$) অতিক্রম করে Overflow হতে পারে।
- **সমাধান:** যোগফল রাখার জন্য `long` ডাটা টাইপ ব্যবহার করুন।
  ```java
  long sum = (long) arr[left] + arr[right];
  ```

ইন্টারভিউয়াররা ইচ্ছে করেই এমন ইনপুট দেয় যেন আপনি Integer Overflow এর ফাঁদে পড়েন। তাই Constraints দেখে ডাটা টাইপ নির্বাচন করা খুবই জরুরি।


## Time Complexity (টাইম কমপ্লেক্সিটি)

Two Pointer প্যাটার্নের সবচেয়ে বড় সুবিধা হলো এটি টাইম কমপ্লেক্সিটি অনেক কমিয়ে আনে।

### 1. $O(N)$ Time Complexity
অধিকাংশ Two Pointer প্রবলেমের টাইম কমপ্লেক্সিটি $O(N)$ হয়। 
- **কেন?** কারণ `left` এবং `right` পয়েন্টারগুলো পুরো Array এর ওপর দিয়ে মাত্র একবার ট্রাভার্স করে। হয় `left` বাড়ে, নাহয় `right` কমে। প্রতিটি এলিমেন্ট বড়জোর একবার ভিজিট করা হয়। 
- **উদাহরণ:** Valid Palindrome, Two Sum II, Container With Most Water।

### 2. $O(N \log N)$ Time Complexity
কিছু প্রবলেমে Two Pointer অ্যাপ্লাই করার আগে Array টিকে সর্ট (Sort) করে নিতে হয়।
- **কেন?** Two Pointer এর (Opposite direction) মুভমেন্ট লজিক তখনই কাজ করে যখন ডাটা সর্টেড থাকে। 
- সর্ট করতে সময় লাগে $O(N \log N)$। এরপর Two Pointer চালাতে সময় লাগে $O(N)$। 
- মোট সময়: $O(N \log N) + O(N) = O(N \log N)$। (Big-O নোটিশনে ছোট টার্মটি বাদ দেওয়া হয়)।
- **উদাহরণ:** 3Sum, 4Sum। (3Sum এর ব্রুট ফোর্স $O(N^3)$, কিন্তু Two Pointer দিয়ে $O(N^2)$ এ নামিয়ে আনা যায়)।

## Space Complexity (স্পেস কমপ্লেক্সিটি)

Two Pointer এর আরেকটি বড় সুবিধা হলো এর মেমোরি এফিশিয়েন্সি।

### 1. $O(1)$ Space Complexity (Constant Space)
Two Pointer অ্যালগরিদমে আমরা সাধারণত শুধু দুটি ইন্টিজার ভেরিয়েবল (`left`, `right`) ব্যবহার করি। ইনপুটের সাইজ যতই বড় হোক না কেন, আমাদের অতিরিক্ত কোনো মেমোরি লাগে না।
- তাই প্রায় সব Two Pointer প্রবলেমের Auxiliary Space Complexity হলো $O(1)$।
- ইন্টারভিউয়াররা ইন-প্লেস (In-place) মডিফিকেশন চাইলে এই $O(1)$ স্পেসই তাদের মূল লক্ষ্য থাকে।

### 2. $O(N)$ Space Complexity (Rare cases)
কখনো কখনো আউটপুট রিটার্ন করার জন্য নতুন একটি Array বা List তৈরি করতে হয়।
- **উদাহরণ:** Squares of a Sorted Array। এখানে আমাদের রেজাল্ট রাখার জন্য একই সাইজের একটি নতুন Array নিতে হয়, তাই Space Complexity $O(N)$ হয়। তবে লক্ষ্যণীয় যে, ইন্টারভিউতে আউটপুট স্টোর করার স্পেসকে অনেক সময় Auxiliary Space হিসেবে ধরা হয় না।
