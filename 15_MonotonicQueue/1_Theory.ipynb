# Monotonic Queue (মনোটোনিক কিউ) - The Software Company Standard Theory 📊

## ১. ভূমিকা (Detailed Introduction)
মনোটোনিক কিউ (Monotonic Queue) হলো কিউ ডেটা স্ট্রাকচারের একটি উন্নত ব্যবহার, যা সাধারণত স্লাইডিং উইন্ডো (Sliding Window) প্রবলেমে ব্যবহৃত হয়। এটি নিশ্চিত করে যে কিউয়ের ভেতরে ডেটাগুলো সব সময় একটি নির্দিষ্ট অর্ডারে (ক্রম) থাকবে। এর মূল বৈশিষ্ট্য হলো এটি আমাদের যেকোনো উইন্ডোর ম্যাক্সিমাম বা মিনিমাম ইলিমেন্ট মাত্র $O(1)$ টাইমে দিতে পারে। Software Company ইন্টারভিউতে "Sliding Window Maximum" সংক্রান্ত সমস্যার এটিই সবচেয়ে অপ্টিমাইজড সমাধান।

সহজ কথায়, মনোটোনিক কিউ হলো একটি "এলিট কিউ"। এখানে শুধুমাত্র তারাই থাকতে পারে যারা শক্তিশালী (বড় বা ছোট) এবং যাদের ভবিষ্যতে বিজয়ী হওয়ার সম্ভাবনা আছে।

---

## ২. কেন মনোটোনিক কিউ? (The O(1) Window Max/Min)
যদি আমাদের প্রতিটি $K$ সাইজের উইন্ডোর ম্যাক্সিমাম বের করতে বলা হয়:
- **ব্রুট ফোর্স:** প্রতিটি উইন্ডোর জন্য লুপ চালানো। টাইম: $O(N \times K)$।
- **প্রায়োরিটি কিউ (Heap):** হিপ ব্যবহার করা। টাইম: $O(N \log K)$।
- **মনোটোনিক কিউ:** ডি-কিউ (Deque) ব্যবহার করা। টাইম: $O(N)$।

এটি ডাটা স্ট্রিমিং এবং রিয়েল-টাইম মনিটরংয়ের জন্য অত্যন্ত কার্যকর।

---

## ৩. ডি-কিউ (Deque) এর ব্যবহার
মনোটোনিক কিউ ইমপ্লিমেন্ট করার জন্য আমরা `Deque` (Double Ended Queue) ব্যবহার করি। কারণ আমাদের কিউয়ের সামনে এবং পিছনে—উভয় দিকেই অপারেশন করতে হয়।
- **পছন থেকে (Back):** নতুন আসা ইলিমেন্টের চেয়ে ছোট/বড় ইলিমেন্টগুলো পপ করা হয়।
- **সামনে থেকে (Front):** উইন্ডোর বাইরে চলে যাওয়া ইলিমেন্টগুলো পপ করা হয়।

## ৪. রিয়েল-ওয়ার্ল্ড Software Company উদাহরণ (Real-World Case Studies)

### ৪.১. গুগল - সার্চ ট্রেন্ডস রিয়েল-টাইম মনিটর
গুগল যখন রিয়েল-টাইমে সার্চ ট্রেন্ড মনিটর করে যে গত ১ ঘণ্টায় সবচেয়ে বেশি কোন টপিক সার্চ হয়েছে, তখন তারা মনোটোনিক কিউয়ের মতো উইন্ডো বেজড অ্যালগরিদম ব্যবহার করে। এটি তাদের মিলি-সেকেন্ডের মধ্যে সর্বোচ্চ সার্চ ভলিউম আপডেট করতে সাহায্য করে।

### ৪.২. আমাজন - রিয়েল-টাইম ডাইনামিক প্রাইসিং
আমাজনের অ্যালগরিদম যখন গত কয়েক মিনিটের প্রাইস ফ্ল্যাকচুয়েশন এনালাইসিস করে প্রোডাক্টের দাম কমায় বা বাড়ায়, তখন তারা স্লাইডিং উইন্ডো ম্যাক্সিমাম/মিনিমাম ব্যবহার করে সবচেয়ে সঠিক ডিসিশন নেয়।

### ৪.৩. নেটওয়ার্কিং - ব্যান্ডউইথ শেপিং (Packet Jitter)
নেটওয়ার্ক প্যাকেটগুলো যখন স্ট্রিমিং সার্ভারে আসে, তখন গত কয়েক সেকেন্ডের ম্যাক্সিমাম ল্যাটেন্সি (Latency) ট্র্যাক করার জন্য মনোটোনিক কিউ ব্যবহার করা হয়। এটি ভিডিও বা অডিওর বাফারিং কমাতে সাহায্য করে।

# Monotonic Queue (মনোটোনিক কিউ) 🚂

## Monotonic Queue কী?
আমরা Monotonic Stack সম্পর্কে জেনেছি। **Monotonic Queue** হলো তার ভাই! এটি একটি Double Ended Queue বা **Deque** (ডেক), যার দুপাশ (First এবং Last) দিয়েই ইলিমেন্ট ঢোকানো এবং বের করা যায়।

Monotonic Queue তেও ইলিমেন্টগুলো একটি নির্দিষ্ট অর্ডারে (ছোট থেকে বড় অথবা বড় থেকে ছোট) সাজানো থাকে।

## Stack এর সাথে Queue এর পার্থক্য কী?
- **Stack:** শুধুমাত্র একপাশ (Top) দিয়ে কাজ করা যায়।
- **Queue (Deque):** Monotonic Queue এর সবচেয়ে বড় সুবিধা হলো, এটি **Sliding Window** এর সাথে দারুণভাবে কাজ করে! উইন্ডো যখন স্লাইড করে, তখন পুরোনো ইলিমেন্টগুলো Queue এর সামনের দিক (Front/First) থেকে খুব সহজেই বাদ দিয়ে দেওয়া যায়।

## কখন বুঝবেন যে Monotonic Queue ব্যবহার করতে হবে? (Pattern Recognition)
১. **Sliding Window + Maximum/Minimum:** প্রবলেমে যদি বলা হয় "প্রতিটি K সাইজের Sliding Window এর ম্যাক্সিমাম বা মিনিমাম ভ্যালু বের করো"।
২. **O(N) Requirement:** Sliding Window এর ভেতরে ম্যাক্স/মিন বের করতে $O(K)$ সময় লাগে (টোটাল $O(N \times K)$)। কিন্তু Monotonic Queue এটি $O(N)$ এ নামিয়ে আনে।

## কীভাবে কাজ করে? (Sliding Window Maximum)
ধরা যাক আমরা K সাইজের উইন্ডোর ম্যাক্সিমাম বের করতে চাই। আমাদের টার্গেট হলো: Deque এর একদম সামনে (Front) যেন সবসময় উইন্ডোর ম্যাক্সিমাম ভ্যালুটি থাকে।

**৩টি রুলস:**
১. **উইন্ডো থেকে বাদ দেওয়া (Remove from Front):** উইন্ডো যখন এক ঘর ডানে স্লাইড করবে, তখন যদি Deque এর সামনের ইলিমেন্টটি উইন্ডোর বাইরে চলে যায়, তবে তাকে `pollFirst()` করে বাদ দিন।
২. **অর্ডার মেইনটেইন করা (Remove from Back):** নতুন একটি ইলিমেন্ট Deque তে ঢোকানোর আগে, Deque এর পেছন (Back) থেকে এমন সব ইলিমেন্টকে `pollLast()` করে বাদ দিন, যারা নতুন ইলিমেন্টের চেয়ে ছোট! (কারণ ছোট ইলিমেন্টগুলো আর কোনোদিনই Maximum হতে পারবে না, যেহেতু তাদের চেয়ে বড় একজন চলে এসেছে)।
৩. **নতুন ইলিমেন্ট যোগ করা:** নতুন ইলিমেন্টটিকে Deque এর পেছনে `offerLast()` করুন।

পরবর্তী ফাইলে আমরা এর সবচেয়ে বিখ্যাত প্রবলেম (Sliding Window Maximum) এর ইমপ্লিমেন্টেশন দেখব।
