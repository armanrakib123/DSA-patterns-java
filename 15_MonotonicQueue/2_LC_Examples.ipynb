## ১. Sliding Window Maximum (LC 239) 🔴
**টাইপ:** Basic Monotonic Deque

### প্রবলেম স্টেটমেন্ট:
প্রতিটি $K$ সাইজের উইন্ডোর মধ্যে থাকা সর্বোচ্চ সংখ্যাটি বের করুন। এটি একটি অত্যন্ত পপুলার FAANG ইন্টারভিউ প্রবলেম।

### সমাধান:
১. একটি `Deque` ব্যবহার করুন যা ইনডেক্স স্টোর করবে।
২. নতুন ইলিমেন্ট আসার আগে ডি-কিউ এর শেষ থেকে ছোট ইলিমেন্টগুলো বের করে দিন। এটি ডি-কিউকে সবসময় ডিক্রিজিং অর্ডারে রাখবে।
৩. ডি-কিউ এর প্রথম ইলিমেন্টটিই হবে রেজাল্ট।
৪. যখন উইন্ডো স্লাইড হবে, তখন চেক করুন ডি-কিউ এর প্রথম ইলিমেন্টটি উইন্ডো থেকে বের হয়ে গেছে কি না।

## ২. Constrained Subsequence Sum (LC 1425) 🔴
**টাইপ:** DP + Monotonic Queue

### প্রবলেম স্টেটমেন্ট:
একটি সাব-সিকোয়েন্সের সর্বোচ্চ যোগফল বের করুন যেখানে পাশাপাশি দুটি ইলিমেন্টের ইনডেক্সের দূরত্ব $K$ এর বেশি হবে না।

### সমাধান:
১. এটি মূলত একটি DP প্রবলেম: `dp[i] = nums[i] + max(0, max_dp_in_window_k)`।
২. এই $K$ সাইজের উইন্ডোর মধ্যে ম্যাক্সিমাম DP ভ্যালু খুঁজে বের করার জন্য আমরা মনোটোনিক কিউ ব্যবহার করি।
৩. এটি $O(N^2)$ এর প্রবলেমকে $O(N)$ এ নিয়ে আসে।

## ৩. Longest Continuous Subarray With Absolute Diff <= Limit (LC 1438) 🟡
**টাইপ:** Two Pointer + Two Monotonic Queues

### প্রবলেম স্টেটমেন্ট:
সবচেয়ে বড় সাব-অ্যারের দৈর্ঘ্য বের করুন যেখানে যেকোনো দুটি ইলিমেন্টের পার্থক্য একটি নির্দিষ্ট লিমিটের মধ্যে।

### সমাধান:
১. দুটি মনোটোনিক কিউ ব্যবহার করুন: একটি ম্যাক্সিমাম এবং একটি মিনিমাম ভ্যালু ট্র্যাক করার জন্য।
২. যখন `max_queue - min_queue > limit` হবে, তখন `left` পয়েন্টার ডানে সরান এবং কিউগুলো আপডেট করুন।

## ৪. Shortest Subarray with Sum at Least K (LC 862) 🔴
**টাইপ:** Prefix Sum + Monotonic Queue

### প্রবলেম স্টেটমেন্ট:
সবচেয়ে ছোট সাব-অ্যারের দৈর্ঘ্য বের করুন যার যোগফল কমপক্ষে $K$। (অ্যারেতে নেগেটিভ নাম্বার থাকতে পারে)।

### সমাধান:
১. প্রথমে প্রিফিক্স সাম (Prefix Sum) ক্যালকুলেট করুন।
২. একটি ইনক্রিজিং মনোটোনিক কিউ ব্যবহার করে প্রিফিক্স সামগুলো স্টোর করুন।
৩. যখনই কোনো প্রিফিক্স সাম থেকে কিউয়ের প্রথম ইলিমেন্টের পার্থক্য $\ge K$ হবে, তখন `minLength` আপডেট করুন এবং কিউ থেকে সরাুন।
