# 🌊 Pattern 24: Breadth-First Search (BFS) - গভীর বিশ্লেষণ

ব্রেডথ-ফার্স্ট সার্চ (BFS) হলো গ্রাফ এবং ট্রি ট্রাভার্সাল অ্যালগরিদমগুলোর মধ্যে অন্যতম একটি স্তম্ভ। যদি আমরা কল্পনা করি একটি পুকুরে পাথর ছুড়লে যেমন ঢেউ চারিদিকে ছড়িয়ে পড়ে, BFS ঠিক সেভাবেই কাজ করে। এটি সোর্স নোড থেকে শুরু করে ধাপে ধাপে (Level by Level) বাইরের দিকে অগ্রসর হয়।

---

## 🧠 BFS এর মূল দর্শন (Core Philosophy)

BFS এর মূল মন্ত্র হলো: **"আগে চারপাশ দেখো, তারপর গভীরে যাও।"** 
এটি শর্টেস্ট পাথ (Shortest Path) খুঁজে বের করার জন্য সবচেয়ে কার্যকর যদি গ্রাফের প্রতিটি এজের ওয়েট (Weight) সমান হয়।

### ১. লেভেল-অর্ডার ট্রাভার্সাল (Level-Order Traversal)
BFS এ আমরা প্রথমে লেভেল ০ (সোর্স) ভিজিট করি, তারপর লেভেল ১ (সোর্সের প্রতিবেশী), তারপর লেভেল ২ এবং এভাবে চলতে থাকে।

### ২. কিউ (Queue) ডেটা স্ট্রাকচার
BFS বাস্তবায়নের জন্য আমাদের একটি **First-In-First-Out (FIFO)** কিউ প্রয়োজন। এটি আমাদের মনে করিয়ে দেয় যে আমরা কোন নোডগুলো দেখেছি কিন্তু এখনো তাদের প্রতিবেশীদের এক্সপ্লোর করিনি।

---

## 🚀 BFS অ্যালগরিদমের ধাপসমূহ (Step-by-Step)

- ১. একটি খালি কিউ তৈরি করুন এবং সোর্স নোডটিকে এতে পুশ করুন।
- ২. সোর্স নোডটিকে "Visited" হিসেবে মার্ক করুন (যাতে আমরা লুপে না পড়ি)।
- ৩. যতক্ষণ কিউ খালি না হয়:
   - কিউ থেকে একটি নোড `current` বের করে আনুন।
   - `current` নোডের প্রতিটি প্রতিবেশীকে (Neighbor) চেক করুন।
   - যদি প্রতিবেশীটি আগে ভিজিট করা না হয়ে থাকে:
     - সেটিকে ভিজিট করুন এবং কিউতে পুশ করুন।

---

## 🛠️ কেন BFS ইন্টারভিউতে এত গুরুত্বপূর্ণ?

- ১. **Shortest Path in Unweighted Graphs:** এটিই এর সবচেয়ে বড় শক্তি। আপনি যদি গোলকধাঁধায় (Maze) আটকা পড়েন, তবে BFS আপনাকে দ্রুততম পথ বের করে দেবে।
- ২. **Layer-wise Analysis:** যখন আমাদের কোনো ডেটা লেয়ার বাই লেয়ার প্রসেস করতে হয় (যেমন- নেটওয়ার্ক হপিং)।
- ৩. **Connectivity:** একটি গ্রাফে দুটি নোড কানেক্টেড কিনা তা দ্রুত চেক করা যায়।

---

## 🚧 BFS বনাম DFS: কখন কোনটি ব্যবহার করবেন?

| বৈশিষ্ট্য | BFS | DFS |
| :--- | :--- | :--- |
| **ডেটা স্ট্রাকচার** | Queue | Stack / Recursion |
| **মেমোরি** | বেশি (লেভেলের সব নোড রাখতে হয়) | কম |
| **পাথ** | শর্টেস্ট পাথ খুঁজে পায় | ডেপথ বা গভীরে যায় |
| **ব্যবহার** | Level order, Shortest Path | Cycle detection, Topological sort |

---

## 📝 BFS এর টেমপ্লেট (Java)

```java
Queue<Node> queue = new LinkedList<>();
Set<Node> visited = new HashSet<>();

queue.add(source);
visited.add(source);

while (!queue.isEmpty()) {
    int levelSize = queue.size(); // বর্তমান লেভেলের নোড সংখ্যা
    for (int i = 0; i < levelSize; i++) {
        Node current = queue.poll();
        // আপনার লজিক এখানে
        for (Node neighbor : current.neighbors) {
            if (!visited.contains(neighbor)) {
                visited.add(neighbor);
                queue.add(neighbor);
            }
        }
    }
}
```

## 💡 ইন্টারভিউ টিপস: "Level Size" ট্রিক
যখন আপনাকে বলা হবে প্রতি লেভেলের গড় (Average) বা লেভেল আলাদাভাবে প্রিন্ট করতে, তখন কিউ থেকে এলিমেন্ট বের করার আগে `queue.size()` ভেরিয়েবলে সেভ করে নিন। এটিই আপনার বর্তমান লেভেলের সীমানা নির্ধারণ করবে।

# Breadth First Search (BFS) বা প্রস্থ-ভিত্তিক অনুসন্ধান 🌊

## BFS কী?
Breadth First Search (BFS) হলো ট্রি (Tree) বা গ্রাফ (Graph) ট্রাভার্স করার আরেকটি অ্যালগরিদম। 
DFS যেমন গভীরে যাওয়ার চেষ্টা করে, BFS ঠিক তার উল্টো। এটি **"Breadth" (প্রস্থ)** বা লেভেল (Level) কে অগ্রাধিকার দেয়। অর্থাৎ, এটি প্রথমে কাছাকাছি (current level) সব নোড ভিজিট করে, তারপর এক ধাপ দূরে (next level) যায়।

*উদাহরন:* পাথরে ঢিল ছুঁড়লে যেমন পানিতে চারদিকে সমানভাবে ঢেউ ছড়িয়ে পড়ে, BFS ঠিক সেভাবেই কাজ করে!

## BFS কীভাবে কাজ করে?
BFS মূলত **Queue (কিউ)** বা লাইন ডেটা স্ট্রাকচার ব্যবহার করে কাজ করে (First In, First Out)। 
১. প্রথমে শুরুর নোডটিকে Queue তে ঢোকানো হয়।
২. Queue থেকে একটি নোড বের করা হয়, এবং তার সব প্রতিবেশীকে (Neighbors/Children) Queue তে ঢোকানো হয়।
৩. গ্রাফের ক্ষেত্রে, Infinite Loop এড়াতে `visited` সেট ব্যবহার করা হয়।

## কখন বুঝবেন যে BFS ব্যবহার করতে হবে? (Pattern Recognition)
১. **Shortest Path (সবচেয়ে ছোট রাস্তা):** একটি আনওয়েটেড (Unweighted) গ্রাফে বা 2D গ্রিডে এক পয়েন্ট থেকে অন্য পয়েন্টে যাওয়ার সবচেয়ে ছোট রাস্তা বের করতে বললে (যেমন: Maze solver)।
২. **Level Order Traversal:** ট্রিতে যদি বলা হয় "প্রতিটি লেভেলের নোডগুলোকে একসাথে প্রিন্ট করো" বা "ডান দিক থেকে দেখলে কোন নোডগুলো দেখা যাবে?" (Right Side View)।
৩. **Spreading / Connected Components:** যখন কোনো কিছু চারদিকে ছড়িয়ে পড়ে। যেমন: জম্বি ইনফেকশন (Zombie infection), পচা কমলা (Rotting Oranges), অথবা পানিতে আগুন ছড়িয়ে পড়া।

## DFS বনাম BFS: কোনটি কখন ব্যবহার করবেন?
- **Shortest Path:** অবশ্যই BFS। DFS দিয়ে Shortest Path পাওয়া যায় না (পেলেও অনেক সময় লাগে)।
- **Memory/Space:** যদি ট্রি অনেক চওড়া (Wide) হয়, তবে BFS এ মেমরি বেশি লাগবে (Queue তে অনেক নোড থাকবে)। তখন DFS ভালো। আর যদি ট্রি অনেক গভীর (Deep) হয়, তবে DFS এ Call Stack Overflow হতে পারে, তখন BFS ভালো।
- **Connectivity:** দুটির যেকোনোটি দিয়েই করা যায়, তবে কোড লেখার সুবিধার জন্য বেশিরভাগ মানুষ DFS ব্যবহার করে।

## Multi-Source BFS (অ্যাডভান্সড)
সাধারণত BFS একটি পয়েন্ট থেকে শুরু হয়। কিন্তু যদি বলা হয় "একটি গ্রিডে একাধিক জায়গা থেকে একই সাথে আগুন ছড়াচ্ছে", তখন আমরা শুরুর দিকেই Queue তে সবগুলো আগুনের পয়েন্ট একসাথে ঢুকিয়ে দিই। একে বলা হয় Multi-Source BFS (যেমন: LC 994 Rotting Oranges)।

পরবর্তী ফাইলগুলোতে আমরা এর টেমপ্লেট এবং উদাহরণ দেখব।

# Breadth First Search (BFS) - The Software Company Standard Theory 🌊

## ১. ভূমিকা (Detailed Introduction)
ব্রেডথ ফার্স্ট সার্চ (Breadth First Search বা BFS) হলো গ্রাফ এবং ট্রি ট্রাভার্সাল করার একটি অত্যন্ত কার্যকর অ্যালগরিদম। DFS এর উল্টো, এটি আগে "প্রস্থে" (Breadth) যায় এবং তারপর "গভীরে" (Depth) ঢোকে। সহজ কথায়, এটি স্টার্টিং নোড থেকে লেয়ার বাই লেয়ার (Layer by Layer) সব নোড ভিজিট করে।

Software Company ইন্টারভিউতে BFS সবচেয়ে বেশি জনপ্রিয় "Shortest Path" বা "Minimum Steps" সংক্রান্ত সমস্যা সমাধানের জন্য। আপনি যদি জানতে চান আপনার সবচেয়ে কাছের রেস্টুরেন্ট কোনটি, তবে BFS হবে আপনার সেরা সলিউশন।

---

## ২. কেন BFS? (The Power of Shortest Path)
BFS এর সবচেয়ে বড় বৈশিষ্ট্য হলো, আন-ওয়েটেড (Unweighted) গ্রাফে এটি গ্যারান্টি দিয়ে সবচেয়ে ছোট রাস্তা (Shortest Path) খুঁজে বের করে।
- **ব্যবহার:** শর্টেস্ট পাথ ফাইন্ডিং (Shortest Path), লেভেল অর্ডার ট্রাভার্সাল (Level Order Traversal), সোশ্যাল নেটওয়ার্ক কানেক্টিভিটি (Degrees of Separation)।

এটি সাধারণত একটি কিউ (Queue) ডেটা স্ট্রাকচার ব্যবহার করে ইমপ্লিমেন্ট করা হয়, যা "FIFO" (First In, First Out) প্রিন্সিপাল মেনে চলে।

---

## ৩. BFS কিভাবে কাজ করে? (Algorithm Logic)
১. একটি স্টার্টিং নোড সিলেক্ট করুন এবং তাকে কিউ-তে পুশ (Push) করুন।
২. নোডটিকে 'Visited' হিসেবে মার্ক করুন।
৩. যতক্ষণ কিউ খালি না হচ্ছে:
   - কিউ থেকে একটি নোড পপ (Pop) করুন।
   - তার সব আন-ভিজিটেড নেইবারগুলোকে ভিজিট করুন এবং কিউ-তে পুশ করুন।
   - প্রতিবার কিউ থেকে পপ করার সময় আপনি এক একটি "লেয়ার" শেষ করছেন।

## ৪. রিয়েল-ওয়ার্ল্ড Software Company উদাহরণ (Real-World Case Studies)

### ৪.১. লিঙ্কডইন - ডিগ্রি অফ কানেকশন (1st, 2nd, 3rd Degree)
লিঙ্কডইন যখন দেখায় একজন ইউজার আপনার কততম ডিগ্রি কানেকশন, তখন তারা BFS ব্যবহার করে। আপনার সরাসরি ফ্রেন্ডরা হলো ১ম লেয়ার (1st degree), তাদের ফ্রেন্ডরা ২য় লেয়ার (2nd degree), এবং এভাবেই চলতে থাকে। BFS নিশ্চিত করে যে আপনি সবচেয়ে ছোট দূরত্বের ডিগ্রিটিই দেখতে পাচ্ছেন।

### ৪.২. নেটফ্লিক্স এবং আমাজন - রেকমেন্ডেশন ইঞ্জিন
রেকমেন্ডেশন ইঞ্জিনগুলো অনেক সময় গ্রাফ ব্যবহার করে। যদি দুইজন ইউজার একই ধরনের মুভি পছন্দ করে, তবে তাদের মধ্যে একটি লিংক তৈরি হয়। BFS ব্যবহার করে সিস্টেম খুঁজে বের করে আপনার কত কাছাকাছি অন্য ইউজাররা আছে এবং তাদের পছন্দ অনুযায়ী আপনাকে মুভি রেকমেন্ড করে।

### ৪.৩. উবার এবং গুগল ম্যাপস - নেভিগেশন (Dijkstra's Base)
যদিও গুগল ম্যাপস Dijkstra বা A* ব্যবহার করে, কিন্তু তার মূল ভিত্তি হলো BFS। সিটি ম্যাপকে একটি গ্রাফ হিসেবে চিন্তা করলে, BFS ব্যবহার করে এক পয়েন্ট থেকে অন্য পয়েন্টে যাওয়ার সবচেয়ে কম ট্রাফিক বা সবচেয়ে কম দূরত্বের রাস্তা খুঁজে বের করা হয়।

## ৫. মাল্টি-সোর্স BFS (Multi-Source BFS) - Advanced
এটি একটি অত্যন্ত শক্তিশালী কনসেপ্ট। যখন আপনার গ্রাফে একাধিক স্টার্টিং পয়েন্ট থাকে এবং আপনি জানতে চান প্রতিটি নোড থেকে সবচেয়ে কাছের স্টার্টিং পয়েন্ট কোনটি, তখন আমরা সব স্টার্টিং পয়েন্টকে একসাথে কিউ-তে পুশ করে BFS শুরু করি। (যেমন: LC 994: Rotting Oranges)।

## ৬. ইন্টারভিউ চেকলিস্ট (Interview Checklist)
১. **Queue Implementation:** জাভায় সাধারণত `LinkedList` বা `ArrayDeque` ব্যবহার করা হয় কিউ হিসেবে।
২. **Level Tracking:** যদি আপনাকে লেভেলের হিসাব রাখতে হয়, তবে প্রতিবার কিউ-এর সাইজ (`queue.size()`) নিয়ে একটি লুপ চালান।
৩. **Visited Check:** নেইবারকে কিউ-তে পুশ করার ঠিক আগেই 'Visited' মার্ক করে দিন যাতে একই নোড বারবার কিউ-তে না ঢোকে।

## ৭. উপসংহার (Conclusion)
BFS হলো অপ্টিমাইজড সার্চিংয়ের প্রাণ। এটি আপনার কোডকে অনেক বেশি প্রেডিক্টেবল এবং ফাস্ট করে তোলে যখন আপনি শর্টেস্ট পাথ নিয়ে কাজ করেন। পরবর্তী ফাইলে আমরা এর বিভিন্ন টেমপ্লেট এবং লেটকোড উদাহরণ দেখব।
