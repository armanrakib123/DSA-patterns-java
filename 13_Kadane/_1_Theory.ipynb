# Kadane's Algorithm (ক্যাডেন্স অ্যালগরিদম) - The FAANG Standard Theory ⚡

## ১. ভূমিকা (Detailed Introduction)
ক্যাডেন্স অ্যালগরিদম (Kadane's Algorithm) হলো সাব-অ্যারে (Subarray) সংক্রান্ত সমস্যা সমাধানের একটি মাস্টারপিস। এটি মূলত একটি অ্যারে থেকে এমন একটি সাব-অ্যারে খুঁজে বের করে যার যোগফল সর্বোচ্চ (Maximum Sum Subarray)। FAANG ইন্টারভিউতে এটি অত্যন্ত জনপ্রিয় কারণ এটি একটি $O(N^2)$ বা $O(N^3)$ এর প্রবলেমকে মাত্র $O(N)$ টাইম এবং $O(1)$ স্পেস কমপ্লেক্সিটিতে সমাধান করতে পারে।

সহজ কথায়, ক্যাডেন্স অ্যালগরিদম আমাদের শেখায় কখন আগের যোগফল সাথে রাখা উচিত এবং কখন নতুন করে শুরু করা উচিত।

## ২. কেন ক্যাডেন্স অ্যালগরিদম? (The $O(N)$ Optimization)
ধরা যাক আমাদের কাছে একটি অ্যারে আছে: `[-2, 1, -3, 4, -1, 2, 1, -5, 4]`
- **ব্রুট ফোর্স:** সব পসিবল সাব-অ্যারে বের করে যোগফল চেক করা। টাইম: $O(N^2)$।
- **ক্যাডেন্স অ্যালগরিদম:** মাত্র একবার পুরো অ্যারে স্ক্যান করে সর্বোচ্চ যোগফল বের করা। টাইম: $O(N)$।

এটি ডাটা প্রসেসিংয়ের সময় এবং রিসোর্স অকল্পনীয়ভাবে সাশ্রয় করে।

## ৩. অ্যালগরিদম লজিক (The Core Logic)
ক্যাডেন্স অ্যালগরিদম একটি সিম্পল গ্রিডি (Greedy) বা ডাইনামিক প্রোগ্রামিং (DP) এপ্রোচ ফলো করে:
১. একটি `currentSum` মেইনটেইন করুন যা বর্তমান ইনডেক্স পর্যন্ত যোগফল স্টোর করবে।
২. একটি `maxSum` মেইনটেইন করুন যা এখন পর্যন্ত পাওয়া সর্বোচ্চ যোগফল সেভ করবে।
৩. প্রতিটি ইনডেক্সে:
   - `currentSum` এ বর্তমান ভ্যালু যোগ করুন।
   - যদি `currentSum` বর্তমান ভ্যালুর চেয়ে ছোট হয়, তবে নতুন করে বর্তমান ভ্যালু থেকেই শুরু করুন (`currentSum = max(num, currentSum + num)`)।
   - `maxSum` আপডেট করুন।

# Kadane's Algorithm (ক্যাডেন্স অ্যালগরিদম) 📈

## Kadane's Algorithm কী?
Kadane's Algorithm (উচ্চারণ: ক্যাডেন-স) হলো একটি গ্রিডি / ডাইনামিক প্রোগ্রামিং অ্যাপ্রোচ, যা মূলত **Maximum Subarray Sum** (একটি অ্যারে থেকে এমন একটি সাব-অ্যারে খুঁজে বের করা যার যোগফল সর্বোচ্চ) বের করতে ব্যবহৃত হয়।

Brute Force দিয়ে এটি করতে $O(N^2)$ বা $O(N^3)$ সময় লাগে। কিন্তু Kadane's Algorithm এটি মাত্র একটি লুপে **$O(N)$** সময়ে সলভ করে দেয়!

## অ্যালগরিদমের মূল কথা (The Core Idea)
পুরো অ্যালগরিদমটি একটি সিম্পল প্রশ্নের ওপর দাঁড়িয়ে আছে:
> "আমি কি আমার বর্তমান ইলিমেন্টকে আগের সাব-অ্যারের যোগফলের সাথে যুক্ত করব? নাকি আমি নিজেই নতুন করে একটি সাব-অ্যারে শুরু করব?"

**লজিক:**
- আমরা যখন অ্যারের ওপর দিয়ে লুপ চালাব, তখন `currentSum` এ ভ্যালু যোগ করতে থাকব।
- যদি কোনো কারণে `currentSum` এর মান **নেগেটিভ (Negative / < 0)** হয়ে যায়, তার মানে এই অংশটি পরবর্তীতে আমাদের যোগফলকে শুধুই কমিয়ে দেবে (ক্ষতি করবে)। তাই আমরা `currentSum` কে সাথে সাথে `0` (Zero) করে দেব! অর্থাৎ আগের সব নেগেটিভ ব্যাগেজ (Baggage) ফেলে দিয়ে নতুন করে শুরু করব।
- এর মাঝেই আমরা একটি `maxSum` ভেরিয়েবলে সবচেয়ে বড় যোগফলটি সেভ করে রাখব।

## কখন বুঝবেন যে Kadane's Algorithm ব্যবহার করতে হবে?
১. **Maximum Subarray Sum:** সরাসরি এই প্রবলেম দেওয়া থাকলে।
২. **Circular Subarray:** যখন অ্যারেটি সাইক্লিক (Circular) হয় এবং ম্যাক্সিমাম যোগফল বের করতে বলা হয়।
৩. **Maximum Product Subarray:** যোগের বদলে যখন গুণফল (Product) বের করতে বলা হয় (একটু মডিফিকেশন লাগে)।
৪. **Buying and Selling Stocks:** শেয়ার বাজারের লাভ-ক্ষতির প্রবলেমে।
