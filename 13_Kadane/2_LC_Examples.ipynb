## ১. Maximum Subarray (LC 53) 🟢
**টাইপ:** Basic Kadane's

### প্রবলেম স্টেটমেন্ট:
একটি অ্যারে দেওয়া আছে। এমন একটি সাব-অ্যারে খুঁজে বের করুন যার যোগফল সর্বোচ্চ।

### সমাধান:
১. `currentSum` এবং `maxSum` প্রথম ইলিমেন্ট দিয়ে শুরু করুন।
২. লুপের মাধ্যমে চেক করুন: `currentSum = Math.max(num, currentSum + num)`।
৩. `maxSum = Math.max(maxSum, currentSum)`।
এটি ও(এন) টাইমে সমাধান দেয়।

## ২. Maximum Product Subarray (LC 152) 🟡
**টাইপ:** Kadane's with Negatives

### প্রবলেম স্টেটমেন্ট:
এমন একটি সাব-অ্যারে খুঁজে বের করুন যার গুণফল (Product) সর্বোচ্চ।

### সমাধান:
১. যেহেতু দুটি নেগেটিভ নাম্বারের গুণফল পজিটিভ হয়ে যায়, তাই আমাদের `minProduct` ও ট্র্যাক করতে হবে।
২. প্রতি ধাপে `maxProduct`, `minProduct` এবং বর্তমান সংখ্যাটি তুলনা করে আপডেট করুন।
৩. যদি বর্তমান সংখ্যা নেগেটিভ হয়, তবে `max` এবং `min` সোয়াপ (Swap) করুন।

## ৩. Maximum Sum Circular Subarray (LC 918) 🟡
**টাইপ:** Circular Kadane's

### প্রবলেম স্টেটমেন্ট:
যদি অ্যারেটি সার্কুলার হয়, তবে সর্বোচ্চ যোগফল কত?

### সমাধান:
১. সাধারণ ক্যাডেন্স চালিয়ে `maxSum` বের করুন।
২. ইনভার্টেড ক্যাডেন্স চালিয়ে (সব সংখ্যার সাইন উল্টে দিয়ে) `minSum` বের করুন।
৩. উত্তর হবে: `max(maxSum, totalSum - minSum)`।
৪. নোট: যদি সব সংখ্যা নেগেটিভ হয়, তবে শুধু `maxSum` রিটার্ন করুন।

## ৪. Best Time to Buy and Sell Stock (LC 121) 🟢
**টাইপ:** Greedy / Kadane's Variant

### প্রবলেম স্টেটমেন্ট:
একবার কিনে একবার বিক্রি করলে সর্বোচ্চ কত প্রফিট সম্ভব?

### সমাধান:
১. `minPrice` ট্র্যাক করুন।
২. বর্তমান প্রাইস থেকে `minPrice` বিয়োগ করে প্রফিট বের করুন এবং ম্যাক্সিমামটি সেভ করুন।
৩. এটিও ক্যাডেন্সের একটি সিম্পল ভ্যারিয়েশন।
