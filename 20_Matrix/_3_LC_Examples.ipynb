# LeetCode Matrix Examples - FAANG Master Class 🚀

এই ফাইলে আমরা ম্যাট্রিক্স (Matrix / 2D Array) প্যাটার্নের সেরা কিছু লেটকোড প্রবলেম এবং সেগুলোর বিস্তারিত সমাধান দেখব।

## ১. Spiral Matrix (LC 54) 🟡
**টাইপ:** Matrix Traversal

### প্রবলেম স্টেটমেন্ট:
একটি $M \times N$ ম্যাট্রিক্সকে স্পাইরাল (Spiral) অর্ডারে প্রিন্ট করুন।

### সমাধান:
১. চারটি বাউন্ডারি মেইনটেইন করুন: `top`, `bottom`, `left`, `right`।
২. একটি লুপের মাধ্যমে প্রথমে বাম থেকে ডানে (Top row), তারপর উপর থেকে নিচে (Right column), তারপর ডান থেকে বামে (Bottom row), এবং শেষে নিচ থেকে উপরে (Left column) যান।
৩. প্রতি ধাপ শেষে সংশ্লিষ্ট বাউন্ডারিটি সংকুচিত (Shrink) করুন।

## ২. Rotate Image (LC 48) 🟡
**টাইপ:** In-place Transformation

### প্রবলেম স্টেটমেন্ট:
একটি ২ডি ইমেজকে (ম্যাট্রিক্স) ৯০ ডিগ্রি ক্লকওয়াইজ রোটেট করুন। অতিরিক্ত মেমরি ব্যবহার করা যাবে না।

### সমাধান:
১. প্রথমে ম্যাট্রিক্সটি ট্রান্সপোজ (Transpose) করুন (রো কে কলাম এবং কলাম কে রো বানান)।
২. এরপর প্রতিটি রো কে রিভার্স (Reverse) করুন।
ব্যাস! আপনার ম্যাট্রিক্স ৯০ ডিগ্রি ঘুরে গেছে।

## ৩. Set Matrix Zeroes (LC 73) 🟡
**টাইপ:** Space Optimization

### প্রবলেম স্টেটমেন্ট:
যদি ম্যাট্রিক্সের কোনো সেল ০ হয়, তবে তার পুরো রো এবং কলামকে ০ করে দিন।

### সমাধান:
১. ব্রুট ফোর্সে $O(M \times N)$ স্পেস লাগে। কিন্তু আমরা প্রথম রো এবং প্রথম কলামকে মার্কার (Marker) হিসেবে ব্যবহার করে $O(1)$ স্পেসে এটি সমাধান করতে পারি।
২. প্রথমে চেক করুন প্রথম রো বা কলামে ০ আছে কি না।
৩. বাকি ম্যাট্রিক্স স্ক্যান করে ০ পাওয়া গেলে সংশ্লিষ্ট প্রথম রো ও কলামে মার্ক করুন।
৪. মার্কার অনুযায়ী সব ০ করে দিন।

## ৪. Search a 2D Matrix (LC 74) 🟢
**টাইপ:** Binary Search on Matrix

### প্রবলেম স্টেটমেন্ট:
একটি সর্টেড ২ডি ম্যাট্রিক্সে একটি সংখ্যা খুঁজুন।

### সমাধান:
১. যেহেতু ম্যাট্রিক্সটি সর্টেড, একে একটি বিশাল ১ডি অ্যারে হিসেবে কল্পনা করা যায়।
২. টোটাল ইলিমেন্ট $N \times M$ এর ওপর বাইনারি সার্চ চালান।
৩. ইনডেক্স থেকে ম্যাট্রিক্স কোঅর্ডিনেট বের করার সূত্র: `row = index / cols`, `col = index % cols`।

## ৫. Word Search (LC 79) 🟡
**টাইপ:** Matrix DFS / Backtracking

### প্রবলেম স্টেটমেন্ট:
গ্রিডে কি কোনো নির্দিষ্ট শব্দ খুঁজে পাওয়া সম্ভব?

### সমাধান:
১. প্রতিটি সেল থেকে DFS শুরু করুন।
২. চার দিকে (উপরে, নিচে, ডানে, বামে) চেক করুন ক্যারেক্টার ম্যাচ করছে কি না।
৩. ভিজিটেড সেলগুলো সাময়িকভাবে মার্ক করুন এবং ব্যাকট্র্যাক করার সময় আনমার্ক করুন।
