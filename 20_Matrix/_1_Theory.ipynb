# Matrix (ম্যাট্রিক্স / ২ডি অ্যারে) - The FAANG Standard Theory 🔳

## ১. ভূমিকা (Detailed Introduction)
ম্যাট্রিক্স বা ২ডি অ্যারে (2D Array) হলো ডেটা অর্গানাইজ করার একটি গ্রিড-বেজড (Grid-based) পদ্ধতি। এটি মূলত রো (Row) এবং কলামের (Column) সমন্বয়ে গঠিত। FAANG ইন্টারভিউতে ম্যাট্রিক্স অত্যন্ত গুরুত্বপূর্ণ কারণ এটি ইমেজ প্রসেসিং, গেম ডেভেলপমেন্ট এবং ভৌগোলিক ডেটা এনালাইসিসের ভিত্তি।

সহজ কথায়, ম্যাট্রিক্স হলো একটি "দাবা বোর্ড" বা "এক্সেল শিট"। প্রতিটি ঘরের একটি নির্দিষ্ট ইনডেক্স $[i][j]$ থাকে যা দিয়ে আমরা ওই ঘরের ডেটা এক্সেস করতে পারি।

---

## ২. ম্যাট্রিক্স ট্রাভার্সাল পদ্ধতি (Traversal Techniques)
ম্যাট্রিক্সের ওপর কাজ করার জন্য মূলত ৩টি প্রধান ট্রাভার্সাল পদ্ধতি ব্যবহৃত হয়:

১. **Row-wise / Column-wise:** লিনিয়ারলি প্রতিটি রো বা কলাম স্ক্যান করা।
২. **Spiral Traversal:** ম্যাট্রিক্সের চারপাশ থেকে ঘুরতে ঘুরতে মাঝখানে পৌঁছানো।
৩. **Diagonal Traversal:** কোণাকোণিভাবে ডেটা স্ক্যান করা।

ইন্টারভিউতে প্রায়ই এই ট্রাভার্সালগুলো ইন-প্লেস (In-place) বা নির্দিষ্ট অর্ডারে করতে বলা হয়।

ম্যাট্রিক্স আমাদের এমন ডেটা হ্যান্ডেল করতে সাহায্য করে যার একাধিক ডাইমেনশন আছে।
- **ব্যবহার:** Image Processing, Path Finding (BFS/DFS on Grid), Matrix Multiplication, Dynamic Programming (2D Table).

## ৪. রিয়েল-ওয়ার্ল্ড FAANG উদাহরণ (Real-World Case Studies)

### ৪.১. গুগল ম্যাপস - গ্রিড-বেজড এরিয়া সার্চ
গুগল ম্যাপস পুরো পৃথিবীকে ছোট ছোট গ্রিড বা ম্যাট্রিক্সের মতো ব্লকে ভাগ করে। আপনি যখন কোনো এরিয়া সার্চ করেন, সিস্টেম ওই নির্দিষ্ট ম্যাট্রিক্স ব্লকগুলো প্রসেস করে রেজাল্ট দেখায়। এটি সার্চ টাইম অনেক কমিয়ে দেয়।

### ৪.২. ফেসবুক এবং ইনস্টাগ্রাম - ইমেজ ফিল্টার ও প্রসেসিং
প্রতিটি ডিজিটাল ছবি মূলত একটি বিশাল পিক্সেল ম্যাট্রিক্স। যখন আপনি ইনস্টাগ্রামে কোনো ফিল্টার অ্যাপ্লাই করেন, তখন একটি ইমেজ প্রসেসিং অ্যালগরিদম ওই পিক্সেল ম্যাট্রিক্সের প্রতিটি ভ্যালুকে একটি গাণিতিক ফাংশন দিয়ে পরিবর্তন করে। এটি রিয়েল-টাইমে করার জন্য অপ্টিমাইজড ম্যাট্রিক্স ম্যানিপুলেশন প্রয়োজন।

### ৪.৩. আমাজন - ওয়্যারহাউস অপ্টিমাইজেশন
আমাজনের ওয়্যারহাউস ফ্লোরকে একটি ২ডি গ্রিড বা ম্যাট্রিক্স হিসেবে চিন্তা করা হয়। রোবটগুলো যখন এক জায়গা থেকে অন্য জায়গায় যায়, তারা ম্যাট্রিক্স পাথ-ফাইন্ডিং অ্যালগরিদম (যেমন: BFS) ব্যবহার করে সবচেয়ে ছোট রাস্তা খুঁজে বের করে।

## ৫. ইন্টারভিউ চেকলিস্ট (Interview Checklist)
১. **Boundary Check:** লুপ চালানোর সময় ইনডেক্স কি ম্যাট্রিক্সের বাইরে চলে যাচ্ছে? (`i < 0 || i >= rows || j < 0 || j >= cols`)—এটি সবচেয়ে কমন বাগ।
২. **Visited Array:** যদি গ্রিডে ট্রাভার্স করেন, তবে একই ঘরে বারবার যাওয়া এড়াতে একটি `visited` ম্যাট্রিক্স ব্যবহার করুন।
৩. **In-place Transformation:** অনেক সময় এক্সট্রা মেমরি ব্যবহার না করে ম্যাট্রিক্সটি মডিফাই করতে বলা হয় (যেমন: Rotate Matrix 90 degrees)।

# 2D Matrix (২ডি ম্যাট্রিক্স) 🧮

## 2D Matrix কী?
2D Matrix বা ২-মাত্রিক অ্যারে (2D Array) হলো ডেটার একটি টেবুলার (Tabular) ফরম্যাট, যেখানে ডেটাগুলো রো (Rows) এবং কলাম (Columns) এ সাজানো থাকে। এটি দেখতে অনেকটা দাবার বোর্ডের (Chessboard) বা এক্সেল শিটের মতো।

জাভাতে এটি `int[][] matrix` আকারে লেখা হয়।

## Matrix Traversal (ম্যাট্রিক্স ট্রাভার্সাল)
সাধারণত একটি ম্যাট্রিক্সের সবগুলো ইলিমেন্ট দেখতে বা প্রসেস করতে দুটি Nested Loop (নেস্টেড লুপ) ব্যবহার করা হয়:
- **Outer Loop:** রো (Rows) কন্ট্রোল করে। (0 থেকে $m-1$)
- **Inner Loop:** কলাম (Columns) কন্ট্রোল করে। (0 থেকে $n-1$)

```java
int rows = matrix.length;
int cols = matrix[0].length;
for(int i = 0; i < rows; i++) {
    for(int j = 0; j < cols; j++) {
        // matrix[i][j] প্রসেস করুন
    }
}
```

## কখন বুঝবেন যে Matrix প্যাটার্ন ব্যবহার করতে হবে?
১. **Grid Search:** একটি গ্রিডের মধ্যে নির্দিষ্ট কোনো শব্দ, রাস্তা বা ডেটা খুঁজতে বললে। (এগুলোর জন্য আমরা গ্রিড DFS/BFS ব্যবহার করি, যা মডিউল 07 ও 08 এ আলোচনা করা হয়েছে)।
২. **In-place State Update:** ম্যাট্রিক্সের ডেটা পরিবর্তন করতে বললে, কিন্তু এক্সট্রা মেমোরি ব্যবহার করতে নিষেধ করলে (যেমন: Set Matrix Zeroes, Game of Life)।
৩. **Spiral / Diagonal Traversal:** ম্যাট্রিক্সটিকে স্পাইরাল (প্যাঁচানো) বা কোণাকুণি (Diagonal) প্রিন্ট করতে বললে।

## সাধারণ ভুল এবং সতর্কতা (Pitfalls)
১. **Index Out of Bounds:** ম্যাট্রিক্সের সবচেয়ে বড় সমস্যা হলো বাউন্ডারি ক্রস করে ফেলা। যেকোনো দিকে (উপরে, নিচে, ডানে, বামে) যাওয়ার আগে অবশ্যই চেক করতে হবে:
   `if (i >= 0 && i < rows && j >= 0 && j < cols)`
২. **Empty Matrix:** যদি `matrix.length == 0` হয়, তবে `matrix[0].length` চেক করতে গেলে `NullPointerException` খাবেন! তাই শুরুতেই এজ কেস হ্যান্ডেল করুন।

পরবর্তী ফাইলগুলোতে আমরা এর কিছু ফেমাস প্রবলেম এবং ইমপ্লিমেন্টেশন দেখব।
