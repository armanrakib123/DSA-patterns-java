# 🌲 Pattern 26: Depth-First Search (DFS) - গভীর বিশ্লেষণ

ডেপথ-ফার্স্ট সার্চ (DFS) হলো গ্রাফ এবং ট্রির ওপর ভিত্তি করে তৈরি অ্যালগরিদমগুলোর মধ্যে সবচেয়ে রহস্যময় এবং শক্তিশালী। এটি কোনো নোড থেকে শুরু করে যতটা গভীরে যাওয়া সম্ভব ততটা গভীরে যায় এবং যখন আর কোনো পথ থাকে না, তখন পেছনে ফিরে (Backtrack) অন্য পথে যাওয়ার চেষ্টা করে।

---

## 🧠 DFS এর মূল দর্শন (Core Philosophy)

DFS এর মূল মন্ত্র হলো: **"আগে শেষ পর্যন্ত যাও, তারপর অন্য পথ দেখো।"** 
এটি মূলত **রিকার্সন (Recursion)** এবং **স্ট্যাক (Stack)** ডেটা স্ট্রাকচারের ওপর ভিত্তি করে কাজ করে।

### ১. মেমরি অফ পাথ (Memory of Path)
DFS সবসময় একটি নির্দিষ্ট পাথ ধরে এগোয়। এটি মেমরিতে খুব সাশ্রয়ী কারণ এটি বিএফএস এর মতো পুরো লেভেল মনে রাখে না, শুধু বর্তমান পাথটি মনে রাখে।

### ২. ব্যাকট্র্যাকিং (Backtracking)
যখন আপনি একটি ডেড-এন্ডে (Dead-end) পৌঁছান, রিকার্সন কলটি ফিরে এসে আপনাকে আগের নোডে নিয়ে যায় যেখান থেকে আপনি অন্য কোনো প্রতিবেশীকে এক্সপ্লোর করতে পারেন।

---

## 🚀 DFS অ্যালগরিদমের ধাপসমূহ (Step-by-Step)

- ১. একটি নোড থেকে শুরু করুন এবং সেটিকে "Visited" মার্ক করুন।
- ২. নোডটির এমন একটি প্রতিবেশীকে খুঁজে বের করুন যা এখনো ভিজিট করা হয়নি।
- ৩. সেই প্রতিবেশীর ওপর আবার একই DFS কল করুন (Recursion)।
- ৪. যদি এমন কোনো প্রতিবেশী না থাকে, তবে আগের নোডে ফিরে যান।

---

## 🛠️ কেন DFS ইন্টারভিউতে এত গুরুত্বপূর্ণ?

- ১. **Connectivity & Components:** একটি গ্রাফে কতগুলো আলাদা আলাদা দ্বীপ (Islands) আছে তা বের করার জন্য DFS সেরা।
- ২. **Cycle Detection:** ডিরেক্টেড বা আনডিরেক্টেড গ্রাফে সাইকেল আছে কিনা তা DFS দিয়ে সহজেই বের করা যায়।
- ৩. **Path Finding:** যদি শর্টেস্ট পাথ না খুঁজে শুধু যেকোনো একটি পাথ খুঁজে বের করতে হয়।
- ৪. **Topological Sorting:** কাজের ওপর ভিত্তি করে কাজের ক্রম নির্ধারণ করা।

---

## 🚧 DFS বনাম BFS: কখন কোনটি ব্যবহার করবেন?

| বৈশিষ্ট্য | DFS | BFS |
| :--- | :--- | :--- |
| **পছন্দ** | গভীরতা বা ডেপথ | প্রশস্থতা বা লেভেল |
| **ডেটা স্ট্রাকচার** | Stack / Recursion | Queue |
| **শর্টেস্ট পাথ** | গ্যারান্টি দেয় না | গ্যারান্টি দেয় (Unweighted) |
| **ব্যবহার** | Topological sort, Path finding | Level order, Shortest Path |

---

## 📝 DFS এর টেমপ্লেট (Java)

```java
void dfs(int node, boolean[] visited, List<List<Integer>> adj) {
    visited[node] = true;
    // আপনার লজিক এখানে
    for (int neighbor : adj.get(node)) {
        if (!visited[neighbor]) {
            dfs(neighbor, visited, adj);
        }
    }
}
```

## 💡 ইন্টারভিউ ট্রিক: "Stack Overflow" থেকে সাবধান!
ইন্টারভিউতে যদি গ্রাফটি অনেক বড় হয় (Deep recursion), তবে রিকার্সিভ DFS করলে স্ট্যাক ওভারফ্লো এর ভয় থাকে। সেক্ষেত্রে আপনি ম্যানুয়ালি একটি `Stack` ডেটা স্ট্রাকচার ব্যবহার করে ইটারেটিভ DFS ইমপ্লিমেন্ট করতে পারেন। এটি আপনার এক্সপার্টিজ প্রমাণ করবে!

# Depth First Search (DFS) বা গভীরতা-ভিত্তিক অনুসন্ধান 🌳

## DFS কী?
Depth First Search (DFS) হলো ট্রি (Tree) বা গ্রাফ (Graph) ট্রাভার্স (Traverse/ভিজিট) করার একটি অ্যালগরিদম। 
নাম শুনেই বোঝা যাচ্ছে, এটি **"Depth" (গভীরতা)** কে অগ্রাধিকার দেয়। অর্থাৎ, এটি একটি রাস্তা ধরে যতদূর সম্ভব গভীরে (লিফ নোড পর্যন্ত) চলে যায়। যখন আর সামনে এগোনোর পথ থাকে না, তখন এটি ব্যাকট্র্যাক (Backtrack) করে ফিরে আসে এবং অন্য রাস্তা ধরে এগোয়।

## DFS কীভাবে কাজ করে?
DFS মূলত স্ট্যাক (Stack) ডেটা স্ট্রাকচার ব্যবহার করে কাজ করে। আমরা চাইলে নিজেরা Stack বানিয়ে (Iterative) কোড লিখতে পারি, তবে ৯৯% ক্ষেত্রে আমরা **রিকার্সন (Recursion)** ব্যবহার করে DFS লিখি, যা সিস্টেমের "Call Stack" ব্যবহার করে খুব সহজেই কাজটা করে দেয়।

**ট্রি (Tree) এর ক্ষেত্রে DFS (৩ প্রকার):**
১. **Pre-order:** Root -> Left -> Right (প্রথমে নিজেকে প্রসেস করা, তারপর চাইল্ড)।
২. **In-order:** Left -> Root -> Right (Binary Search Tree তে এটি ছোট থেকে বড় সর্টেড ডেটা দেয়)।
৩. **Post-order:** Left -> Right -> Root (প্রথমে সব চাইল্ডের কাজ শেষ করা, তারপর নিজের কাজ করা। যেমন: ট্রি এর হাইট বের করা)।

## Tree DFS বনাম Graph DFS
- **Tree DFS:** ট্রিতে কোনো সাইকেল (Cycle) বা লুপ থাকে না। অর্থাৎ, ওপর থেকে নিচে গেলে আবার ওপরে ফিরে যাওয়ার কোনো রাস্তা নেই। তাই এখানে ট্র্যাক রাখার দরকার নেই যে কোন নোড ভিজিট করা হয়েছে।
- **Graph DFS:** গ্রাফে সাইকেল থাকতে পারে। আপনি A থেকে B, B থেকে C, এবং C থেকে আবার A তে চলে আসতে পারেন! এভাবে ইনফিনিট লুপ (Infinite Loop) তৈরি হতে পারে। তাই গ্রাফে DFS করার সময় আমাদের একটি `visited` সেট (HashSet বা boolean array) ব্যবহার করতে হয়, যাতে একই নোড দুবার ভিজিট না হয়।

## কখন বুঝবেন যে DFS ব্যবহার করতে হবে? (Pattern Recognition)
- ১. **Trees:** প্রায় সব Binary Tree প্রবলেম DFS (বা রিকার্সন) দিয়ে সলভ করা যায়।
- ২. **Connectivity / Islands:** গ্রাফ বা 2D Grid (Matrix) এ যদি বলা হয় "কতগুলো দ্বীপ (Islands) আছে?" বা "এই নোড থেকে ওই নোডে যাওয়া যায় কি না?"।
- ৩. **Path Finding:** "A থেকে B তে যাওয়ার কোনো রাস্তা আছে কি না?"
- ৪. **Exhaustive Search:** যখন সম্ভাব্য সব পথ খুঁজে দেখতে হয় (এটি মূলত ব্যাকট্র্যাকিং এর অংশ)।

## DFS এর সুবিধা
- **Memory/Space:** BFS এর তুলনায় DFS এ মেমরি কম লাগে। কারণ DFS শুধু বর্তমান পথের (Path) নোডগুলোকে স্ট্যাকে রাখে (Space $O(H)$, যেখানে H হলো Height)।

পরবর্তী ফাইলগুলোতে আমরা Tree এবং Graph এর জন্য DFS টেমপ্লেট ও উদাহরণ দেখব।



সহজ কথায়, DFS হলো একটি গোলকধাঁধায় (Maze) পথ খোঁজার মতো। আপনি একটি রাস্তায় একদম শেষ পর্যন্ত যাবেন, যখনই কোনো রাস্তা পাবেন না, তখন এক ধাপ পিছিয়ে এসে অন্য রাস্তা ট্রাই করবেন। এই পদ্ধতিটি "LIFO" (Last In, First Out) প্রিন্সিপাল মেনে চলে।

---

## ২. কেন DFS? (The Structural Exploration)
DFS আমাদের একটি স্ট্রাকচারের একদম গভীরে গিয়ে তার বৈশিষ্ট্যগুলো বুঝতে সাহায্য করে।
- **ব্যবহার:** পাথ ফাইন্ডিং (Path Finding), কানেক্টেড কম্পোনেন্ট (Connected Components), সাইকেল ডিটেকশন (Cycle Detection), টপোলজিক্যাল সর্ট (Topological Sort)।

এটি সাধারণত রিকার্সন (Recursion) বা স্ট্যাক (Stack) ব্যবহার করে ইমপ্লিমেন্ট করা হয়, যা কোডকে অনেক বেশি ক্লিন এবং ছোট করে তোলে।

---

## ৩. DFS কিভাবে কাজ করে? (Algorithm Logic)
- ১. একটি স্টার্টিং নোড (Root/Source) সিলেক্ট করুন।
- ২. নোডটিকে 'Visited' হিসেবে মার্ক করুন।
- ৩. নোডটির কোনো আন-ভিজিটেড (Unvisited) নেইবার (Neighbor) থাকলে সেখানে যান এবং রিকার্সিভলি ১ ও ২ নম্বর ধাপ ফলো করুন।
- ৪. যদি কোনো নেইবার না থাকে, তবে আগের নোডটিতে ফিরে আসুন (Backtrack)।

## ৪. রিয়েল-ওয়ার্ল্ড Software Company উদাহরণ (Real-World Case Studies)

### ৪.১. ফেসবুক - মিউচুয়াল ফ্রেন্ডস ও কানেকশন
ফেসবুক যখন দেখে আপনি কার সাথে কানেক্টেড এবং আপনার ফ্রেন্ডলিস্টের মাধ্যমে তারা কতদূর বিস্তৃত, তখন তারা গ্রাফ ট্রাভার্সাল অ্যালগরিদম (যেমন DFS) ব্যবহার করে। এটি আপনার এবং অন্য একজন ইউজারের মধ্যে "Path" খুঁজে বের করতে সাহায্য করে।

### ৪.২. গুগল - ওয়েব ক্রলিং (Web Crawling)
গুগল বট (Googlebot) যখন ইন্টারনেটে নতুন ওয়েব পেজ খোঁজে, তারা প্রায়ই DFS বা BFS এর মডিফাইড ভার্সন ব্যবহার করে। একটি পেজের ভেতর থাকা লিংকগুলো ফলো করে তারা আরও গভীরে যায় এবং নতুন নতুন ইনফরমেশন ইনডেক্স করে।

### ৪.৩. অপারেটিং সিস্টেম - ফাইল সিস্টেম ট্রাভার্সাল
আপনার কম্পিউটারে যখন আপনি কোনো ফোল্ডারের ভেতর সার্চ করেন, তখন অপারেটিং সিস্টেম DFS ব্যবহার করে। এটি প্রথমে একটি ফোল্ডারের ভেতর ঢোকে, তার ভেতর সাব-ফোল্ডারে যায়, একদম ফাইল পর্যন্ত পৌঁছে আবার ফিরে এসে অন্য সাব-ফোল্ডারে ঢুকে সার্চ করে।

---

## ৫. DFS বনাম BFS
ইন্টারভিউতে এটি একটি ক্লাসিক প্রশ্ন:
- **DFS:** মেমরি সাশ্রয়ী (যদি গ্রাফ অনেক বেশি চওড়া হয়)। এটি আগে গভীরে যায়।
- **BFS:** শর্টেস্ট পাথ (Shortest Path) খোঁজার জন্য সেরা। এটি আগে আশেপাশে সব চেক করে।

## ৬. ইন্টারভিউ চেকলিস্ট (Interview Checklist)
- ১. **Visited Set:** গ্রাফের ক্ষেত্রে অবশ্যই একটি `visited` সেট বা অ্যারে ব্যবহার করুন, তা না হলে কোড ইনফিনিট লুপে পড়ে যাবে।
- ২. **Base Case:** রিকার্সনের ব্রেক কন্ডিশন কি ঠিক আছে?
- ৩. **Stack Overflow:** যদি গ্রাফ খুব বেশি গভীর হয় (যেমন ১০ লক্ষ নোড), তবে রিকার্সনের বদলে ইটারেটিভ (Iterative) স্ট্যাক ব্যবহার করা নিরাপদ।

---

## ৭. উপসংহার (Conclusion)
DFS হলো অ্যালগরিদমিক থিঙ্কিংয়ের একটি স্তম্ভ। এটি মাস্টারি করলে আপনি শুধু ট্রি বা গ্রাফ নয়, বরং অনেক জটিল সিস্টেমের ইন্টারনাল স্ট্রাকচার বুঝতে পারবেন। পরবর্তী ফাইলে আমরা এর বিভিন্ন টেমপ্লেট এবং লেটকোড উদাহরণ দেখব।
