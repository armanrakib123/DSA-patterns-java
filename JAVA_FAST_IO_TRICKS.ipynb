# 🚀 Java Fast I/O Guide

কম্পিটিটিভ প্রোগ্রামিং (Competitive Programming) বা হ্যাকারর‌্যাঙ্ক (Hackerrank)-এর মতো প্ল্যাটফর্মে জাভাতে সাধারণ `Scanner` এবং `System.out.print` ব্যবহার করলে অনেক সময় `Time Limit Exceeded (TLE)` এরর দেয়। এর কারণ হলো `Scanner` ক্লাসটি ইনপুট পার্স করার সময় রেগুলার এক্সপ্রেশন ব্যবহার করে এবং সিঙ্ক্রোনাইজড, যা বেশ ধীর গতির।

FAANG ইন্টারভিউতে সাধারণত লিটকোডের মতো প্ল্যাটফর্ম থাকে যেখানে ইনপুট/আউটপুট হ্যান্ডেল করতে হয় না, শুধু ফাংশন লিখতে হয়। তবে অনলাইন অ্যাসেসমেন্ট (Online Assessment - OA) রাউন্ডে অনেক সময় ম্যানুয়াল ইনপুট নিতে হয়। তখন Fast I/O জানা থাকা অত্যন্ত জরুরি।

---

## 🐢 সাধারণ পদ্ধতি (যা এড়িয়ে চলবেন)
```java
import java.util.Scanner;

public class SlowIO {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);
        int n = sc.nextInt();
        for (int i = 0; i < n; i++) {
            System.out.println(sc.nextInt()); // TLE হতে পারে বড় ইনপুটের জন্য
        }
    }
}
```

## ⚡ Fast I/O টেমপ্লেট (ইন্ডাস্ট্রি ও সিপি স্ট্যান্ডার্ড)

এই টেমপ্লেটটিতে ইনপুটের জন্য `BufferedReader` এবং `StringTokenizer` ব্যবহার করা হয়েছে, যা `Scanner` থেকে বহুগুণ দ্রুত। আর আউটপুটের জন্য `PrintWriter` বা `BufferedWriter` ব্যবহার করা হয়েছে।

আপনার যেকোনো অনলাইন অ্যাসেসমেন্টে কোড ফাইলের নিচে এই ক্লাসটি কপি-পেস্ট করে নিতে পারেন।

```java
import java.io.BufferedReader;
import java.io.IOException;
import java.io.InputStreamReader;
import java.io.PrintWriter;
import java.util.StringTokenizer;

public class FastIOTemplate {

    // ফাস্ট রিডার ক্লাস
    static class FastReader {
        BufferedReader br;
        StringTokenizer st;

        public FastReader() {
            br = new BufferedReader(new InputStreamReader(System.in));
        }

        String next() {
            while (st == null || !st.hasMoreElements()) {
                try {
                    st = new StringTokenizer(br.readLine());
                } catch (IOException e) {
                    e.printStackTrace();
                }
            }
            return st.nextToken();
        }

        int nextInt() {
            return Integer.parseInt(next());
        }

        long nextLong() {
            return Long.parseLong(next());
        }

        double nextDouble() {
            return Double.parseDouble(next());
        }

        String nextLine() {
            String str = "";
            try {
                if (st != null && st.hasMoreTokens()) {
                    str = st.nextToken("\n");
                } else {
                    str = br.readLine();
                }
            } catch (IOException e) {
                e.printStackTrace();
            }
            return str;
        }
    }

    public static void main(String[] args) {
        FastReader in = new FastReader();
        PrintWriter out = new PrintWriter(System.out);

        // ইনপুট পড়া
        int n = in.nextInt();
        
        for (int i = 0; i < n; i++) {
            long val = in.nextLong();
            // আউটপুট বাফারে লেখা (System.out.println এর চেয়ে অনেক দ্রুত)
            out.println(val * 2); 
        }

        // প্রোগ্রাম শেষে অবশ্যই বাফার ফ্লাশ করতে হবে, নাহলে আউটপুট কনসোলে প্রিন্ট হবে না
        out.flush();
    }
}
```

## 🧠 কেন এটি দ্রুত?
১. **BufferedReader:** এটি বড় চাঙ্ক (Chunk) আকারে ডেটা মেমোরিতে রিড করে রাখে, ফলে প্রতিবার হার্ড ডিস্ক বা কনসোল থেকে রিড করতে হয় না।
২. **StringTokenizer:** এটি শুধু স্পেস বা নিউলাইন দিয়ে স্ট্রিংকে ভাগ করে, কোনো রেগুলার এক্সপ্রেশন চেক করে না, যা `Scanner.nextInt()` করে থাকে।
৩. **PrintWriter:** `System.out.println` প্রতিবার কনসোলে আউটপুট রাইট করে (I/O অপারেশন খুব কস্টলি)। কিন্তু `PrintWriter` আউটপুটগুলো মেমোরিতে জমিয়ে রাখে এবং শেষে `flush()` কল করার পর একবারে প্রিন্ট করে।

> **টিপস:** লিটকোডে এর প্রয়োজন নেই। কিন্তু হ্যাকারর‌্যাঙ্ক, কোডফোর্সেস বা কোম্পানির নিজস্ব পোর্টালে (যেমন: মেটা হ্যাকারকাপ) এটি অবশ্যই ব্যবহার করবেন।


# 🎩 Common Tricks & Hacks (কোডিং ইন্টারভিউয়ের ম্যাজিক)

ইন্টারভিউতে অনেক সময় এমন কিছু ছোট ছোট ট্রিকস (Tricks) ব্যবহার করা যায়, যা আপনার কোডকে অনেক বেশি ক্লিন (Clean) এবং ফাস্ট (Fast) করে তোলে। এটি ইন্টারভিউয়ারের কাছে আপনার গভীর নলেজ প্রমাণ করে।

এখানে জাভা এবং ডেটা স্ট্রাকচারের কিছু মোস্ট ইউজড ট্রিকস দেওয়া হলো:

---

## ১. Array & Indexing Tricks

**Circular Array / Wrap Around:**
অ্যারের শেষ মাথায় পৌঁছানোর পর আবার শুরুতে ফিরে আসতে `%` (মডুলো) ব্যবহার করুন।
```java
int[] arr = {10, 20, 30, 40};
int n = arr.length;
for(int i = 0; i < 10; i++) {
    System.out.println(arr[i % n]); // 0, 1, 2, 3, 0, 1, 2, 3...
}
```

**অক্ষর (Character) থেকে ইনডেক্স বের করা:**
স্ট্রিংয়ের `char` কে `int` ইনডেক্সে কনভার্ট করার সবচেয়ে সহজ উপায়।
```java
char ch = 'c';
int index = ch - 'a'; // index হবে 2 ('a'=0, 'b'=1, 'c'=2)
```

## ২. Bit Manipulation Magic (বিট হ্যাকস)

**জোড়-বিজোড় (Even-Odd) চেক করা:**
`% 2` এর চেয়ে বিটওয়াইজ `& 1` বেশি ফাস্ট।
```java
if ((n & 1) == 0) {
    // Even
} else {
    // Odd
}
```

**২ দিয়ে গুণ বা ভাগ করা (Shift Operators):**
```java
int multiplyByTwo = n << 1; // n * 2
int divideByTwo = n >> 1;   // n / 2
```

**একটি সংখ্যা ২ এর পাওয়ার (Power of Two) কিনা চেক করা:**
```java
boolean isPowerOfTwo = (n > 0) && ((n & (n - 1)) == 0);
```

**সর্বডানের Set Bit (১) ক্লিয়ার করা:**
```java
n = n & (n - 1); // এটি অনেক অ্যালগরিদমে ব্যবহার হয় (যেমন Fenwick Tree)
```

## ৩. Math & Numbers

**Integer Overflow এড়ানো:**
যখন দুটি বড় সংখ্যা যোগ করে অর্ধেক করতে হয় (যেমন Binary Search-এ), তখন এভাবে করবেন:
```java
// ভুল উপায় (Integer Overflow হতে পারে)
int mid = (left + right) / 2; 

// সঠিক উপায়
int mid = left + (right - left) / 2;
```

**ডিজিটের সংখ্যা (Number of Digits) বের করা:**
লুপ চালিয়ে ভাগ করার দরকার নেই।
```java
int numDigits = (int) Math.log10(n) + 1; // n = 12345 হলে 5 রিটার্ন করবে
```

**Max/Min আপডেট করা (শর্টহ্যান্ড):**
```java
// বড় কোড
if (val > maxVal) { maxVal = val; }

// ক্লিন কোড
maxVal = Math.max(maxVal, val);
```

## ৪. HashMap & Set Tricks

**ডিফল্ট ভ্যালু রিট্রিভ করা (Frequency Count):**
অ্যারেতে কোন এলিমেন্ট কতবার আছে তা গোনার জন্য `getOrDefault` সবচেয়ে ক্লিন উপায়।
```java
Map<Integer, Integer> freq = new HashMap<>();
for (int num : arr) {
    // যদি num ম্যাপে না থাকে তবে 0 দেবে, আর থাকলে তার সাথে 1 যোগ করবে
    freq.put(num, freq.getOrDefault(num, 0) + 1); 
}
```

**অ্যারে থেকে ডুপ্লিকেট রিমুভ করা:**
অ্যারে থেকে HashSet তৈরি করা সবচেয়ে সহজ উপায়।
```java
Integer[] arr = {1, 2, 2, 3, 3}; // (int না, Integer হতে হবে)
Set<Integer> uniqueSet = new HashSet<>(Arrays.asList(arr));
```

## ৫. String Building
লুপের ভেতরে কখনো `String` ভেরিয়েবলের সাথে `+` দিয়ে কনক্যাটেনেট করবেন না। এতে মেমোরিতে বারবার নতুন অবজেক্ট তৈরি হয় এবং এটি `O(N^2)` টাইম নেয়। সবসময় `StringBuilder` ব্যবহার করবেন, এটি `O(N)` টাইম নেয়।

```java
// ❌ ভুল উপায় (ইন্টারভিউতে রেড ফ্ল্যাগ)
String s = "";
for (int i = 0; i < 1000; i++) {
    s += "a";
}

// ✅ সঠিক উপায়
StringBuilder sb = new StringBuilder();
for (int i = 0; i < 1000; i++) {
    sb.append("a");
}
String s = sb.toString();
```
