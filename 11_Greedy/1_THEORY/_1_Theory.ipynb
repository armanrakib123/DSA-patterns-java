# Greedy Algorithms (গ্রিডি বা লোভী অ্যালগরিদম) 🤑

## Greedy Algorithm কী?
"Greedy" শব্দের অর্থ লোভী। গ্রিডি অ্যালগরিদম প্রতিটি ধাপে সবচেয়ে সেরা (লোভনীয়) অপশনটি বেছে নেয়, এই আশায় যে প্রতিটি ছোট ছোট সেরা সিদ্ধান্ত (Local Optimum) মিলে শেষ পর্যন্ত একটি চূড়ান্ত সেরা সমাধান (Global Optimum) তৈরি করবে।

এটি ভবিষ্যতের কথা চিন্তা করে না বা অতীতে ফিরে যায় না (No Backtracking)। একবার একটি সিদ্ধান্ত নিয়ে নিলে তা চূড়ান্ত।

*উদাহরন:* আপনাকে বলা হলো সবচেয়ে কম সংখ্যক কয়েন (Coin) ব্যবহার করে ৯৯ টাকা দিতে। আপনার কাছে ১, ২, ৫, ১০, ২০, ৫০ টাকার কয়েন আছে। আপনি গ্রিডি হলে কী করবেন? 
প্রথমে সবচেয়ে বড় কয়েন (৫০) নেবেন। এরপর বাকি ৪৯ টাকার জন্য আবার সবচেয়ে বড় (২০) নেবেন। এভাবে চলতে থাকবে।

## কখন বুঝবেন যে Greedy ব্যবহার করতে হবে? (Pattern Recognition)
- ১. **Optimization Problems:** "Minimum number of...", "Maximum amount of...", "Shortest time to..." ইত্যাদি বের করতে বললে।
- ২. **Intervals & Scheduling:** অনেকগুলো কাজ বা মিটিং দেওয়া আছে, বলা হলো ম্যাক্সিমাম কয়টি মিটিং করা সম্ভব?
- ৩. **Sorting:** যদি দেখেন প্রবলেমের ডেটাকে সর্ট (Sort) করলে একটি সহজ প্যাটার্ন পাওয়া যাচ্ছে। প্রায় ৯৫% গ্রিডি প্রবলেমে আগে ডেটা সর্ট করে নিতে হয়।

## Greedy এর সীমাবদ্ধতা (Limitation)
সব প্রবলেমে গ্রিডি কাজ করে না। 
ধরি, আমাদের কাছে ১, ৩ এবং ৪ টাকার কয়েন আছে। আমাদের ৬ টাকা বানাতে হবে।
- গ্রিডি অ্যালগরিদম: প্রথমে ৪ টাকার কয়েন নেবে। এরপর বাকি ২ টাকার জন্য দুটি ১ টাকার কয়েন নেবে। (মোট ৩টি কয়েন)।
- কিন্তু আসল সেরা সমাধান হলো: দুটি ৩ টাকার কয়েন নেওয়া! (মোট ২টি কয়েন)।
এসব ক্ষেত্রে আমাদের **Dynamic Programming (DP)** ব্যবহার করতে হয়। 

ইন্টারভিউতে একটি প্রবলেমে গ্রিডি কাজ করবে কি না তা বোঝার জন্য প্রথমে কয়েকটি ছোট এক্সাম্পল (Test cases) ড্রাই-রান (Dry run) করে দেখতে হয়।

## 💡 Pattern 20: Greedy (গ্রীডি প্যাটার্ন)

গ্রীডি (Greedy) বা "লোভী" অ্যালগরিদম হলো এমন একটি কৌশল যেখানে প্রতিটি ধাপে সবচেয়ে **ভালো (Optimal)** অপশনটি বেছে নেওয়া হয় এই আশায় যে, প্রতিটি ধাপে ভালো করলে চূড়ান্ত রেজাল্টটিও ভালো হবে। 

---

## 🧠 ইন্টুইশন (Intuition)

সহজ কথায়, গ্রীডি হলো "বর্তমান নিয়ে ভাবা"। এটি ভবিষ্যতের কথা চিন্তা না করে বর্তমান মুহূর্তে যেটি সবচেয়ে লাভজনক মনে হয়, সেটিকেই গ্রহণ করে। 

**উদাহরণ:**
আপনার কাছে ৫ টাকা, ১০ টাকা এবং ২০ টাকার নোট আছে। আপনাকে সর্বনিম্ন সংখ্যক নোট ব্যবহার করে ৩৫ টাকা বানাতে হবে। 
- গ্রীডি লজিক: সবার আগে সবচেয়ে বড় নোটটি নাও (২০ টাকা)। বাকি রইল ১৫ টাকা। এরপর বড় নোট নাও (১০ টাকা)। বাকি রইল ৫ টাকা। শেষে নাও ৫ টাকা। মোট ৩টি নোট। 
এটি একটি অপ্টিমাল সল্যুশন। (তবে সবসময় গ্রীডি কাজ করে না, যেমন ১, ৩, ৪ টাকার কয়েন দিয়ে ৬ টাকা বানাতে গেলে)।

---

## 🚀 কেন এটি ব্যবহার করবেন?

গ্রীডি অ্যালগরিদম সাধারণত ডাইনামিক প্রোগ্রামিং (DP) এর চেয়ে অনেক দ্রুত কাজ করে কারণ এখানে আমাদের সব অপশন ট্রাই করতে হয় না। 

**বৈশিষ্ট্য:**
- ১. এটি খুব দ্রুত এবং সহজ (Simple to implement)।
- ২. টাইম কমপ্লেক্সিটি সাধারণত O(N log N) বা O(N) হয়।
- ৩. একবার কোনো অপশন বেছে নিলে তা আর পরিবর্তন করা হয় না (No backtracking)।

---

## 🚨 কখন বুঝবেন Greedy ব্যবহার করতে হবে?

ইন্টারভিউতে গ্রীডি চেনা একটু ট্রিকি, কারণ এটি অনেক সময় DP এর মতো মনে হতে পারে। তবে নিচের সিগন্যালগুলো লক্ষ্য করবেন:
- ১. প্রবলেমে যদি **"Minimum"**, **"Maximum"**, বা **"Optimize"** এর কথা বলা থাকে।
- ২. যদি দেখা যায় প্রতিটি ধাপে স্থানীয়ভাবে (locally) সেরা অপশনটি বেছে নিলেই গ্লোবাল (globally) সেরা উত্তর পাওয়া যাচ্ছে।
- ৩. যদি প্রবলেমটি "ইন্টারভ্যাল" বা "শিডিউলিং" রিলেটেড হয়।

---

## 🛠️ গ্রীডি বনাম ডাইনামিক প্রোগ্রামিং (DP)

- **Greedy:** প্রতিটি ধাপে শুধু সেরা অপশনটি বেছে নেয়। (ফাস্ট কিন্তু সবসময় কাজ করে না)।
- **DP:** সব সম্ভাব্য অপশন চেক করে সেরাটি বেছে নেয়। (স্লো কিন্তু সবসময় সঠিক উত্তর দেয়)।
