# Linked List (লিঙ্কড লিস্ট) - The FAANG Standard Theory 🔗

## ১. ভূমিকা (Detailed Introduction)
লিঙ্কড লিস্ট (Linked List) হলো একটি লিনিয়ার ডেটা স্ট্রাকচার যেখানে ইলিমেন্টগুলো মেমরিতে পাশাপাশি থাকে না (যেমনটা অ্যারেতে থাকে)। প্রতিটি ইলিমেন্ট বা "নোড" (Node) তার পরবর্তী নোডের অ্যাড্রেস বা রেফারেন্স স্টোর করে রাখে। FAANG ইন্টারভিউতে লিঙ্কড লিস্ট অত্যন্ত গুরুত্বপূর্ণ কারণ এটি মেমরি ম্যানেজমেন্ট এবং পয়েন্টার ম্যানিপুলেশন (Pointer Manipulation) বোঝার জন্য সেরা টুল।

সহজ কথায়, লিঙ্কড লিস্ট হলো একটি "ট্রেন"। প্রতিটি বগি (Node) তার পরের বগির সাথে কানেক্টেড। আপনি যদি একটি বগি মাঝখানে যোগ করতে চান, তবে আপনাকে শুধু শিকলটি (Pointer) পরিবর্তন করতে হবে, পুরো ট্রেনটি সরাতে হবে না।

## ২. লিঙ্কড লিস্ট বনাম অ্যারে (The Trade-off)
ইন্টারভিউতে এটি একটি মাস্ট-আস্কড কোয়েশ্চন:
- **অ্যারে:** র্যান্ডম এক্সেস ($O(1)$) সম্ভব কিন্তু মাঝখানে ইনসার্ট বা ডিলিট করা স্লো ($O(N)$)।
- **লিঙ্কড লিস্ট:** ইনসার্ট বা ডিলিট করা ফাস্ট ($O(1)$) যদি আপনার কাছে ওই পজিশনের রেফারেন্স থাকে, কিন্তু র্যান্ডম এক্সেস স্লো ($O(N)$)।

লিঙ্কড লিস্ট তখন ব্যবহার করা হয় যখন আমাদের ডেটা সাইজ ডাইনামিক এবং প্রচুর পরিমাণে ইনসার্ট/ডিলিট করতে হয়।

## ৩. লিঙ্কড লিস্টের প্রকারভেদ
১. **Singly Linked List:** প্রতিটি নোড শুধু তার পরের নোডকে চেনে।
২. **Doubly Linked List:** প্রতিটি নোড তার আগের এবং পরের—উভয় নোডকেই চেনে। (এটি ব্রাউজার হিস্ট্রি বা মিউজিক প্লেয়ার লিস্টে ব্যবহৃত হয়)।
৩. **Circular Linked List:** শেষ নোডটি আবার প্রথম নোডের সাথে কানেক্টেড থাকে।

## ৪. রিয়েল-ওয়ার্ল্ড FAANG উদাহরণ (Real-World Case Studies)

### ৪.১. গুগল ক্রোম - ব্রাউজার ব্যাক এবং ফরোয়ার্ড বাটন
আপনার ব্রাউজারে যখন আপনি কোনো ওয়েবসাইটে যান এবং 'Back' বা 'Forward' বাটন প্রেস করেন, সেটি মূলত একটি ডাবলি লিঙ্কড লিস্ট (Doubly Linked List) মেইনটেইন করে। প্রতিটি পেজ একটি নোড এবং আপনি পয়েন্টার সরিয়ে এক পেজ থেকে অন্য পেজে যাতায়াত করেন।

### ৪.২. আমাজন এবং নেটফ্লিক্স - মিউজিক/ভিডিও প্লেলিস্ট
মিউজিক প্লেয়ারের প্লেলিস্টে গানগুলো লিঙ্কড লিস্ট হিসেবে থাকে। যখন আপনি 'Next' বা 'Previous' চাপেন, প্লেয়ার শুধু পয়েন্টারটি চেঞ্জ করে। সার্কুলার লিঙ্কড লিস্ট ব্যবহার করা হয় যখন আপনি 'Repeat All' মোড অন করেন, যাতে শেষ গানের পর আবার প্রথম গানটি শুরু হয়।

### ৪.৩. লিনাক্স কার্নেল - মেমরি এবং প্রসেস ম্যানেজমেন্ট
অপারেটিং সিস্টেম যখন মেমরি অ্যালোকেট করে বা প্রসেস শিডিউল করে, তখন তারা লিঙ্কড লিস্ট ব্যবহার করে ফ্রি মেমরি ব্লক এবং রানিং প্রসেসগুলোর লিস্ট মেইনটেইন করে। এটি খুব দ্রুত মেমরি ব্লক যোগ করতে বা সরাতে সাহায্য করে।

## ৫. ইন্টারভিউ চেকলিস্ট (Interview Checklist)
১. **The Dummy Node:** অনেক প্রবলেমে একটি `dummy` নোড ব্যবহার করলে আপনার এজ-কেস (যেমন: হেড রিমুভ করা) হ্যান্ডেল করা অনেক সহজ হয়ে যায়।
২. **Reverse Logic:** লিঙ্কড লিস্ট রিভার্স করা একটি বেসিক কিন্তু গুরুত্বপূর্ণ স্কিল। এটি বারবার প্র্যাকটিস করুন।
৩. **Fast & Slow Pointers:** লিঙ্কড লিস্টে সাইকেল ডিটেক্ট করা বা মিডল নোড খুঁজে বের করার জন্য এটি সবচেয়ে অপ্টিমাইজড পদ্ধতি।

লিঙ্কড লিস্ট হলো ডেটা স্ট্রাকচারের ভিত্তি। এটি আয়ত্ত করলে আপনার পয়েন্টার এবং মেমরি কনসেপ্ট অনেক ক্লিয়ার হবে, যা আপনাকে হার্ড লেভেলের গ্রাফ বা ট্রি প্রবলেম সলভ করতে সাহায্য করবে। পরবর্তী ফাইলে আমরা এর বাস্তব উদাহরণ এবং লেটকোড প্রবলেম দেখব।

# Linked List (লিংকড লিস্ট) 🔗

## Linked List কী?
Linked List হলো এমন একটি ডেটা স্ট্রাকচার যেখানে ডেটাগুলো মেমোরিতে পাশাপাশি (Contiguous) থাকে না (Array এর মতো নয়)। এর প্রতিটি ইলিমেন্টকে **Node** বলা হয়। 

প্রতিটি Node এর দুটি অংশ থাকে:
১. **Value (val):** আসল ডেটা।
২. **Next Pointer (next):** পরবর্তী নোডের মেমোরি অ্যাড্রেস (Reference)।

শেষ নোডের next পয়েন্টারটি `null` থাকে, যা বোঝায় লিস্ট এখানে শেষ।

## Array বনাম Linked List
- **সার্চ (Search):** Array তে ইনডেক্স দিয়ে $O(1)$ এ ডেটা পাওয়া যায়। কিন্তু Linked List এ শুরু থেকে খুঁজতে হয় $O(N)$।
- **ইনসার্ট/ডিলিট (Insert/Delete):** Array তে মাঝখানে ডেটা ঢোকাতে হলে বাকিদের সরাতে (Shift) হয় $O(N)$। কিন্তু Linked List এ শুধু পয়েন্টার পরিবর্তন করলেই হয় $O(1)$ (যদি নোডের পজিশন জানা থাকে)।

## কখন বুঝবেন যে Linked List প্যাটার্ন ব্যবহার করতে হবে?
১. **ইন-প্লেস মডিফিকেশন (In-place Modification):** এক্সট্রা মেমোরি ($O(1)$ Space) ব্যবহার না করে ডেটা রিভার্স (Reverse) বা রিমুভ (Remove) করতে বললে।
২. **পয়েন্টার ম্যানিপুলেশন:** ডেটার স্ট্রাকচার পরিবর্তন করার কথা থাকলে (যেমন: Swap Nodes, Rotate List)।

## কমন টেকনিকস ও প্যাটার্নস
১. **Fast & Slow Pointer:** সাইকেল (Cycle) বের করতে বা মাঝখানের নোড (Middle Node) খুঁজতে। (এটি আমরা 03_FastSlowPointer মডিউলে বিস্তারিত দেখেছি)।
২. **Dummy Node:** হেড (Head) নোড পরিবর্তন হতে পারে এমন ক্ষেত্রে একটি Dummy Node তৈরি করে নেওয়া খুব উপকারে আসে। (`ListNode dummy = new ListNode(0); dummy.next = head;`)
৩. **Multiple Pointers (Prev, Curr, Next):** লিস্ট রিভার্স (Reverse) করার জন্য ৩টি পয়েন্টার মেইনটেইন করা।

পরবর্তী ফাইলগুলোতে আমরা রিভার্সিং এবং মার্জিং এর ইমপ্লিমেন্টেশন দেখব।
