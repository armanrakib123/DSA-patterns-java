## ১. Reverse Linked List (LC 206) 🟢
**টাইপ:** Basic Pointer Manipulation

### প্রবলেম স্টেটমেন্ট:
একটি সিংলি লিঙ্কড লিস্ট রিভার্স (Reverse) করুন।

### সমাধান:
১. তিনটি পয়েন্টার ব্যবহার করুন: `prev`, `curr`, এবং `next`.
২. লুপের মাধ্যমে বর্তমান নোডের `next` পয়েন্টারকে `prev` এ সেট করুন।
৩. পয়েন্টারগুলো এক ধাপ করে ডানে সরান।
এটি ও(এন) টাইমে এবং ও(১) স্পেসে সম্পন্ন হয়।

## ২. Merge Two Sorted Lists (LC 21) 🟢
**টাইপ:** Two Pointer / Dummy Node

### প্রবলেম স্টেটমেন্ট:
দুটি সর্টেড লিঙ্কড লিস্ট মার্জ করে একটি নতুন সর্টেড লিস্ট তৈরি করুন।

### সমাধান:
১. একটি `dummy` নোড তৈরি করুন যা রেজাল্ট লিস্টের শুরু হিসেবে কাজ করবে।
২. দুটি লিস্টের হেড তুলনা করুন এবং ছোটটি রেজাল্ট লিস্টে যোগ করুন।
৩. লুপ শেষে যদি কোনো লিস্টে নোড বাকি থাকে, তবে সেটি সরাসরি রেজাল্ট লিস্টের শেষে যুক্ত করে দিন।

## ৩. Remove Nth Node From End of List (LC 19) 🟡
**টাইপ:** Two Pointer (Fast & Slow)

### প্রবলেম স্টেটমেন্ট:
লিস্টের শেষ থেকে $N$-তম নোডটি রিমুভ করুন।

### সমাধান:
১. একটি `fast` পয়েন্টারকে $N$ ঘর আগে পাঠিয়ে দিন।
২. এরপর `slow` এবং `fast` পয়েন্টার একসাথে এক ঘর করে সরান।
৩. যখন `fast` লিস্টের শেষে পৌঁছাবে, `slow` পয়েন্টার ঠিক $N$-তম নোডের আগে থাকবে।

## ৪. Add Two Numbers (LC 2) 🟡
**টাইপ:** Math + LinkedList

### প্রবলেম স্টেটমেন্ট:
দুটি লিঙ্কড লিস্টের মাধ্যমে দুটি রিভার্সড নাম্বার দেওয়া আছে। তাদের যোগফল বের করে লিঙ্কড লিস্ট হিসেবে রিটার্ন করুন।

### সমাধান:
১. দুটি লিস্টের নোডগুলো যোগ করুন এবং একটি `carry` মেইনটেইন করুন।
২. যদি যোগফল ১০ বা তার বেশি হয়, তবে `carry` ১ হিসেবে পরবর্তী নোডে পাস করুন।
৩. লুপ শেষে যদি কোনো `carry` বাকি থাকে, তবে নতুন একটি নোড তৈরি করুন।

## ৫. Copy List with Random Pointer (LC 138) 🟡
**টাইপ:** HashMap / In-place Manipulation

### প্রবলেম স্টেটমেন্ট:
একটি লিঙ্কড লিস্টের ডিপ কপি (Deep Copy) তৈরি করুন যেখানে প্রতিটি নোডে একটি র‍্যান্ডম পয়েন্টার আছে।

### সমাধান:
১. একটি HashMap ব্যবহার করুন অরিজিনাল নোড থেকে কপি নোড ম্যাপ করতে।
২. প্রথমে সব নোড কপি করুন এবং ম্যাপে রাখুন।
৩. দ্বিতীয় লুপে ম্যাপ ব্যবহার করে `next` এবং `random` পয়েন্টারগুলো সেট করুন।
