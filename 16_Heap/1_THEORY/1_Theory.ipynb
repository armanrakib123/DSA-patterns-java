# Heap / Priority Queue (হিপ / প্রায়োরিটি কিউ) - The FAANG Standard Theory 🏔️

## ১. ভূমিকা (Detailed Introduction)
হিপ (Heap) হলো একটি বিশেষ ধরনের বাইনারি ট্রি (Binary Tree) যা একটি নির্দিষ্ট শর্ত মেনে চলে। এটি মূলত প্রায়োরিটি কিউ (Priority Queue) ইমপ্লিমেন্ট করার জন্য ব্যবহৃত হয়। যখন আমাদের কোনো ডেটাসেট থেকে সবচেয়ে ছোট বা সবচেয়ে বড় ইলিমেন্টটি খুব দ্রুত ($O(1)$ টাইমে) খুঁজে বের করতে হয়, তখন হিপ আমাদের সবচেয়ে বড় হাতিয়ার। FAANG ইন্টারভিউতে "Top K elements" বা "Median finding" সংক্রান্ত সমস্যার এটিই সবচেয়ে অপ্টিমাইজড সমাধান।

সহজ কথায়, হিপ হলো একটি "অটো-সর্টেড" কালেকশন যেখানে সবসময় রাজা (ম্যাক্সিমাম বা মিনিমাম) একদম উপরে থাকে।

---

## ২. কেন হিপ? (The Power of Logarithmic Maintenance)
যদি আমাদের কাছে একটি লিস্ট থাকে এবং আমরা প্রতিবার সবচেয়ে ছোট ইলিমেন্টটি চাই:
- **আন-সর্টেড অ্যারে:** প্রতিবার সার্চ করতে $O(N)$ সময় লাগবে।
- **সর্টেড অ্যারে:** সর্ট করতে $O(N \log N)$ লাগবে এবং ইনসার্ট করতে $O(N)$ লাগবে।
- **হিপ:** ইনসার্ট এবং রিমুভ—উভয়ই মাত্র $O(\log N)$ টাইমে করা সম্ভব এবং টপ ইলিমেন্ট পাওয়া যায় $O(1)$ এ।

এটি বিশাল পরিমাণ ডাইনামিক ডেটা হ্যান্ডেল করার জন্য অত্যন্ত কার্যকর।

## ৩. হিপের প্রকারভেদ
ইন্টারভিউতে মূলত দুই ধরনের হিপ ব্যবহৃত হয়:

১. **Min-Heap (মিন-হিপ):** এখানে প্যারেন্ট নোড তার চিলড্রেনদের চেয়ে সবসময় ছোট হয়। ফলে সবচেয়ে ছোট ইলিমেন্টটি রুটে (Root) থাকে।
২. **Max-Heap (ম্যাক্স-হিপ):** এখানে প্যারেন্ট নোড তার চিলড্রেনদের চেয়ে সবসময় বড় হয়। ফলে সবচেয়ে বড় ইলিমেন্টটি রুটে থাকে।

## ৪. রিয়েল-ওয়ার্ল্ড FAANG উদাহরণ (Real-World Case Studies)

### ৪.১. গুগল - সার্চ রেজাল্ট র‍্যাঙ্কিং
গুগল যখন আপনার কোয়েরির জন্য হাজার হাজার রেজাল্ট পায়, তখন তারা সেগুলোকে র‍্যাঙ্কিং স্কোর অনুযায়ী সাজায়। "Top K" মোস্ট রিলেভেন্ট রেজাল্ট দেখানোর জন্য তারা হিপ ব্যবহার করে। এটি নিশ্চিত করে যে সবচেয়ে সেরা রেজাল্টগুলোই আপনি আগে দেখতে পাচ্ছেন।

### ৪.২. আমাজন - রিয়েল-টাইম অর্ডার প্রাইওরিটাইজেশন
আমাজনের ওয়্যারহাউসে যখন হাজার হাজার অর্ডার আসে, তখন ডেলিভারি ডেট এবং মেম্বারশিপ টাইপ (Prime vs Regular) অনুযায়ী অর্ডারগুলোকে প্রায়োরিটি কিউতে রাখা হয়। হিপ ব্যবহার করে সিস্টেম দ্রুত ডিসাইড করে কোন অর্ডারটি আগে প্রসেস করতে হবে।

### ৪.৩. অপারেটিং সিস্টেম - টাস্ক শিডিউলিং (CPU Scheduling)
আপনার কম্পিউটার বা স্মার্টফোনে যখন অনেকগুলো অ্যাপ একসাথে চলে, তখন অপারেটিং সিস্টেম প্রতিটি প্রসেসের প্রায়োরিটি অনুযায়ী CPU টাইম অ্যালোকেট করে। এটি মেইনটেইন করার জন্য হিপ ব্যবহার করা হয় যাতে হাই-প্রায়োরিটি টাস্কগুলো (যেমন: ভিডিও কল) ল্যাগ না করে।

# Heap / Priority Queue (হিপ বা প্রায়োরিটি কিউ) ⛰️

## Heap কী?
হিপ (Heap) হলো একটি বিশেষ ধরনের বাইনারি ট্রি, যা Array এর মাধ্যমে ইমপ্লিমেন্ট করা হয়। এর মূল বৈশিষ্ট্য হলো: এটি সব সময় সবচেয়ে ছোট (Min) বা সবচেয়ে বড় (Max) ভ্যালুটিকে একদম ওপরে (Root) রাখে।

জাভাতে (Java) Heap কে **PriorityQueue** ক্লাস দিয়ে ইমপ্লিমেন্ট করা হয়। সাধারণ Queue তে যে আগে ঢোকে সে আগে বের হয় (FIFO)। কিন্তু Priority Queue তে যার প্রায়োরিটি (Priority) বেশি, সে আগে বের হয়!

## ২ প্রকারের Heap
১. **Min Heap (মিন হিপ):** সবচেয়ে ছোট ইলিমেন্টটি টপে (Top) থাকে। জাভাতে এটি বাই ডিফল্ট (Default) থাকে।
   - `PriorityQueue<Integer> minHeap = new PriorityQueue<>();`
২. **Max Heap (ম্যাক্স হিপ):** সবচেয়ে বড় ইলিমেন্টটি টপে থাকে। জাভাতে এটি বানাতে Comparator ব্যবহার করতে হয়।
   - `PriorityQueue<Integer> maxHeap = new PriorityQueue<>(Collections.reverseOrder());`

## Time Complexity
- টপে থাকা ইলিমেন্ট (Min/Max) দেখতে (Peek): $O(1)$
- নতুন ইলিমেন্ট যোগ করতে (Add/Offer): $O(\log N)$
- টপের ইলিমেন্ট বের করতে (Poll/Remove): $O(\log N)$
- একটি Array থেকে পুরো Heap বানাতে (Heapify): $O(N)$

## কখন বুঝবেন যে Heap ব্যবহার করতে হবে? (Pattern Recognition)
১. **Top / K-th Elements:** প্রবলেমে যদি বলা হয় "Find the K-th largest", "K-th smallest", "Top K frequent elements"। এটি Heap এর সবচেয়ে কমন প্যাটার্ন!
২. **Streaming Data:** যদি ডেটা একে একে আসতে থাকে (Stream) এবং আপনাকে যেকোনো মুহূর্তে ম্যাক্সিমাম বা মিনিমাম ভ্যালু বলতে হয় (যেমন: Median in a Data Stream)।
৩. **Sorting by Frequency/Value:** অনেকগুলো অবজেক্টকে তাদের ফ্রিকোয়েন্সি বা নির্দিষ্ট কোনো ভ্যালুর ওপর ভিত্তি করে বারবার সর্ট করতে হলে।

## K-th Element বের করার রুল (The Golden Rule)
- K-th **Largest** (সবচেয়ে বড়) বের করতে বললে: **Min Heap** ব্যবহার করবেন! (সাইজ $K$ এর মধ্যে লিমিট রাখবেন। ছোটগুলো পপ হয়ে যাবে, টপে K-তম বড়টি থাকবে)।
- K-th **Smallest** (সবচেয়ে ছোট) বের করতে বললে: **Max Heap** ব্যবহার করবেন!

পরবর্তী ফাইলগুলোতে আমরা এর ইমপ্লিমেন্টেশন দেখব।
