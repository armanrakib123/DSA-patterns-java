# Advanced Graphs & Shortest Path (অ্যাডভান্সড গ্রাফ) 🌐

## গ্রাফের পূর্বপাঠ (Recap)
আমরা 07_DFS এবং 08_BFS মডিউলে গ্রাফের বেসিক ট্রাভার্সাল (Traversal), Connected Components, এবং Matrix/Grid প্রবলেমগুলো দেখেছি। টপোলজিক্যাল সর্ট (09_TopologicalSort) এবং সাইকেল ডিটেকশন (10_UnionFind) ও গ্রাফেরই অংশ।

এই মডিউলে আমরা গ্রাফের সবচেয়ে ইম্পর্টেন্ট এবং অ্যাডভান্সড একটি টপিক দেখব: **Shortest Path on Weighted Graphs (ওজনযুক্ত গ্রাফে সবচেয়ে ছোট রাস্তা)**।

## Unweighted vs Weighted Graph
- **Unweighted (ওজনহীন):** প্রতিটি এজের (রাস্তার) দূরত্ব বা খরচ 1। এখানে Shortest Path বের করতে আমরা সাধারণ **BFS (Queue)** ব্যবহার করি।
- **Weighted (ওজনযুক্ত):** প্রতিটি এজের একটি নির্দিষ্ট খরচ (Weight) থাকে (যেমন: A থেকে B তে যেতে ৫ টাকা লাগে)। এখানে সাধারণ BFS কাজ করে না। তখন আমাদের **Dijkstra's Algorithm (ডাইকস্ট্রা)** ব্যবহার করতে হয়।

## Dijkstra's Algorithm (ডাইকস্ট্রা) কী?
Dijkstra হলো একটি গ্রিডি (Greedy) অ্যালগরিদম যা Weighted Graph এ শর্টেস্ট পাথ বের করতে ব্যবহৃত হয়।
এটি সাধারণ BFS এর মতোই, শুধু দুটি পার্থক্য আছে:
১. Queue এর বদলে **Priority Queue (Min Heap)** ব্যবহার করতে হয়।
২. একটি `distances` অ্যারে রাখতে হয়, যা ট্র্যাক করে কোন নোডে পৌঁছাতে এখন পর্যন্ত সবচেয়ে কম কত খরচ হয়েছে।

## অ্যালগরিদম কীভাবে কাজ করে?
১. শুরুর নোডটির দূরত্ব $0$ ধরে Priority Queue তে ঢোকানো হয়। বাকি সব নোডের দূরত্ব অসীম ($\infty$ বা `Integer.MAX_VALUE`) ধরা হয়।
২. Queue থেকে সবসময় সবচেয়ে কম খরচের (Min Distance) নোডটি বের করা হয়।
৩. এরপর তার প্রতিবেশীদের (Neighbors) কাছে যাওয়ার খরচ হিসাব করা হয়: 
   `newDistance = currentDistance + edgeWeight`
৪. যদি `newDistance` ওই প্রতিবেশীর আগের জানা দূরত্বের চেয়ে কম হয়, তবে তার দূরত্ব আপডেট করা হয় এবং তাকে Queue তে ঢোকানো হয়।

## Limitation (সীমাবদ্ধতা)
Dijkstra's Algorithm **Negative Edge Weights (নেগেটিভ ওজনের রাস্তা)** এর জন্য কাজ করে না! যদি নেগেটিভ সাইকেল থাকে, তবে এটি ইনফিনিট লুপে পড়ে যেতে পারে। তখন Bellman-Ford অ্যালগরিদম ব্যবহার করতে হয়।

পরবর্তী ফাইলে আমরা Dijkstra's Algorithm এর ফুল ইমপ্লিমেন্টেশন দেখব।


# Greedy Algorithm (গ্রিডি অ্যালগরিদম) - The Software Company Standard Theory 💰

## ১. ভূমিকা (Detailed Introduction)
গ্রিডি অ্যালগরিদম (Greedy Algorithm) হলো এমন একটি প্রবলেম সল্ভিং পদ্ধতি যা প্রতিটি ধাপে বর্তমানে সবচেয়ে লাভজনক বা "সেরা" ডিসিশনটি নেয়। এটি ভবিষ্যতের কথা চিন্তা না করে শুধুমাত্র বর্তমান মুহূর্তের গ্লোবাল অপ্টিমাম (Global Optimum) অর্জনের চেষ্টা করে। Software Company ইন্টারভিউতে গ্রিডি অ্যালগরিদম খুবই জনপ্রিয় কারণ এটি সাধারণত খুব দ্রুত এবং মেমরি সাশ্রয়ী হয়।

সহজ কথায়, গ্রিডি হলো "লোভী" হওয়া। আপনি যদি একটি বাক্সে অনেকগুলো টাকা ভরতে চান এবং আপনাকে বলা হয় যত বেশি সম্ভব টাকা নিতে, আপনি প্রথমে সবচেয়ে বড় নোটগুলো আগে নেবেন। এটিই গ্রিডি অ্যাপ্রোচ।

## ২. কেন গ্রিডি অ্যালগরিদম? (Speed and Efficiency)
অনেক সমস্যা আছে যা ডাইনামিক প্রোগ্রামিং (DP) দিয়ে সমাধান করা যায়, কিন্তু গ্রিডি ব্যবহার করলে সেটি অনেক কম সময়ে করা সম্ভব।
- **সুবিধা:** ইমপ্লিমেন্ট করা সহজ এবং টাইম কমপ্লেক্সিটি অনেক কম (সাধারণত $O(N)$ বা $O(N \log N)$)।
- **সীমাবদ্ধতা:** এটি সব প্রবলেমের জন্য কাজ করে না। শুধুমাত্র তখনই কাজ করে যখন স্থানীয়ভাবে সেরা সিদ্ধান্তটি গ্লোবাল সেরা সিদ্ধান্তের দিকে নিয়ে যায় (Greedy Choice Property)।

---

## ৩. গ্রিডি অ্যালগরিদমের শর্ত (Properties)
একটি প্রবলেম গ্রিডি দিয়ে সমাধান করার জন্য দুটি শর্ত থাকতে হবে:
১. **Greedy Choice Property:** বর্তমান ধাপে সেরা সিদ্ধান্তটি নিলে সেটি ফাইনাল সেরা সমাধানে পৌঁছাতে সাহায্য করবে।
২. **Optimal Substructure:** একটি বড় সমস্যার সমাধান তার ছোট ছোট সাব-প্রবলেমের সেরা সমাধানের সমন্বয়ে তৈরি হয়।

## ৪. রিয়েল-ওয়ার্ল্ড Software Company উদাহরণ (Real-World Case Studies)

### ৪.১. গুগল ম্যাপস - হাফম্যান কোডিং (Data Compression)
গুগল যখন বিশাল পরিমাণ ডেটা ট্রান্সফার করে, তখন তারা 'Huffman Coding' ব্যবহার করে ডেটা কম্প্রেস করে। এটি একটি গ্রিডি অ্যালগরিদম যা সবচেয়ে বেশিবার আসা ক্যারেক্টারকে সবচেয়ে ছোট কোড দেয়। এটি ব্যান্ডউইথ এবং স্টোরেজ সাশ্রয় করে।

### ৪.২. আমাজন - ডেলিভারি রুট অপ্টিমাইজেশন
আমাজনের ডেলিভারি ড্রাইভাররা যখন অনেকগুলো পার্সেল ডেলিভারি করে, তখন তাদের রুট প্ল্যানিংয়ে গ্রিডি লজিক ব্যবহার করা হয়। এটি প্রতিটি ধাপে সবচেয়ে কাছের ডেলিভারি পয়েন্টটি সিলেক্ট করে যাতে সময় এবং তেল সাশ্রয় হয়।

### ৪.৩. স্পটিফাই এবং ইউটিউব - ক্যাশ ম্যানেজমেন্ট (Caching)
সিস্টেমগুলো যখন ডিসাইড করে কোন গান বা ভিডিও ক্যাশ (Cache) থেকে মুছে ফেলবে, তখন তারা 'Least Recently Used' (LRU) বা গ্রিডি-ভিত্তিক পলিসি ব্যবহার করে। এটি নিশ্চিত করে যে সবচেয়ে গুরুত্বপূর্ণ ডেটাগুলো মেমরিতে থাকে।

## ৫. গ্রিডি বনাম ডাইনামিক প্রোগ্রামিং
ইন্টারভিউতে এটি সবচেয়ে কনফিউজিং পার্ট:
- **Greedy:** প্রতিটি ধাপে একটি সিদ্ধান্ত নেয় এবং আর পেছনে ফিরে তাকায় না।
- **DP:** সব ধরনের সম্ভাবনা চেক করে এবং সেরাটি বেছে নেয়। এটি মেমরি বেশি ব্যবহার করে কিন্তু গ্যারান্টি দেয় যে রেজাল্ট সঠিক হবে।

## ৬. ইন্টারভিউ চেকলিস্ট (Interview Checklist)
১. **Sorting:** গ্রিডি প্রবলেমের ৯০% ক্ষেত্রে শুরুতে সর্টিং করতে হয় (যেমন: ফিনিশ টাইম বা ভ্যালু অনুযায়ী)।
২. **Counter Example:** যদি আপনি গ্রিডি ট্রাই করতে চান, তবে নিজেকে জিজ্ঞেস করুন—এমন কোনো কেস কি আছে যেখানে এই লজিকটি ফেল করবে? (যেমন: Coin Change problem with custom denominations)।
৩. **Priority Queue:** অনেক সময় বর্তমানে সেরাটি খুঁজে বের করতে ‘হিপ’ (Heap/Priority Queue) ব্যবহার করা হয়।
