# Bit Manipulation (বিট ম্যানিপুলেশন) 🔟

## Bit Manipulation কী?
কম্পিউটার ডেটা সেভ করে 0 এবং 1 (Bits) আকারে। বিট ম্যানিপুলেশন হলো সরাসরি এই 0 এবং 1 গুলোকে নিয়ে কাজ করা। এটি সাধারণ গাণিতিক অপারেশনের (যেমন যোগ, গুণ, ভাগ) চেয়ে অনেক বেশি ফাস্ট (Fast) এবং কম মেমোরি ব্যবহার করে।

## Common Bitwise Operators
১. **AND (`&`):** দুটি বিটই 1 হলে রেজাল্ট 1 হবে, নাহলে 0। (Masking এর জন্য ব্যবহৃত হয়)।
২. **OR (`|`):** যেকোনো একটি বিট 1 হলেই রেজাল্ট 1। (বিট সেট বা অন করার জন্য ব্যবহৃত হয়)।
৩. **XOR (`^`):** দুটি বিট আলাদা হলে রেজাল্ট 1, একরকম হলে 0। (Toggle বা জোড়া খোঁজার জন্য ব্যবহৃত হয়)।
৪. **NOT (`~`):** 0 কে 1 এবং 1 কে 0 করে দেয় (Invert)।
৫. **Left Shift (`<<`):** বিটগুলোকে বাম দিকে সরিয়ে দেয়। $x \ll 1$ মানে হলো $x$ কে 2 দিয়ে গুণ করা।
৬. **Right Shift (`>>`):** বিটগুলোকে ডান দিকে সরিয়ে দেয়। $x \gg 1$ মানে হলো $x$ কে 2 দিয়ে ভাগ করা।

## Magic Bit Tricks (ম্যাজিক ট্রিকস)

১. **জোড়/বিজোড় চেক করা (Even/Odd):**
   - `if ((n & 1) == 0)` -> Even (জোড়)
   - `if ((n & 1) == 1)` -> Odd (বিজোড়)

২. **XOR এর জাদুকরী প্রপার্টি:**
   - কোনো সংখ্যার সাথে তার নিজের XOR করলে 0 হয়: `a ^ a = 0`
   - 0 এর সাথে কোনো সংখ্যার XOR করলে ওই সংখ্যাই হয়: `a ^ 0 = a`
   - এই ট্রিকটি ব্যবহার করে Array থেকে Single/Missing নাম্বার বের করা যায়।

৩. **সবচেয়ে ডানের (Right-most) 1-বিট রিমুভ করা:**
   - `n = n & (n - 1)`
   - এটি দিয়ে খুব দ্রুত বের করা যায় একটি সংখ্যাতে কতগুলো 1 আছে।

৪. **Power of 2 চেক করা:**
   - যদি `(n > 0)` এবং `(n & (n - 1)) == 0` হয়, তবে সংখ্যাটি 2 এর পাওয়ার (Power of 2)।

পরবর্তী ফাইলগুলোতে আমরা এই ট্রিকসগুলোর ইমপ্লিমেন্টেশন দেখব।
