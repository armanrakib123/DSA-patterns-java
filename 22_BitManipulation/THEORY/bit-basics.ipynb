# Bit Manipulation Basics (বিট ম্যানিপুলেশন বেসিকস)

## What is Bit Manipulation?
কম্পিউটার সবকিছুই ০ এবং ১ (Bits) হিসেবে স্টোর করে। Bit Manipulation হলো সরাসরি এই বিটগুলোর ওপর কাজ করার একটি পদ্ধতি। এটি অত্যন্ত ফাস্ট এবং মেমরি সাশ্রয়ী।

## Binary Representation
একটি সংখ্যা (যেমন ৫) বাইনারিতে হবে `101`। 
- **LSB (Least Significant Bit):** সবার ডানের বিটটি।
- **MSB (Most Significant Bit):** সবার বামের বিটটি।

## Core Operators (মূল অপারেটরসমূহ)
১. **AND (&):** দুটি বিটই ১ হলে রেজাল্ট ১।
   - `5 (101) & 3 (011) = 1 (001)`
২. **OR (|):** অন্তত একটি বিট ১ হলে রেজাল্ট ১।
   - `5 (101) | 3 (011) = 7 (111)`
৩. **XOR (^):** বিট দুটি আলাদা হলে ১, একই হলে ০। (সবচেয়ে ম্যাজিক্যাল অপারেটর!)
   - `5 (101) ^ 3 (011) = 6 (110)`
   - **XOR-এর বৈশিষ্ট্য:** $a \wedge a = 0$ এবং $a \wedge 0 = a$।
৪. **NOT (~):** বিট উল্টে দেয় (১ থাকলে ০, ০ থাকলে ১)।
৫. **Shift Operators:**
   - **Left Shift (<<):** বিট বামে সরায়। $x \ll 1$ মানে $x \times 2$।
   - **Right Shift (>>):** বিট ডানে সরায়। $x \gg 1$ মানে $x / 2$।



# Common Bit Tricks (কিছু প্রয়োজনীয় বিট ট্রিকস)

ইন্টারভিউতে এই ছোট ছোট ট্রিকগুলো আপনার অনেক সময় বাঁচিয়ে দিবে।

## 1. Check if a number is Even or Odd
আমরা সাধারণত `% 2` ব্যবহার করি, কিন্তু বিট দিয়ে এটি আরও ফাস্ট:
```java
if ((n & 1) == 0) // Even
else // Odd
```

## 2. Check if a number is Power of 2
একটি সংখ্যা যদি ২-এর পাওয়ার হয় (যেমন: ২, ৪, ৮, ১৬), তবে তার বাইনারিতে শুধুমাত্র একটি '১' থাকে।
```java
boolean isPowerOfTwo = (n > 0) && ((n & (n - 1)) == 0);
```

## 3. Clear the Lowest Set Bit
একটি সংখ্যার বাইনারিতে থাকা একদম ডানের '১'-টিকে ০ করে দেওয়া:
```java
n = n & (n - 1);
```
এটি ব্যবহার করে একটি সংখ্যায় কয়টি '১' আছে (Hamming Weight) তা খুব দ্রুত বের করা যায়।

## 4. Get the i-th Bit
$i$ তম বিটটি ১ কি না তা চেক করা:
```java
boolean isSet = (n & (1 << i)) != 0;
```

## 5. Swap two numbers without extra variable
```java
a = a ^ b;
b = a ^ b;
a = a ^ b;
```
