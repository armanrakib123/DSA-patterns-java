# Binary Search (বাইনারি সার্চ) 🔍

## Binary Search কী?
Binary Search হলো এমন একটি সার্চিং অ্যালগরিদম যা একটি **সর্টেড (Sorted)** ডেটা স্ট্রাকচার (যেমন Array) থেকে কোনো ইলিমেন্ট খুব দ্রুত খুঁজে বের করতে সাহায্য করে। এটি প্রতি ধাপে সার্চ স্পেসকে (যেখানে খুঁজছি) অর্ধেক করে ফেলে।

- **Linear Search (লিনিয়ার সার্চ):** একপাশ থেকে একটা একটা করে খুঁজা ($O(N)$)।
- **Binary Search:** মাঝখানে চেক করা। যদি কাঙ্ক্ষিত ভ্যালু মাঝখানের চেয়ে ছোট হয়, তবে বাম দিকে খোঁজা, আর বড় হলে ডান দিকে খোঁজা ($O(\log N)$)।

*উদাহরন:* ১ থেকে ১০০ পর্যন্ত একটি সংখ্যা গেস করতে বললে, আপনি কি ১, ২, ৩ করে গুনে গুনে খুঁজবেন? না! আপনি প্রথমে ৫০ গেস করবেন। যদি উত্তর "ছোট" হয়, তবে এরপর ২৫ গেস করবেন। এটাই বাইনারি সার্চ!

## কখন বুঝবেন যে Binary Search ব্যবহার করতে হবে? (Pattern Recognition)
- ১. **Sorted Array:** প্রবলেমে যদি বলা থাকে Array টি "Sorted" বা "Ascending/Descending order" এ আছে, এবং কিছু খুঁজতে বলা হয়।
- ২. **Time Complexity Constraint:** যদি প্রবলেমের রিকোয়ারমেন্টে বলা থাকে যে $O(\log N)$ এ সলভ করতে হবে (যেমন LC 33: Search in Rotated Sorted Array)।
- ৩. **Finding a Boundary / Threshold:** যখন বলা হয় এমন একটি "মিনিমাম" বা "ম্যাক্সিমাম" ভ্যালু বের করতে যা একটি নির্দিষ্ট কন্ডিশন পূরণ করে (এটিকে বলা হয় Binary Search on Answer)।
- ৪. **Monotonic Property:** যদি কোনো ফাংশন বা গ্রাফ সবসময় বাড়তে থাকে (Monotonically Increasing) বা কমতে থাকে, সেখানেও বাইনারি সার্চ কাজ করে।

## তিনটি প্রধান ভ্যারিয়েশন (Variations)

১. **Classic Binary Search:** 
   সরাসরি একটি টার্গেট ভ্যালু খোঁজা। (যেমন Array তে $5$ আছে কি না)।

২. **Lower Bound / Upper Bound (খুবই গুরুত্বপূর্ণ):**
   - **Lower Bound:** টার্গেটের সমান বা তার চেয়ে বড় প্রথম ইলিমেন্টের ইনডেক্স। (First occurrence / First position)।
   - **Upper Bound:** টার্গেটের চেয়ে বড় প্রথম ইলিমেন্টের ইনডেক্স। (Last occurrence বের করতে এর সাহায্য নেওয়া হয়)।
   
৩. **Binary Search on Answer (অ্যাডভান্সড):**
   এখানে Array দেওয়া থাকে না। বরং একটি টার্গেট റേঞ্জ (Range) দেওয়া থাকে, এবং আমরা একটি উত্তর গেস (Guess) করি। এরপর একটি `isValid(guess)` ফাংশন লিখে চেক করি ওই উত্তরটি সম্ভব কি না। সম্ভব হলে আমরা আরও ভালো উত্তরের (Minimum/Maximum) দিকে এগোই। (যেমন: Koko Eating Bananas - LC 875)।

পরবর্তী ফাইলগুলোতে আমরা এই ভ্যারিয়েশনগুলোর টেমপ্লেট এবং উদাহরণ দেখব।


# Binary Search (বাইনারি সার্চ) - The Software Company Standard Theory 🔍

## ১. ভূমিকা (Detailed Introduction)
বাইনারি সার্চ (Binary Search) হলো অ্যালগরিদম জগতের সবচেয়ে শক্তিশালী এবং দক্ষ সার্চিং টেকনিক। এটি মূলত একটি সর্টেড (Sorted) ডেটা সেটকে বারবার দুই ভাগে ভাগ করে কাঙ্ক্ষিত ফলাফল খুঁজে বের করে। Software Company ইন্টারভিউতে বাইনারি সার্চ অত্যন্ত গুরুত্বপূর্ণ কারণ এটি $O(N)$ এর প্রবলেমকে $O(\log N)$ এ নামিয়ে আনতে পারে।

কল্পনা করুন আপনি ১ থেকে ১০০ এর মধ্যে একটি সংখ্যা খুঁজছেন। আপনি যদি এক এক করে জিজ্ঞেস করেন তবে আপনার ১০০ বার পর্যন্ত লাগতে পারে। কিন্তু বাইনারি সার্চ ব্যবহার করলে আপনি মাত্র ৭টি প্রশ্নে সংখ্যাটি খুঁজে পাবেন। এই "Divide and Conquer" পদ্ধতিটিই বাইনারি সার্চের মূল শক্তি।

---

## ২. কেন বাইনারি সার্চ? (The Logarithmic Power)
ইন্টারনেট স্কেলে যখন ডেটা কোটি কোটি হয়, তখন লিনিয়ার সার্চ (Linear Search) কাজ করে না।
- **লিনিয়ার সার্চ:** ১ বিলিয়ন ডেটার জন্য ১ বিলিয়ন অপারেশন লাগে।
- **বাইনারি সার্চ:** ১ বিলিয়ন ডেটার জন্য মাত্র ৩০টি অপারেশন লাগে! ($2^{30} \approx 1$ Billion)।

এই অকল্পনীয় গতিই বাইনারি সার্চকে মডার্ন কম্পিউটিংয়ের ভিত্তি বানিয়ে দিয়েছে।

---

## ৩. বাইনারি সার্চের শর্ত (The Prerequisite)
বাইনারি সার্চ ব্যবহারের প্রধান শর্ত হলো ডেটা সর্টেড থাকতে হবে। কিন্তু ইন্টারভিউতে অনেক সময় ডেটা সরাসরি সর্টেড থাকে না। সেক্ষেত্রে আমাদের "Search Space" সর্টেড কি না তা খুঁজে বের করতে হয়। যদি কোনো কন্ডিশন এমন হয় যে সেটি একটি পয়েন্ট পর্যন্ত 'False' এবং তারপর সব 'True' (যেমন: $FFFFTTTT$), তবেই সেখানে বাইনারি সার্চ ব্যবহার করা সম্ভব।

## ৪. রিয়েল-ওয়ার্ল্ড Software Company উদাহরণ (Real-World Case Studies)

### ৪.১. গুগল সার্চ - ইন্ডেক্সিং (Indexing)
গুগল যখন আপনার সার্চ কোয়েরির উত্তর দেয়, তারা বিলিয়ন বিলিয়ন ওয়েব পেজের ইন্ডেক্সের ওপর বাইনারি সার্চের মতো এডভান্সড টেকনিক ব্যবহার করে। এটি তাদের মিলি-সেকেন্ডের মধ্যে রেজাল্ট দেখাতে সাহায্য করে।

### ৪.২. আমাজন - শিপিং লজিস্টিক্স (Logistics)
আমাজনের ওয়্যারহাউসে যখন কোনো আইটেম খোঁজা হয়, তাদের ইনভেন্টরি ম্যানেজমেন্ট সিস্টেম সর্টেড আইটেম লিস্টের ওপর বাইনারি সার্চ চালিয়ে কয়েক সেকেন্ডের মধ্যে আইটেমটির সঠিক লোকেশন বলে দেয়।

### ৪.৩. সফটওয়্যার ডিবাগিং - গিট বাইসেক্ট (Git Bisect)
ডেভেলপাররা যখন কোডে কোনো বাগ (Bug) খুঁজে পায় এবং জানতে চায় কোন কমিটে (Commit) এই বাগটি এসেছে, তখন তারা `git bisect` কমান্ড ব্যবহার করে। এটি শত শত কমিটের ওপর বাইনারি সার্চ চালিয়ে মাত্র কয়েকটি ট্রায়ালে বাগি কমিটটি খুঁজে বের করে।

## ৫. অ্যাডভান্সড বাইনারি সার্চ প্যাটার্নস (Search Space)
ইন্টারভিউতে বাইনারি সার্চ শুধু অ্যারেতে সার্চ করার জন্য আসে না। এটি মূলত দুই ধরনের হয়:
- ১. **Binary Search on Index:** সর্টেড অ্যারেতে ইনডেক্স খোঁজা।
- ২. **Binary Search on Answer:** যখন সরাসরি অ্যারে নেই, কিন্তু আমরা জানি যে আমাদের উত্তরটি একটি নির্দিষ্ট রেঞ্জের (Search Space) মধ্যে আছে। (যেমন: Kokane Eating Bananas, Capacity to Ship Packages)।

## ৬. ইন্টারভিউ চেকলিস্ট (Interview Checklist)
- ১. **Overlow Prevention:** `mid = low + (high - low) / 2` ব্যবহার করুন, `(low + high) / 2` নয়।
- ২. **Infinite Loop:** লুপ কন্ডিশন `low <= high` নাকি `low < high` তা প্রবলেম অনুযায়ী ঠিক করুন।
- ৩. **Upper/Lower Bound:** আপনি কি ফার্স্ট অকারেন্স খুঁজছেন নাকি লাস্ট? এটি বাইনারি সার্চের একটি কমন ভ্যারিয়েশন।

## ৭. উপসংহার (Conclusion)
বাইনারি সার্চ হলো অপ্টিমাইজেশনের একটি মাস্টারপিস। এটি আয়ত্ত করলে আপনি বুঝতে পারবেন কিভাবে বিশাল পরিমাণ ডেটা খুব সামান্য সময়ে প্রসেস করা সম্ভব। পরবর্তী ফাইলে আমরা এর বিভিন্ন টেমপ্লেট এবং লেটকোড উদাহরণ দেখব।
