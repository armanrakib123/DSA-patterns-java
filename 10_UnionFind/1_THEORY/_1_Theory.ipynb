# 🔗 Pattern 41: Union-Find / DSU (ডিসজয়েন্ট সেট ইউনিয়ন)

ডিসজয়েন্ট সেট ইউনিয়ন (Disjoint Set Union) বা সংক্ষেপে DSU হলো গ্রাফ থিওরির অন্যতম শক্তিশালী ডেটা স্ট্রাকচার। এটি মূলত কতগুলো এলিমেন্টকে আলাদা আলাদা সেটে ভাগ করা এবং দুটি সেটকে একত্র করার কাজে ব্যবহৃত হয়।

---

## 🧠 DSU এর মূল দর্শন (Core Philosophy)

কল্পনা করুন একটি পার্টিতে অনেক মানুষ আছে। শুরুতে সবাই একা। যখন দুইজন বন্ধু হয়, তারা একটি ছোট গ্রুপ তৈরি করে। DSU আমাদের সাহায্য করে জানতে যে—"অমুক ব্যক্তি কি অমুক গ্রুপের সদস্য?" অথবা "এই দুইজন কি একই গ্রুপে আছে?"

### প্রধান অপারেশনসমূহ:
- ১. **Find:** একটি এলিমেন্ট কোন গ্রুপের (বা সেটের) লিডার বা রিপ্রেজেন্টেটিভ, তা খুঁজে বের করে।
- ২. **Union:** দুটি আলাদা গ্রুপকে একত্র করে একটি বড় গ্রুপ বানায়।

---

## 🚀 অপ্টিমাইজেশন টেকনিক (ইন্টারভিউয়ের জন্য অত্যন্ত জরুরি)

DSU কে সুপারফাস্ট করার জন্য দুটি টেকনিক ব্যবহার করা হয়:

### ১. পাথ কম্প্রেশন (Path Compression)
যখন আপনি `find` অপারেশন চালান, তখন আপনি সরাসরি সব নোডকে তাদের গ্র্যান্ড-প্যারেন্টের সাথে কানেক্ট করে দেন। এর ফলে পরবর্তী সময়ে সার্চিং ও(১) এর কাছাকাছি হয়।

### ২. ইউনিয়ন বাই র‍্যাঙ্ক বা সাইজ (Union by Rank/Size)
দুটি সেট একত্র করার সময় আমরা সবসময় ছোট সেটটিকে বড় সেটের নিচে যুক্ত করি। এটি ট্রির উচ্চতা (Height) বাড়তে দেয় না।

---

## 🛠️ কেন DSU ইন্টারভিউতে গুরুত্বপূর্ণ?

- ১. **Connectivity:** গ্রাফের দুটি নোড কানেক্টেড কি না তা ও(১) সময়ে চেক করার জন্য।
- ২. **Cycle Detection:** আনডিরেক্টেড গ্রাফে সাইকেল ডিটেকশনের জন্য এটিই সেরা পদ্ধতি।
- ৩. **Dynamic Grouping:** যখন ডাটা ডাইনামিক্যালি গ্রুপ করতে হয় (যেমন- Accounts Merge).

---

## 🚧 DSU টেমপ্লেট (Java)

```java
class DSU {
    int[] parent;
    int[] rank;

    public DSU(int n) {
        parent = new int[n];
        rank = new int[n];
        for (int i = 0; i < n; i++) parent[i] = i;
    }

    public int find(int i) {
        if (parent[i] == i) return i;
        return parent[i] = find(parent[i]); // Path Compression
    }

    public void union(int i, int j) {
        int rootI = find(i);
        int rootJ = find(j);
        if (rootI != rootJ) {
            // Union by Rank
            if (rank[rootI] < rank[rootJ]) parent[rootI] = rootJ;
            else if (rank[rootI] > rank[rootJ]) parent[rootJ] = rootI;
            else {
                parent[rootI] = rootJ;
                rank[rootJ]++;
            }
        }
    }
}
```

# Union Find (Disjoint Set) 🔗

## Union Find কী?
Union Find (যাকে **Disjoint Set Union বা DSU** বলা হয়) হলো একটি ডেটা স্ট্রাকচার যা মূলত অনেকগুলো আলাদা আলাদা গ্রুপ (বা Set) কে ট্র্যাক করতে এবং তাদেরকে যুক্ত (Union) করতে ব্যবহৃত হয়।

এটি মূলত দুটি প্রধান প্রশ্নের উত্তর খুব দ্রুত ($O(\alpha(N)) \approx O(1)$) দিতে পারে:
- ১. **Find:** একটি ইলিমেন্ট কোন গ্রুপের অন্তর্ভুক্ত? (Who is the boss?)
- ২. **Union:** দুটি আলাদা গ্রুপকে একসাথে মিলিয়ে একটি গ্রুপ করা।

## কখন বুঝবেন যে Union Find ব্যবহার করতে হবে? (Pattern Recognition)
- ১. **Dynamic Connectivity:** যখন আপনাকে ডাইনামিকালি দুটি নোড কানেক্ট করতে বলা হয় এবং জানতে চাওয়া হয় তারা কানেক্টেড কি না।
- ২. **Connected Components:** গ্রাফে কতগুলো আলাদা আলাদা গ্রুপ বা আইল্যান্ড (Island) আছে তা বের করা (DFS/BFS দিয়েও করা যায়, তবে DSU অনেক সময় ফাস্ট ও ক্লিন)।
- ৩. **Cycle Detection (Undirected Graph):** আনডাইরেক্টেড গ্রাফে সাইকেল আছে কি না তা বের করতে। (যদি দুটি নোড আগে থেকেই একই গ্রুপে থাকে এবং তাদের মধ্যে নতুন করে এজ (Edge) দেওয়া হয়, তার মানে সাইকেল তৈরি হলো!)
- ৪. **Kruskal's Algorithm:** Minimum Spanning Tree (MST) তৈরি করতে।

## কীভাবে কাজ করে? (The Boss Concept)
কল্পনা করুন অনেকগুলো মানুষ আলাদা আলাদা দাঁড়িয়ে আছে।
- শুরুতে সবাই নিজের নিজের বস (Parent is self)। `parent[i] = i`
- **Union(A, B):** A এবং B হাত মেলাল। এখন আমরা ঠিক করব এদের মধ্যে বস কে হবে। ধরি, B কে বস বানালাম। তাহলে `parent[A] = B`।
- **Find(A):** A এর বস কে? আমরা `parent[A]` তে যাব। যদি সে নিজেই নিজের বস না হয়, তবে তার বসের কাছে যাব। এভাবে পরম বস (Root/Ultimate Boss) বের করব।

## দুটি জাদুকরী অপ্টিমাইজেশন (Optimizations)
বেসিক DSU তে একটি সমস্যা আছে। যদি A এর বস B, B এর বস C, C এর বস D হয়, তবে A এর পরম বস খুঁজতে $O(N)$ সময় লাগতে পারে (Linked List এর মতো)। এটিকে $O(1)$ এ নামিয়ে আনতে দুটি অপ্টিমাইজেশন ব্যবহার করা হয়:

- ১. **Path Compression (পাথ কম্প্রেশন - Find এর সময়):** 
   যখন আমরা A এর পরম বস D কে খুঁজে পাই, তখন আমরা সরাসরি A, B এবং C সবার বস হিসেবে D কে সেট করে দিই। এতে পরবর্তীতে খোঁজার সময় সরাসরি D কে পাওয়া যায়।

- ২. **Union by Rank / Size (ইউনিয়ন বাই র‍্যাঙ্ক - Union এর সময়):**
   যখন দুটি গ্রুপের মধ্যে ইউনিয়ন করা হয়, তখন ছোট গ্রুপটিকে বড় গ্রুপের নিচে যুক্ত করা হয়। এতে ট্রির গভীরতা (Depth) বাড়ে না।

পরবর্তী ফাইলগুলোতে আমরা এই অপ্টিমাইজেশনগুলোর ইমপ্লিমেন্টেশন দেখব।


# Union Find (ইউনিয়ন ফাইন্ড) - The FAANG Standard Theory 🕸️

## ১. ভূমিকা (Detailed Introduction)
ইউনিয়ন ফাইন্ড (Union Find), যা ডিসজয়েন্ট সেট ইউনিয়ন (Disjoint Set Union বা DSU) নামেও পরিচিত, একটি অত্যন্ত দক্ষ ডেটা স্ট্রাকচার। এটি মূলত একগুচ্ছ ইলিমেন্টকে আলাদা আলাদা গ্রুপে (Sets) ভাগ করে এবং তারা একই গ্রুপে আছে কি না তা খুব দ্রুত চেক করতে সাহায্য করে। FAANG ইন্টারভিউতে গ্রাফের কানেক্টিভিটি (Connectivity) এবং ডাইনামিক গ্রুপ ম্যানেজমেন্টের জন্য এটি একটি মাস্ট-হ্যাভ টুল।

সহজ কথায়, ইউনিয়ন ফাইন্ড হলো "পরিবার" বা "গোষ্ঠী" ম্যানেজ করার মতো। আপনি যদি জানতে চান দুইজন ব্যক্তি একই বংশের কি না, তবে আপনি তাদের "গোষ্ঠী প্রধান" (Representative) চেক করবেন। ইউনিয়ন ফাইন্ড ঠিক এই কাজটিই কম্পিউটার সায়েন্সে করে।

---

## ২. কেন ইউনিয়ন ফাইন্ড? (The Power of Constant Time)
সাধারণ গ্রাফ ট্রাভার্সাল (DFS/BFS) ব্যবহার করে দুইজন নোড কানেক্টেড কি না তা বের করতে $O(V+E)$ সময় লাগে। কিন্তু ইউনিয়ন ফাইন্ড অপ্টিমাইজেশন (Path Compression + Union by Rank) ব্যবহার করলে এটি প্রায় ও(১) বা "Amortized Constant Time" এ করা সম্ভব।

এটি কোটি কোটি কানেকশন হ্যান্ডেল করার জন্য অকল্পনীয় গতি প্রদান করে।

---

## ৩. মূল অপারেশন দুটি (Core Operations)
ইউনিয়ন ফাইন্ডের দুটি প্রধান কাজ আছে:
- ১. **Find:** একটি ইলিমেন্ট কোন গ্রুপের সদস্য তা খুঁজে বের করা। এটি ওই গ্রুপের "প্যারেন্ট" বা "রিপ্রেজেন্টেটিভ" রিটার্ন করে।
- ২. **Union:** দুটি আলাদা গ্রুপকে একীভূত করে একটি বড় গ্রুপ তৈরি করা।

---

## ৪. অপ্টিমাইজেশন টেকনিক (The Magic of DSU)
সাধারণ ইউনিয়ন ফাইন্ড অনেক স্লো হতে পারে (Worst case $O(N)$)। একে ফাস্ট করার জন্য দুটি টেকনিক ব্যবহার করা হয়:

- ১. **Path Compression (পথ সংকোচন):** যখন আমরা `find` করি, তখন প্রতিটি নোডকে সরাসরি তার গ্রুপ লিডারের সাথে কানেক্ট করে দিই। এতে পরের বার সার্চ করলে মাত্র এক ধাপেই রেজাল্ট পাওয়া যায়।
- ২. **Union by Rank/Size:** ছোট গ্রুপকে সবসময় বড় গ্রুপের নিচে যুক্ত করা হয় যাতে ট্রির হাইট বা গভীরতা না বাড়ে।

এই দুটি একসাথে ব্যবহার করলে টাইম কমপ্লেক্সিটি হয় $O(\alpha(N))$, যেখানে $\alpha$ হলো ইনভার্স অ্যাকারম্যান ফাংশন, যা প্র্যাকটিক্যালি ৪ বা ৫ এর বেশি হয় না।

## ৬. ইন্টারভিউ চেকলিস্ট (Interview Checklist)
১. **Initialization:** শুরুতে প্রতিটি নোড নিজের প্যারেন্ট নিজেই।
২. **Cycle Detection:** যদি দুইজন নোডের প্যারেন্ট আগে থেকেই একই হয় এবং তাদের মধ্যে নতুন এজ যুক্ত করা হয়, তবে সেখানে একটি সাইকেল তৈরি হবে।
৩. **Path Compression:** `parent[x] = find(parent[x])` - এই একটি লাইন আপনার কোডকে ১০০ গুণ ফাস্ট করে দিতে পারে।
