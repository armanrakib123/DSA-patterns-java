# 🔙 Pattern 23: Backtracking (ব্যাকট্র্যাকিং প্যাটার্ন)

ব্যাকট্র্যাকিং (Backtracking) হলো একটি ব্রুট-ফোর্স (Brute-force) বা সব সম্ভাব্য অপশন ট্রাই করার একটি পদ্ধতি, যা রিকার্সনের মাধ্যমে কাজ করে। এর বিশেষত্ব হলো, এটি যখন কোনো ভুল পথে (যে পথে সমাধান নেই) চলে যায়, তখন সেটি আবার আগের অবস্থায় ফিরে আসে (Backtrack) এবং অন্য কোনো পথ ট্রাই করে।

---

## 🧠 ইন্টুইশন (Intuition)

ব্যাকট্র্যাকিংকে আপনি একটি গোলকধাঁধার (Maze) সাথে তুলনা করতে পারেন। 
- আপনি একটি পথে হাঁটতে শুরু করলেন। 
- যদি দেখেন পথটি বন্ধ (Dead end), তবে আপনি আবার আগের মোড়ে ফিরে আসবেন। 
- এরপর অন্য একটি পথে যাওয়ার চেষ্টা করবেন। 

**মূল মন্ত্র:** "Explore all possibilities, but stop as soon as you find a path is invalid."

---

## 🚀 কেন এটি ব্যবহার করবেন?

ব্যাকট্র্যাকিং মূলত ব্যবহার করা হয় যখন আমাদের সব সম্ভাব্য সমাধান (Combinations, Permutations, Subsets) বের করতে হয় বা কোনো নির্দিষ্ট শর্ত পূরণ করে এমন সব পথ খুঁজতে হয়।

উদাহরণ:
- **N-Queens:** দাবা বোর্ডে এন-সংখ্যক কুইন বসানো যেন কেউ কাউকে আক্রমণ না করে।
- **Sudoku Solver:** সুডোকু পাজল সলভ করা।
- **Permutations:** একটি স্ট্রিং বা অ্যারের সবরকম সাজানো ব্যবস্থা বের করা।
- **Subset Sum:** একটি সেটের কোন কোন উপাদানের যোগফল নির্দিষ্ট একটি সংখ্যার সমান।

---

## 🛠️ ব্যাকট্র্যাকিং টেমপ্লেট (The Standard Template)

```java
public void backtrack(state, result, current) {
    if (is_solution(state)) {
        result.add(new ArrayList<>(current));
        return;
    }

    for (option : all_options) {
        if (is_valid(option, state)) {
            make_choice(option, state, current);

            backtrack(state, result, current);

            undo_choice(option, state, current);
        }
    }
}
```

---

## 🚨 ব্যাকট্র্যাকিং বনাম রিকার্সন

- **Recursion:** একটি সাধারণ টেকনিক যেখানে ফাংশন নিজেকে কল করে।
- **Backtracking:** রিকার্সনের একটি প্রয়োগ যেখানে আমরা "স্টেট" (State) পরিবর্তন করি এবং রিকার্সিভ কল শেষে সেই স্টেটকে আবার আগের অবস্থায় ফিরিয়ে নিয়ে আসি।

---

## 🏆 এই ফোল্ডারের প্রবলেমগুলো
- ১. **Permutations.java (Medium):** একটি অ্যারের সবরকম বিন্যাস বের করা। (O(N!) Time).
- ২. **Subsets.java (Medium):** একটি সেটের সব সম্ভাব্য উপসেট (Power set) বের করা।
- ৩. **NQueens.java (Hard):** দাবা বোর্ডে কুইন বসানোর ক্লাসিক প্রবলেম। (ইন্টারভিউয়ের বস প্রবলেম)।

প্রথমে নিজে নিজে ট্রাই করুন, না পারলে কোড এবং কমেন্টগুলো পড়ুন!


# Backtracking (ব্যাকট্র্যাকিং) 🔙

## Backtracking কী?
ব্যাকট্র্যাকিং (Backtracking) হলো একটি অ্যালগরিদম ডিজাইন টেকনিক যা প্রবলেমের সব সম্ভাব্য সমাধান (all possible solutions) খুঁজে বের করতে ব্যবহৃত হয়। এটি মূলত **Depth First Search (DFS)** এর একটি ভ্যারিয়েশন।

এটি ধাপে ধাপে (step-by-step) এগোয়। যদি কোনো ধাপে গিয়ে মনে হয় যে "এই পথে গেলে আর সমাধান পাওয়া সম্ভব নয়" (constraint violation), তখন এটি সেই পথ থেকে **পিছিয়ে আসে (backtracks)** এবং অন্য পথে চেষ্টা করে।

*উদাহরন:* গোলকধাঁধায় (Maze) পথ খোঁজা। আপনি একটি পথে গেলেন, গিয়ে দেখলেন সামনে দেয়াল। তখন আপনি কী করবেন? আপনি পেছনের মোড়ে ফিরে আসবেন (backtrack) এবং অন্য পথে যাওয়ার চেষ্টা করবেন।

## DFS এবং Backtracking এর মধ্যে পার্থক্য কী?
- **DFS:** একটি গ্রাফ বা ট্রির সব নোড ভিজিট করা।
- **Backtracking:** এটিও DFS ব্যবহার করে, কিন্তু সব নোড ভিজিট করে না। যদি কোনো নোডে গিয়ে মনে হয় এটি দিয়ে কাজ হবে না, তবে সেটিকে Prune (ছেঁটে ফেলা) করে দেয় এবং ফিরে আসে (Un-choose)।

## কখন বুঝবেন যে Backtracking ব্যবহার করতে হবে? (Pattern Recognition)
- ১. **All Combinations / Permutations:** যদি প্রবলেমে বলা হয় "Find all possible combinations", "Find all subsets", "Find all permutations"।
- ২. **Constraint Satisfaction:** সুডোকু (Sudoku), N-Queens, বা Maze এর মতো প্রবলেম যেখানে অনেকগুলো অপশন থাকে এবং নির্দিষ্ট কিছু রুলস মেনে সলভ করতে হয়।
- ৩. **Small Constraints:** Backtracking এর Time Complexity সাধারণত অনেক বেশি হয় ($O(2^N)$ বা $O(N!)$)। তাই LeetCode এ যদি দেখেন ইনপুট সাইজ খুবই ছোট ($N \le 20$), তাহলে বুঝবেন এখানে Backtracking লাগতে পারে!

## Backtracking এর তিনটি মূল ভিত্তি (3 Pillars)
ব্যাকট্র্যাকিং কোড লেখার সময় সবসময় তিনটি প্রশ্ন নিজেকে করবেন:
- ১. **Choice (পছন্দ):** এই মুহূর্তে আমার কাছে কী কী অপশন আছে?
- ২. **Constraints (শর্ত):** কোন কোন অপশন আমি নিতে পারব আর কোনটি নিতে পারব না?
- ৩. **Goal (লক্ষ্য):** আমি কখন থামব? (Base case কী?)

## The Magic Formula (Choose ➡️ Explore ➡️ Un-choose)
সব ব্যাকট্র্যাকিং প্রবলেম নিচের এই লজিকে কাজ করে:
- 1. বর্তমান ইলিমেন্টটিকে লিস্টে **যোগ করুন (Choose)**।
- 2. রিকার্সন কল করে সামনের দিকে **এগিয়ে যান (Explore)**।
- 3. রিকার্সন থেকে ফিরে আসার পর, ওই ইলিমেন্টটিকে লিস্ট থেকে **মুছে ফেলুন (Un-choose / Backtrack)**। (যাতে অন্য অপশন ট্রাই করা যায়)।

পরবর্তী ফাইলগুলোতে আমরা এর জেনেরিক টেমপ্লেট এবং উদাহরণ দেখব।


# Backtracking (ব্যাকট্র্যাকিং) - The FAANG Standard Theory ↩️

## ১. ভূমিকা (Detailed Introduction)
ব্যাকট্র্যাকিং (Backtracking) হলো একটি অত্যন্ত শক্তিশালী ব্রুট-ফোর্স (Brute-Force) অপ্টিমাইজেশন টেকনিক। এটি মূলত সব ধরনের সম্ভাব্য সমাধান (Possible Solutions) এক্সপ্লোর করার জন্য ব্যবহৃত হয়। কিন্তু সাধারণ ব্রুট ফোর্সের সাথে এর পার্থক্য হলো, ব্যাকট্র্যাকিং যখনই বুঝতে পারে যে বর্তমান পথটি সঠিক সমাধানের দিকে যাবে না, তখনই সে "ব্যাকট্র্যাক" করে বা এক ধাপ পিছিয়ে এসে অন্য পথ খুঁজে দেখে।

সহজ কথায়, ব্যাকট্র্যাকিং হলো একটি মেজ (Maze) বা গোলকধাঁধায় পথ খোঁজার মতো। আপনি একটি রাস্তায় হাঁটলেন, দেখলেন সেটি ডেড-এন্ড (Dead-end), আপনি ফিরে আসলেন এবং অন্য রাস্তায় ট্রাই করলেন। এই ফিরে আসার প্রক্রিয়াটিই হলো ব্যাকট্র্যাকিং।

---

## ২. কেন ব্যাকট্র্যাকিং? (The Decision Space)
ব্যাকট্র্যাকিং মূলত তখনই ব্যবহৃত হয় যখন আমাদের অনেকগুলো ডিসিশন নিতে হয় এবং প্রতিটি ডিসিশনের পর অনেকগুলো নতুন ডিসিশন তৈরি হয়।
- **ব্যবহার:** পারমিউটেশন (Permutations), কম্বিনেশন (Combinations), এন-কুইনস (N-Queens), সুডোকু সলভার (Sudoku Solver)।

এটি আমাদের কোডকে অনেক বেশি ক্লিন এবং রিকার্সিভলি (Recursively) সব পসিবিলিটি চেক করতে সাহায্য করে।

---

## ৩. ব্যাকট্র্যাকিংয়ের মূল তিনটি অংশ (Core Components)
প্রতিটি ব্যাকট্র্যাকিং অ্যালগরিদমে তিনটি প্রধান অংশ থাকে:
- ১. **Choice:** আপনি বর্তমান ধাপে কী ডিসিশন নিচ্ছেন? (যেমন: একটি সংখ্যা নেওয়া)।
- ২. **Constraint:** আপনার ডিসিশনটি কি বৈধ? (যেমন: সংখ্যাটি কি আগে নেওয়া হয়েছে?)।
- ৩. **Goal:** আপনি কি কাঙ্ক্ষিত সমাধানে পৌঁছেছেন? (যেমন: সব ইনডেক্স কি ফিল-আপ হয়েছে?)।

## ৪. রিয়েল-ওয়ার্ল্ড FAANG উদাহরণ (Real-World Case Studies)

### ৪.১. গুগল ম্যাপস - অপ্টিমাল রুট ফাইন্ডিং
যখন আপনি একটি গন্তব্য থেকে অন্য গন্তব্যে যাওয়ার রাস্তা খুঁজছেন, গুগল ম্যাপসের অ্যালগরিদম ব্যাকট্র্যাকিং এবং গ্রাফ ট্রাভার্সাল ব্যবহার করে সব সম্ভাব্য রাস্তা এক্সপ্লোর করে। যখনই সে দেখে কোনো একটি রাস্তা বন্ধ বা অনেক বেশি ট্রাফিক, সে ব্যাকট্র্যাক করে অন্য রাস্তা ট্রাই করে।

### ৪.২. আমাজন - প্যাকেজ প্যাকিং অপ্টিমাইজেশন
আমাজনের রোবটগুলো যখন একটি বক্সে বিভিন্ন সাইজের আইটেম প্যাক করে, তারা ব্যাকট্র্যাকিংয়ের মতো লজিক ব্যবহার করে। তারা একটি আইটেম রেখে চেক করে অন্যগুলো ফিট হবে কি না। যদি না হয়, তারা আইটেমটি সরিয়ে অন্য একটি আইটেম রেখে ট্রাই করে।

### ৪.৩. সাইবার সিকিউরিটি - পাসওয়ার্ড ক্র্যাকিং (Penetration Testing)
পাসওয়ার্ড ক্র্যাকিং বা সিকিউরিটি টেস্টিংয়ের সময় ব্যাকট্র্যাকিং ব্যবহার করে বিভিন্ন ক্যারেক্টার কম্বিনেশন চেক করা হয়। যদিও এটি টাইম কনজিউমিং, কিন্তু এটি সব সম্ভাব্য পাসওয়ার্ড জেনারেট করার একমাত্র উপায়।

---

## ৫. ব্যাকট্র্যাকিং বনাম রিকার্সন
অনেকেই এই দুটিকে গুলিয়ে ফেলেন:
- **রিকার্সন:** একটি ফাংশন নিজেকে কল করা। এটি একটি মাধ্যম।
- **ব্যাকট্র্যাকিং:** রিকার্সন ব্যবহার করে সব সম্ভাব্য সমাধান খোঁজা এবং অপ্রয়োজনীয় পথ বাদ দেওয়া। ব্যাকট্র্যাকিং হলো একটি 'Strategy'।

---

## ৬. ইন্টারভিউ চেকলিস্ট (Interview Checklist)
- ১. **Base Case:** রিকার্সন কখন থামবে তা অবশ্যই ডিফাইন করুন।
- ২. **The Swap/Reset:** ব্যাকট্র্যাকিংয়ের সবচেয়ে গুরুত্বপূর্ণ অংশ হলো স্টেট রিসেট করা (যেমন: `list.remove(list.size() - 1)`)। যদি এটি ভুলে যান, তবে আপনার রেজাল্ট ভুল আসবে।
- ৩. **Pruning:** যদি আপনি আগেই বুঝতে পারেন যে এই পথটি ভুল, তবে রিকার্সন কল করবেন না। এটি আপনার কোডের পারফরম্যান্স বহুগুণ বাড়িয়ে দেবে।

## ৭. উপসংহার (Conclusion)
ব্যাকট্র্যাকিং হলো কোডিং দুনিয়ার 'Exploration' টুল। এটি আপনাকে যেকোনো জটিল সমস্যার সব সম্ভাবনা খুঁজে বের করতে সাহায্য করে। যদিও এর টাইম কমপ্লেক্সিটি সাধারণত এক্সপোনেনশিয়াল ($2^N$ বা $N!$) হয়, কিন্তু সঠিক প্রুনিং (Pruning) এর মাধ্যমে একে অনেক অপ্টিমাইজ করা সম্ভব। পরবর্তী ফাইলে আমরা এর বিভিন্ন টেমপ্লেট এবং লেটকোড উদাহরণ দেখব।
