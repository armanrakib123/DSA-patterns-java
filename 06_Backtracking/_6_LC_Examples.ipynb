## ১. Subsets (LC 78) 🟢
**টাইপ:** Power Set Generation

### প্রবলেম স্টেটমেন্ট:
একটি অ্যারের সব সম্ভাব্য সাব-সেট (Subsets) জেনারেট করুন।

### সমাধান:
১. প্রতিটি ইলিমেন্টের জন্য আমাদের দুটি চয়েস: হয় একে নেব, না হয় নেব না।
২. ব্যাকট্র্যাকিং ব্যবহার করে প্রতিটি ইনডেক্সে এই ডিসিশন নিন।
৩. বেস কেস: যখন ইনডেক্স অ্যারের সাইজের সমান হবে।

## ২. Permutations (LC 46) 🟡
**টাইপ:** Sequence Generation

### প্রবলেম স্টেটমেন্ট:
একটি অ্যারের সব সম্ভাব্য বিন্যাস (Permutations) খুঁজে বের করুন।

### সমাধান:
১. একটি `used` অ্যারে মেইনটেইন করুন যা ট্র্যাক করবে কোন ইলিমেন্ট নেওয়া হয়েছে।
২. একটি লুপের মাধ্যমে প্রতিটি পজিশনে ইউনিক ইলিমেন্টগুলো ট্রাই করুন।
৩. ব্যাকট্র্যাক করার সময় `used[i] = false` করতে ভুলবেন না।

## ৩. Combination Sum (LC 39) 🟡
**টাইপ:** Target Sum Search

### প্রবলেম স্টেটমেন্ট:
অ্যারের ইলিমেন্টগুলো ব্যবহার করে এমন সব কম্বিনেশন বের করুন যাদের যোগফল টার্গেটের সমান। একই ইলিমেন্ট বারবার ব্যবহার করা যাবে।

### সমাধান:
১. প্রতি ধাপে একটি সংখ্যা যোগ করুন।
২. যদি যোগফল টার্গেট ছাড়িয়ে যায়, তবে ব্যাকট্র্যাক করুন।
৩. একই ইলিমেন্ট আবার ব্যবহার করার জন্য রিকার্সিভ কলে ইনডেক্স ইনক্রিমেন্ট করবেন না।

## ৪. N-Queens (LC 51) 🔴
**টাইপ:** Constraint Satisfaction

### প্রবলেম স্টেটমেন্ট:
একটি $N \times N$ বোর্ডে $N$ টি কুইন এমনভাবে বসান যাতে কেউ কাউকে অ্যাটাক করতে না পারে।

### সমাধান:
১. প্রতি রো-তে একটি করে কুইন বসানোর চেষ্টা করুন।
২. চেক করুন কলাম, ডায়াগোনাল এবং অ্যান্টি-ডায়াগোনালে কোনো কুইন আছে কি না।
৩. যদি নিরাপদ হয়, তবে কুইন বসান এবং পরবর্তী রো-তে যান।

## ৫. Word Search (LC 79) 🟡
**টাইপ:** Grid Exploration

### প্রবলেম স্টেটমেন্ট:
একটি ২ডি গ্রিডে কি কোনো নির্দিষ্ট শব্দ (Word) খুঁজে পাওয়া সম্ভব?

### সমাধান:
১. গ্রিডের প্রতিটি সেল থেকে শুরু করুন।
২. চার দিকে (উপরে, নিচে, ডানে, বামে) রিকার্সিভলি সার্চ করুন।
৩. ভিজিটেড সেলগুলো মার্ক করুন এবং ব্যাকট্র্যাক করার সময় আনমার্ক করুন।
